## Syntax and Semantics (Simple Telugu-English Mix)

**Syntax:**
Programming language lo rules and format ni **syntax** antaru.
Simple ga: code ela rayali ane grammar.

**Semantics:**
Code meaning and behavior ni **semantics** antaru.
Simple ga: code correct ga unna, adi actually emi chesthundi ane meaning.

**Memory trick:**
- Syntax = grammar
- Semantics = meaning

## Basic Rules (Easy)

### Syntax rules
1. Python lo indentation correct ga undali (usually 4 spaces).
2. `if`, `for`, `while`, `def` taruvata `:` compulsory.
3. Brackets and quotes properly close cheyyali.
4. Variable names letter/underscore tho start avvali.
5. Keywords ni variable names laga use cheyyakudadhu (example: `if`, `for`, `class`).

### Semantics rules
1. Code compile/run ayina saripodu, output correct undali.
2. Data type correct ga use cheyyali (int, str, list etc.).
3. Condition logic right ga undali (`>` vs `<`, `and` vs `or`).
4. Function purpose clear ga undali (name + behavior match avvali).
5. Real problem ki correct meaning/result isthene code useful.

### Quick check before running
- Format correct aa? (Syntax)
- Meaning correct aa? (Semantics)

```python
# Syntax + Semantics runnable examples

print("=== 1) Syntax (Correct Example) ===")
x = 10
if x > 5:
    print("Correct syntax block run ayyindi")

print("\n=== 2) Syntax Error Demo (Safe) ===")
bad_code = "if x > 5\nprint(x)"
try:
    compile(bad_code, "<string>", "exec")
except SyntaxError as e:
    print("Caught SyntaxError:", e)

print("\n=== 3) Semantics Demo ===")
a = 10 / 2
b = 5
print("a =", a, ", b =", b)
print("Values equal:", a == b)
print("Semantics different: one is division result, one is direct assignment")
```

## Case Sensitivity (Important)

Python **case-sensitive** language.
Ante uppercase and lowercase letters different ga treat chesthundi.

### Rules to remember
1. Variable names lo case same ga use cheyyali.
2. Function call chesetappudu exact case follow avvali.
3. Keywords lowercase lone untayi (`if`, `for`, `while`, `def`).
4. File/module names import time lo case mismatch unte error ravachu.

### Quick tip
- Naming style consistent ga petti confusion avoid cheyyi.
- Common style: variables/functions `snake_case`, classes `PascalCase`.

```python
# Case sensitivity runnable example

name = "Krish"
Name = "Sai"
print("name:", name)
print("Name:", Name)
print("Are they same variable?", name is Name)
print("Variables available:", "name" in locals(), "Name" in locals())
```

## Indentation (Very Important in Python)

Python lo indentation just style kaadu, **syntax part**.
Ante spaces correct ga ivvakapothe code error istundi.

### Basic rule
- Usually **4 spaces** use cheyyali per block.
- Tabs and spaces mix cheyyakudadhu.

### Where indentation is required
`if`, `elif`, `else`, `for`, `while`, `def`, `class`, `try`, `except`, `with` blocks lo.

### Quick tips
1. VS Code lo auto-indent on unchandi.
2. One file lo same indentation style maintain cheyyandi.
3. Paste chesina code ni reformat cheyyandi.
4. Error vaste first indentation check cheyyandi.

```python
# Indentation runnable examples

print("=== Correct indentation example ===")
age = 20
if age >= 18:
    print("Eligible to vote")
    print("You are an adult")
print("Program end")

print("\n=== Indentation error demo (safe) ===")
bad_indent_code = "if True:\nprint('Hi')"
try:
    exec(bad_indent_code)
except IndentationError as e:
    print("Caught IndentationError:", e)

print("\n=== Nested indentation example ===")
for i in range(3):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
```

## Line Continuation

Python lo line chala pedda ga unte next line ki continue cheyyadanini **line continuation** antaru.

### 1) Explicit line continuation (`\\`)
Backslash (`\\`) use chesi line ni next line ki continue cheyyachu.

### 2) Implicit line continuation
Brackets `()`, square brackets `[]`, braces `{}` lopala unte backslash avasaram ledu.

### Important rules
1. `\\` vesi taruvata space/comment pettakudadhu (error ravachu).
2. Possible aina appudu `()`, `[]`, `{}` style use cheyyadam better readability.
3. Long expressions ni clean ga multiple lines lo split cheyyi.

### Quick tip
- Preferred: implicit continuation with parentheses.
- Avoid too many backslashes where possible.

```python
# Line continuation runnable examples

print("=== 1) Explicit continuation with \\\\ ===")
total = 10 + 20 + 30 + \
        40 + 50
print("total:", total)

print("\n=== 2) Implicit continuation with parentheses ===")
message = (
    "Python lo line continuation use chesthe "
    "long lines readable ga untayi."
)
print(message)

print("\n=== 3) List continuation without backslash ===")
nums = [
    1,
    2,
    3,
    4,
]
print("nums:", nums)

print("\n=== 4) Dictionary continuation without backslash ===")
student = {
    "name": "Krish",
    "course": "Python Basics",
    "level": "Beginner",
}
print("student:", student)

print("\nAll line continuation examples ran successfully.")
```

## Multiple Statements on a Single Line

Python lo one line lo multiple statements rayachu using `;` (semicolon).

Example:
- `a = 10; b = 20; print(a + b)`

### Important
1. Technically valid, kani readability thakkuva avvachu.
2. Learning and production code lo usually separate lines better.
3. Small quick tests ki single-line statements use cheyyachu.

### Best practice
- Prefer one statement per line for clean and maintainable code.

```python
# Multiple statements on a single line - runnable examples

print("=== 1) Basic semicolon example ===")
a = 10; b = 20; print("Sum:", a + b)

print("\n=== 2) Multiple assignments in one line ===")
x = 5; y = 6; z = x * y
print("x:", x, "y:", y, "z:", z)

print("\n=== 3) if statement on one line ===")
num = 7
if num > 5: print("num is greater than 5")

print("\n=== 4) Preferred readable style ===")
a = 10
b = 20
print("Sum:", a + b)

print("\nAll single-line statement examples ran successfully.")
```

## Type Inference

Python lo variable ki data type separate ga declare cheyyalsina avasaram ledu.
Python value batti type ni automatic ga guess chesthundi. Dinine **type inference** antaru.

Example:
- `x = 10` -> `x` is `int`
- `name = "Krish"` -> `name` is `str`

### Important points
1. Python dynamically typed language.
2. Same variable ki later different type assign cheyyachu.
3. `type()` function use chesi current type check cheyyachu.

### Best practice
- Type inference flexible ayina, variable names clear ga petti readability maintain cheyyi.

```python
# Type inference runnable examples

print("=== 0) Your exact example ===")
variable = 10
print(type(variable))

variable = "Krish"
print(type(variable))

print("\n=== 1) Basic type inference ===")
x = 10
name = "Krish"
price = 99.5
is_active = True

print("x:", x, "->", type(x).__name__)
print("name:", name, "->", type(name).__name__)
print("price:", price, "->", type(price).__name__)
print("is_active:", is_active, "->", type(is_active).__name__)

print("\n=== 2) Same variable, different type ===")
value = 100
print("value:", value, "->", type(value).__name__)

value = "Hundred"
print("value:", value, "->", type(value).__name__)

print("\n=== 3) Inference from expression ===")
a = 5
b = 2
result = a / b
print("result:", result, "->", type(result).__name__)

print("\nAll type inference examples ran successfully.")
```
