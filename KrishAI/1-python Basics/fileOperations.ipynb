# File Operations in Python — Zero to Pro Guide

## Architecture Diagram

```
[User Code]
    ↓
[open() function — File Handle banddutundi]
    ↓
[Read/Write/Append — File lo data transfer]
    ↓
[close() — File handle close chestundi]
    ↓
[Data in Memory | Data on Disk]
```

## Deep Architecture Notes

**File Operations Flow:**
1. open() — File path ni accept chesi file handle return chestundi
2. Mode parameter — "r" (read), "w" (write), "a" (append), "b" (binary) decide chestundi
3. Read/Write/Append operations — Actual data transfer happen chestundi
4. close() — File handle release chesi resources free chestundi
5. Context managers (with statement) — Automatic close guarantee chestundi

**Key Concepts:**
- File pointer — Current position file lo track chestundi
- Mode types — Different operations (read-only, write-only, read-write)
- Buffering — Data temporary memory lo store avutundi, then disk lo transfer
- Encoding — Text files lo character representation specify chestundi

## Section 5: File Modes in Detail

### Different Modes Explained

| Mode | Name | Description |
|------|------|-------------|
| "r" | Read | File read chestundi (default), file exist avvali |
| "w" | Write | File create/overwrite chestundi |
| "a" | Append | File end lo data add chestundi |
| "x" | Exclusive | New file create chestundi (existing file create cheyyadu) |
| "r+" | Read+Write | File read ani write both chestundi |
| "w+" | Write+Read | File create/overwrite, read+write chestundi |
| "b" | Binary | Binary mode (images, videos, etc.) |
| "t" | Text | Text mode (default) |


````python
# Mode "x" lo file open chestundi — Exclusive create mode
# File already exist chese, error occur chestundi

try:
    with open("new_exclusive.txt", "x") as file_obj:
        file_obj.write("This is a new file created with x mode")
        print("File created successfully!")
except FileExistsError:
    print("File already exists! Cannot create with x mode")
````

````python
# Create sample.txt first
with open("sample.txt", "w") as f:
    f.write("Original content here")

# Mode "r+" lo file open chestundi — Read and Write both
# File already exist avvali

with open("sample.txt", "r+") as file_obj:
    # Read chestundi
    content = file_obj.read()
    print("Content:", content)
    
    # File pointer ki end lo move chestundi
    file_obj.seek(0, 2)  # 0 = position, 2 = end of file
    
    # Write chestundi (existing content preserve avuthundi)
    file_obj.write("\nNew content appended!")
````

## Section 7: Error Handling in File Operations

### Try-Except Blocks use chesi errors handle chestundi


````python
# Try-except lo file operations chestundi

try:
    # File open chestundi
    with open("nonexistent.txt", "r") as file_obj:
        content = file_obj.read()
        print(content)
except FileNotFoundError:
    # File exist kaledu — Error handle chestundi
    print("Error: File not found!")
except IOError:
    # Input/Output error — handle chestundi
    print("Error: Cannot read the file!")
else:
    # Try block success chese, else execute chestundi
    print("File read successfully!")
finally:
    # Always execute chestundi — cleanup work chestundi
    print("File operation completed.")
````

## Section 8: Working with Binary Files

### "b" mode use chesi binary files (images, videos) handle chestundi


````python
# Binary mode lo write chestundi

# Text string binary form lo convert chestundi
text = "Hello Binary World!"
binary_data = text.encode("utf-8")  # String -> bytes

# "wb" mode - Write Binary
with open("binary_output.bin", "wb") as file_obj:
    file_obj.write(binary_data)
    print("Binary data written!")

# Binary data back read chestundi
with open("binary_output.bin", "rb") as file_obj:
    read_data = file_obj.read()
    decoded = read_data.decode("utf-8")
    print(f"Read from binary: {decoded}")
````

````python
# Binary mode lo write chestundi

# Text string binary form lo convert chestundi
text = "Hello Binary World!"
binary_data = text.encode("utf-8")  # String → bytes

# "wb" mode — Write Binary
with open("binary_output.bin", "wb") as file_obj:
    file_obj.write(binary_data)
    print("Binary data written!")

# Binary data back read chestundi
with open("binary_output.bin", "rb") as file_obj:
    read_data = file_obj.read()
    decoded_content = read_data.decode("utf-8")
    print(f"Read from binary: {decoded_content}")
````

## Section 9: File Pointer & Seek Operations

### seek() method — File lo specific position lo move chestundi


````python
# Create sample.txt first
with open("sample.txt", "w") as f:
    f.write("0123456789ABCDEFGHIJ")

# File pointer position control chestundi

with open("sample.txt", "r") as file_obj:
    # File pointer beginning lo unnadi
    content1 = file_obj.read(10)  # First 10 characters read chestundi
    print(f"First read: {content1}")
    
    # tell() — Current position return chestundi
    current_pos = file_obj.tell()
    print(f"Current position: {current_pos}")
    
    # seek(0) — Beginning ki jump chestundi
    file_obj.seek(0)
    content2 = file_obj.read(10)  # Again first 10 characters
    print(f"Second read (after seek): {content2}")
    
    # seek(5) — 5th position lo jump chestundi
    file_obj.seek(5)
    content3 = file_obj.read(10)
    print(f"Third read (from position 5): {content3}")
````

## Section 12: Real-World Example — Copying a Text File

### Read content from source text file and write to destination text file

File copy cheyali antay, source file lo unna content read chesi, destination file lo write chestundi.


````python
# Source file create chestundi (original data with)
source_file = "source.txt"
destination_file = "destination.txt"

# Source file lo data write chestundi
with open(source_file, "w") as f:
    f.write("Original file content — Line 1\n")
    f.write("Line 2: File Operations in Python\n")
    f.write("Line 3: This is copied from source to destination!\n")
    print(f"Source file {source_file} created!")

# Source file lo data read chestundi
with open(source_file, "r") as src_file:
    # Entire content read chestundi
    content = src_file.read()
    print(f"\nContent read from {source_file}:")
    print(content)

# Destination file lo content write chestundi
with open(destination_file, "w") as dest_file:
    # Source file lo read chestundi content likhao
    dest_file.write(content)
    print(f"\nContent written to {destination_file}!")

# Verify: Destination file lo data check chestundi
with open(destination_file, "r") as verify_file:
    copied_content = verify_file.read()
    print(f"\nVerification - Content from {destination_file}:")
    print(copied_content)
    
# Both files same content unnai verify chestundi
if content == copied_content:
    print(f"\n[SUCCESS] File successfully copied! Both files identical.")
else:
    print(f"\n[ERROR] File copy failed!")
````

## Section 13: Real-World Example — File Analysis (Lines, Words, Characters)

### Read a text file and count the number of lines, words, and characters

File lo unna content analyze chesi, total lines, total words, total characters count chestundi.


````python
# Sample text file create chestundi
filename = "sample_document.txt"

# File lo multiple lines data write chestundi
content = """Python is a powerful programming language.
File operations are essential for data processing.
This example demonstrates how to count lines, words, and characters.
Let us learn and practice!
Have fun coding!"""

with open(filename, "w") as f:
    f.write(content)
    print(f"Sample file {filename} created!\n")

# File lo data read chestundi
with open(filename, "r") as f:
    # Entire content read chestundi
    file_content = f.read()
    print("File Content:")
    print(file_content)
    print("\n" + "="*50 + "\n")

# File statistics calculate chestundi
with open(filename, "r") as f:
    # readlines() method use chesi line-by-line read chestundi
    lines = f.readlines()

# Total lines count chestundi
total_lines = len(lines)
print(f"Total Lines: {total_lines}")

# Total characters count chestundi
total_characters = len(file_content)
print(f"Total Characters (including spaces & newlines): {total_characters}")

# Whitespace exclude chesi character count chestundi
characters_without_space = len(file_content.replace(" ", "").replace("\n", ""))
print(f"Total Characters (excluding spaces & newlines): {characters_without_space}")

# Total words count chestundi
# split() method use chesi words separate chestundi
words = file_content.split()
total_words = len(words)
print(f"Total Words: {total_words}")

# Average calculations
avg_words_per_line = total_words / total_lines if total_lines > 0 else 0
avg_chars_per_line = total_characters / total_lines if total_lines > 0 else 0
avg_chars_per_word = total_characters / total_words if total_words > 0 else 0

print(f"\nAverage words per line: {avg_words_per_line:.2f}")
print(f"Average characters per line: {avg_chars_per_line:.2f}")
print(f"Average characters per word: {avg_chars_per_word:.2f}")
````

### Advanced Analysis — Most Common Words

File lo unna words ko count chesi, kondu most common unnai identify chestundi.


````python
# Sample text file create chestundi
filename = "text_analysis.txt"

sample_text = """Python is great for file operations.
Python is widely used in data science.
File handling with Python is easy and powerful.
Learn Python, master programming!"""

with open(filename, "w") as f:
    f.write(sample_text)
    print(f"File {filename} created!\n")

# File lo content read chestundi
with open(filename, "r") as f:
    content = f.read().lower()  # lowercase convert chestundi (case-insensitive)

# Words separate chestundi
words = content.split()

# Dictionary use chesi word frequency track chestundi
word_frequency = {}

for word in words:
    # Punctuation remove chestundi (if any)
    word_clean = word.strip(".,!?;:")
    
    if word_clean:
        # Dictionary lo word count increment chestundi
        if word_clean in word_frequency:
            word_frequency[word_clean] += 1
        else:
            word_frequency[word_clean] = 1

# Sort chesi most common words find chestundi
sorted_words = sorted(word_frequency.items(), key=lambda x: x[1], reverse=True)

print("Word Frequency (Top 5 most common):")
print("-" * 40)
for word, count in sorted_words[:5]:
    print(f"  {word:15} : {count} times")

print(f"\nTotal unique words: {len(word_frequency)}")
print(f"Total words (including duplicates): {len(words)}")
````

## Section 14: Deep Dive — The w+ Mode (Write & Read)

### Understanding w+ mode in Python

The w+ mode in Python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exists, its content is truncated (i.e., the file is overwritten).

**w+ mode characteristics:**
- File open chestundi both read and write operations ke liye
- File exist kaledu, new file create chestundi
- File exist chese, existing content truncate (overwrite) chestundi
- File pointer beginning lo start avuthundi
- Read and write both operations possible


### w+ vs other modes — Comparison Table

| Mode | Create | Overwrite | Read | Write | Seek |
|------|--------|-----------|------|-------|-------|
| r | No | No | Yes | No | Yes |
| w | Yes | Yes | No | Yes | Yes |
| a | Yes | No | No | Yes | No |
| r+ | No | No | Yes | Yes | Yes |
| w+ | Yes | Yes | Yes | Yes | Yes |
| a+ | Yes | No | Yes | Yes | Yes |

**w+ mode idi best choice antay:**
- New file create chesi, immediately read cheyali
- File lo data write chesi, verification read cheyali
- Both operations same file lo cheyali


````python
# w+ mode lo file open chestundi
filename = "w_plus_demo.txt"

# Open file in w+ mode
# Agar file exist chese, content truncate avuthundi (overwrite)
# File exist kaledu, new file create chestundi
with open(filename, "w+") as f:
    # First, write some data
    f.write("Line 1: Introduction to w+ mode\n")
    f.write("Line 2: File opened for both reading and writing\n")
    f.write("Line 3: Content can be modified dynamically\n")
    print(f"Data written to {filename}")
    
    # Now read the data back
    # seek(0) use chesi file pointer beginning lo move chestundi
    f.seek(0)
    
    # Read all content
    content = f.read()
    print(f"\nContent read from {filename}:")
    print(content)
````

### Advanced Example — Writing and then Reading a File

Write chesi, verify chese, modify chesi, again read chestundi.


````python
# w+ mode lo advanced example
filename = "data_processing.txt"

# Open file in w+ mode
with open(filename, "w+") as f:
    # Step 1: Write initial data
    print("Step 1: Writing initial data...")
    initial_data = """AI Learning Progress:
Module 1: Python Basics - COMPLETED
Module 2: File Operations - IN PROGRESS
Module 3: Data Structures - PENDING"""
    
    f.write(initial_data)
    print(f"Initial data written!\n")
    
    # Step 2: Read and verify
    print("Step 2: Reading and verifying data...")
    f.seek(0)  # File pointer beginning lo move chestundi
    content = f.read()
    print(f"Current file content:\n{content}\n")
    
    # Step 3: Append more data
    print("Step 3: Appending additional data...")
    f.seek(0, 2)  # File pointer end lo move chestundi (0 = offset, 2 = end)
    f.write("\n\nUpdated at: 2025-01-01")
    print("Additional data appended!\n")
    
    # Step 4: Read complete updated content
    print("Step 4: Reading complete updated content...")
    f.seek(0)  # Beginning lo move chestundi
    updated_content = f.read()
    print(f"Complete file content:\n{updated_content}")
    
    # Step 5: Get file statistics
    print(f"\nFile Statistics:")
    print(f"  - Total characters: {len(updated_content)}")
    print(f"  - Total lines: {updated_content.count(chr(10)) + 1}")
    print(f"  - Total words: {len(updated_content.split())}")
````

### Real-World Use Case — Configuration File Management

w+ mode use chesi configuration file update chesi, verify chestundi.


````python
# Configuration file create chesi manage chestundi
config_file = "app_config.txt"

# Initial configuration
initial_config = """[Application Settings]
app_name=AILearnings
version=1.0
debug=false
database=localhost
port=5432"""

with open(config_file, "w+") as f:
    # Write initial configuration
    f.write(initial_config)
    print(f"Initial configuration written to {config_file}\n")
    
    # Read and display initial config
    f.seek(0)
    print("Initial Configuration:")
    initial_content = f.read()
    print(initial_content)
    print("\n" + "="*50 + "\n")
    
    # Update configuration
    print("Updating configuration...")
    f.seek(0)
    f.truncate()  # Truncate file lo current position ke baad content delete chestundi
    
    updated_config = """[Application Settings]
app_name=AILearnings
version=1.1
debug=true
database=production_db
port=8080
last_update=2025-01-01"""
    
    f.write(updated_config)
    print("Configuration updated!\n")
    
    # Read and verify updated config
    f.seek(0)
    print("Updated Configuration:")
    updated_content = f.read()
    print(updated_content)
    
    # Parse and display specific settings
    print(f"\nParsed Settings:")
    for line in updated_content.split("\n")[1:]:  # Skip header
        if "=" in line:
            key, value = line.split("=")
            print(f"  {key.strip()}: {value.strip()}")
````

## Section 15: Directory Operations — Creating, Listing, Managing Directories

### Create directories, list files, and manage folder structures

Directory lo files create chesi, existing directories list chesi, folder structure manage chestundi. Python lo `os` module use chesi directory operations chestundi.

**Key os module methods:**
- `os.mkdir()` — Single directory create chestundi
- `os.makedirs()` — Nested directories create chestundi (with intermediate)
- `os.listdir()` — Directory lo files/folders list chestundi
- `os.getcwd()` — Current working directory path return chestundi
- `os.chdir()` — Directory change chestundi
- `os.path.exists()` — File/directory exist chestundi check chestundi

- `os.path.isdir()` — Path idi directory aa? check chestundi


````python
# os module import chestundi
import os

# Current working directory get chestundi
current_directory = os.getcwd()
print(f"Current Working Directory: {current_directory}")

# New directory create chestundi
new_directory = "my_project"

# Directory already exist chese check chestundi
if not os.path.exists(new_directory):
    os.mkdir(new_directory)  # Single directory create chestundi
    print(f"Directory {new_directory} created successfully!")
else:
    print(f"Directory {new_directory} already exists!")

# Nested directories create chestundi (makedirs)
# os.mkdir() single level only create chestundi
# os.makedirs() multiple levels create chestundi
nested_path = "data/raw/input"

if not os.path.exists(nested_path):
    os.makedirs(nested_path)  # Intermediate directories automatically create chestundi
    print(f"Nested directory structure created: {nested_path}")
else:
    print(f"Path {nested_path} already exists!")

print("\n[INFO] Directory creation completed!")
````

### Listing Files and Directories

### Advanced Directory Listing — Recursive with os.walk()

Entire directory structure recursively list chestundi (subdirectories lo files kuda).


````python
# os module import chestundi
import os

# Directory structure setup chestundi
root_dir = "project_structure"
os.makedirs(os.path.join(root_dir, "src"), exist_ok=True)
os.makedirs(os.path.join(root_dir, "data"), exist_ok=True)
os.makedirs(os.path.join(root_dir, "tests"), exist_ok=True)

# Sample files create chestundi
files_to_create = [
    os.path.join(root_dir, "src", "main.py"),
    os.path.join(root_dir, "src", "utils.py"),
    os.path.join(root_dir, "data", "input.csv"),
    os.path.join(root_dir, "tests", "test_main.py")
]

for file_path in files_to_create:
    with open(file_path, "w") as f:
        f.write(f"Sample content")

print(f"Project structure created!\n")
print(f"Listing entire {root_dir} structure recursively:\n")

# os.walk() use chesi recursive listing chestundi
# os.walk() returns tuple: (dirpath, dirnames, filenames)

for dirpath, dirnames, filenames in os.walk(root_dir):
    # Current directory level get chestundi
    level = dirpath.replace(root_dir, "").count(os.sep)
    indent = " " * 2 * level  # Indentation for hierarchy
    print(f"{indent}[FOLDER] {os.path.basename(dirpath)}/")
    
    # Current directory lo files list chestundi
    subindent = " " * 2 * (level + 1)
    for filename in filenames:
        file_full_path = os.path.join(dirpath, filename)
        file_size = os.path.getsize(file_full_path)
        print(f"{subindent}[FILE] {filename} ({file_size} bytes)")
````

### File Filtering — List specific file types

Directory lo unna files lo, specific type (extension) ke files only list chestundi.


````python
# os module import chestundi
import os

# Sample files directory setup chestundi
files_dir = "sample_files"
os.makedirs(files_dir, exist_ok=True)

# Different extensions ke files create chestundi
sample_files = [
    ("document1.txt", "Text file 1"),
    ("document2.txt", "Text file 2"),
    ("script.py", "Python script"),
    ("data.csv", "CSV data"),
    ("readme.md", "Markdown file"),
    ("image.jpg", "Image file")
]

for filename, content in sample_files:
    file_path = os.path.join(files_dir, filename)
    with open(file_path, "w") as f:
        f.write(content)

print(f"Sample files created in {files_dir}!\n")

# Text files only list chestundi
print("Text files (.txt):")
txt_files = [f for f in os.listdir(files_dir) if f.endswith(".txt")]
for file in txt_files:
    file_path = os.path.join(files_dir, file)
    file_size = os.path.getsize(file_path)
    print(f"  - {file} ({file_size} bytes)")

# Python files only list chestundi
print("\nPython files (.py):")
py_files = [f for f in os.listdir(files_dir) if f.endswith(".py")]
for file in py_files:
    file_path = os.path.join(files_dir, file)
    file_size = os.path.getsize(file_path)
    print(f"  - {file} ({file_size} bytes)")

# All files with their types list chestundi
print("\nAll files by type:")
all_files = os.listdir(files_dir)
for file in sorted(all_files):
    file_path = os.path.join(files_dir, file)
    if os.path.isfile(file_path):
        _, extension = os.path.splitext(file)  # File extension extract chestundi
        file_size = os.path.getsize(file_path)
        print(f"  {file:20} {extension:8} ({file_size:5} bytes)")
````

### Directory Information & Statistics

Directory lo unna total files, directories, total size calculate chestundi.


````python
# os module import chestundi
import os

# Directory setup chestundi
target_dir = "stats_directory"
os.makedirs(os.path.join(target_dir, "subfolder"), exist_ok=True)

# Sample files create chestundi
for i in range(3):
    with open(os.path.join(target_dir, f"file{i}.txt"), "w") as f:
        f.write(f"Content of file {i} " * 10)

for i in range(2):
    with open(os.path.join(target_dir, "subfolder", f"subfile{i}.txt"), "w") as f:
        f.write(f"Subfolder content {i} " * 15)

# Directory statistics collect chestundi
total_files = 0
total_dirs = 0
total_size = 0

for dirpath, dirnames, filenames in os.walk(target_dir):
    total_dirs += len(dirnames)
    total_files += len(filenames)
    
    for filename in filenames:
        file_path = os.path.join(dirpath, filename)
        total_size += os.path.getsize(file_path)

# Statistics display chestundi
print(f"Directory Statistics for: {target_dir}")
print("=" * 50)
print(f"Total Directories: {total_dirs}")
print(f"Total Files: {total_files}")
print(f"Total Size: {total_size} bytes ({total_size / 1024:.2f} KB)")
print(f"Average file size: {total_size / total_files:.2f} bytes" if total_files > 0 else "No files")
````

## Section 12: File Operations Checklist

### Important Points to Remember

✅ **Always do:**
- File open chese lo mode correct ga specify cheyyali ("r", "w", "a")
- Context managers ("with" statement) use cheyyali — automatic close guarantee
- Large files lo readlines() or line-by-line reading use cheyyali (memory save)
- Error handling (try-except) implement cheyyali
- Binary files lo "b" mode use cheyyali (images, videos, etc.)

❌ **Never do:**
- File close() forget cheyakudadhu — resource leak avuthundi
- Entire large file read() method lo load cheyakudadhu — memory crash
- File path hard-code cheyakudadhu — os.path use cheyyali
- Error handling skip cheyakudadhu — program crash avuthundi

```python
# Best practice — Always use "with" statement
with open("file.txt", "r") as f:
    data = f.read()  # or f.readlines() for large files
```
