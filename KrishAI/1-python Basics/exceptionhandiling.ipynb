# Architecture Diagram

```text
[Code Execution]
      |
[Exception Occurs?]
      ├─ NO → Continue
      └─ YES ↓
         [Try Block Stops]
              |
         [Match Exception Handler?]
              ├─ YES → Execute Except Block ↓
              └─ NO → Propagate Up ↓
                     [Finally Always Runs]
                          |
                     [Cleanup/Logging]
```

# Deep Architecture Notes

**exception handling** ni understand cheyali. Python lo error ochinappudu normal execution stop aiyindi — **exception** antunnadi.

**Try-Except Flow:**
1. Try block lo code execute chestundi
2. Exception occur ainappudu, try block immediately stop aiyindi
3. Except block run chestundi (matching type if specified)
4. Finally block always run chestundi (cleanup ke)

**Exception Types:**
- FileNotFoundError — File ni find chey leni appudu
- ValueError — Invalid value pass ainappudu
- TypeError — Wrong type unnappudu
- ZeroDivisionError - 0 by divide chey leni appudu

## Why Exception Handling?

Program lo runtime errors vastayi: user wrong input ivvachu, file missing undachu, network fail avvachu. Exception handling valla application sudden ga crash avvakunda meaningful message ichchi, safe recovery path choose chestundi.

> **Important:** Exception ni catch cheyadam ante error ni ignore cheyadam kaadu. Expected error ni handle cheyandi; unexpected error ni context tho re-raise cheyandi.


## 1. Basic `try` and `except`

`try` block lo error raavachu ani expect chese code pedatham. Error vaste Python matching `except` block ki control ni transfer chestundi. Error lekapothe `except` skip avuthundi.

Syntax:
```python
try:
    risky_operation()
except SpecificError as error:
    recover_from_error(error)
```


````python
# User enter chesina text ni integer ga convert cheyadaniki value ni store chestundi.
user_age_text = "twenty"

# Conversion risky kabatti try block ni start chestundi.
try:
    # Text ni integer ga convert cheyadaniki try chestundi.
    user_age = int(user_age_text)
    # Conversion successful aina age ni print chestundi.
    print(f"Your age is {user_age}")
# Invalid numeric text valla vachina ValueError ni specific ga handle chestundi.
except ValueError as conversion_error:
    # User ki understandable message ni error details tho print chestundi.
    print(f"Please enter a whole number. Details: {conversion_error}")
````

## 2. Handle Multiple Exception Types

Different failures ki different recovery logic undali. Specific handlers ni **most specific nundi general** order lo rayandi. `ValueError` and `ZeroDivisionError` rendu expected failures kabatti clear messages ivvachu.


````python
# Division kosam numerator value ni define chestundi.
numerator = 100
# User input laga denominator text ni define chestundi.
denominator_text = "0"

# Conversion and division operations ni safe ga execute cheyadaniki try start chestundi.
try:
    # Denominator text ni integer ga convert chestundi.
    denominator = int(denominator_text)
    # Numerator ni denominator tho divide chestundi.
    result = numerator / denominator
# Non-numeric denominator vachinappudu handle chestundi.
except ValueError:
    # Valid integer enter cheyamani user ki cheptundi.
    print("Denominator must be a valid integer.")
# Zero denominator vachinappudu handle chestundi.
except ZeroDivisionError:
    # Zero tho divide cheyalem ani user ki cheptundi.
    print("Denominator cannot be zero.")
# Error lekunda division complete ainappudu matrame run avuthundi.
else:
    # Successful calculation result ni print chestundi.
    print(f"Division result: {result}")
````

## 3. `else` and `finally`

- `else`: `try` lo exception lekapothe matrame run avuthundi. Success-only logic ki correct place.
- `finally`: exception vachina raakapoina run avuthundi. Resource cleanup (file close, DB connection close) ki use cheyandi.

`finally` lo `return` avoid cheyandi; adi original exception ni accidentally suppress cheyachu.


````python
# Demo kosam valid numeric text ni define chestundi.
score_text = "92"

# Input conversion ni attempt cheyadaniki try start chestundi.
try:
    # Score text ni integer ga convert chestundi.
    score = int(score_text)
# Invalid score input ni handle chestundi.
except ValueError as input_error:
    # Invalid input reason ni print chestundi.
    print(f"Invalid score: {input_error}")
# Conversion successful ayinappudu matrame run avuthundi.
else:
    # Valid score ni print chestundi.
    print(f"Saved score: {score}")
# Success leda failure tarvata cleanup notification ni always run chestundi.
finally:
    # Request complete ayindi ani print chestundi.
    print("Input processing finished.")
````

## 4. Validate Data with `raise`

Business rule break ainappudu Python ki explicit ga exception raise cheyachu. Example: age negative ga undakudadhu. Built-in `ValueError` use cheyandi when supplied value valid range/format lo ledu.


````python
# Age validation ni reusable function ga define chestundi.
def validate_age(age):
    # Negative age invalid ani check chestundi.
    if age < 0:
        # Caller ki clear validation exception ni raise chestundi.
        raise ValueError("Age cannot be negative.")
    # Valid age ni caller ki return chestundi.
    return age

# Invalid sample age ni validate cheyadaniki try start chestundi.
try:
    # Negative age tho validation function ni call chestundi.
    validated_age = validate_age(-5)
# Raised validation error ni handle chestundi.
except ValueError as validation_error:
    # Validation failed ani reason tho print chestundi.
    print(f"Validation failed: {validation_error}")
````

## 5. Custom Exceptions

Domain-specific error meaning important ayinappudu custom exception create cheyandi. `Exception` ni inherit chesina class saripothundi. Example: bank withdrawal amount balance kante ekkuva unte `InsufficientFundsError` use cheyachu.


````python
# Account balance sariponappudu use cheyadaniki custom exception class ni define chestundi.
class InsufficientFundsError(Exception):
    # Default Exception behavior ni use cheyadaniki class body ni pass chestundi.
    pass

# Account nundi amount withdraw cheyadaniki function ni define chestundi.
def withdraw(balance, amount):
    # Requested amount available balance kante ekkuva unda ani check chestundi.
    if amount > balance:
        # Business-specific error ni clear message tho raise chestundi.
        raise InsufficientFundsError("Withdrawal amount exceeds the available balance.")
    # Successful withdrawal tarvata remaining balance ni return chestundi.
    return balance - amount

# Withdrawal operation ni safely test cheyadaniki try start chestundi.
try:
    # Insufficient balance unna withdrawal ni attempt chestundi.
    remaining_balance = withdraw(500, 750)
# Domain-specific exception ni exact ga handle chestundi.
except InsufficientFundsError as funds_error:
    # Failed transaction reason ni print chestundi.
    print(f"Transaction declined: {funds_error}")
````

## 6. File Handling Best Practice: `with`

File open chesetappudu `with open(...)` use cheyandi. File read fail aina leda success aina context manager file ni automatically close chestundi; manual `close()` kosam `finally` rayalsina avasaram ledu.

Below example missing file ni correctly handle chestundi. Existing file path replace chesi success path kuda try cheyandi.


````python
# Demonstration kosam missing file path ni define chestundi.
file_path = "monthly_report.txt"

# File read operation ni safely attempt cheyadaniki try start chestundi.
try:
    # Context manager tho file ni open chesi automatic cleanup ni ensure chestundi.
    with open(file_path, "r", encoding="utf-8") as report_file:
        # File content motham ni read chestundi.
        report_content = report_file.read()
# File lekapothe expected error ni handle chestundi.
except FileNotFoundError:
    # File path verify cheyamani user ki cheptundi.
    print(f"File not found: {file_path}")
# File successful ga read ayinappudu matrame run avuthundi.
else:
    # Read ayina character count ni print chestundi.
    print(f"Read {len(report_content)} characters.")
````

## 7. Preserve Root Cause with Exception Chaining

Low-level error ni application-level message ga convert cheyalsi vaste `raise NewError(...) from original_error` use cheyandi. Idi traceback lo original cause ni preserve chestundi, debugging easy avuthundi.


````python
# Configuration error meaning ni represent cheyadaniki custom exception ni define chestundi.
class ConfigurationError(Exception):
    # Default Exception behavior ni use cheyadaniki class body ni pass chestundi.
    pass

# Port text ni validate chesi integer ga return cheyadaniki function ni define chestundi.
def parse_port(port_text):
    # String conversion ni safely attempt cheyadaniki try start chestundi.
    try:
        # Port text ni integer ga convert chestundi.
        return int(port_text)
    # Conversion failure ni application context tho translate chestundi.
    except ValueError as conversion_error:
        # Original ValueError ni cause ga preserve chesi new error ni raise chestundi.
        raise ConfigurationError("PORT must contain a whole number.") from conversion_error

# Invalid configuration ni demonstrate cheyadaniki outer try start chestundi.
try:
    # Invalid port text tho parser ni call chestundi.
    server_port = parse_port("eighty")
# Clear application-level exception ni handle chestundi.
except ConfigurationError as configuration_error:
    # Configuration correction message ni print chestundi.
    print(f"Configuration problem: {configuration_error}")
````

## Best Practices Checklist

1. **Catch only expected errors:** `except ValueError` is better than bare `except:`.
2. **Keep `try` small:** only risky statements ni try lo petti, accidental bugs ni catch avvakunda chudandi.
3. **Use `else` for success logic:** try block readable ga untundi.
4. **Use `finally` or context managers for cleanup:** especially files, sockets, database connections.
5. **Do not use exceptions for normal control flow:** simple `if` conditions better when no error actually occurred.
6. **Give actionable messages:** user/employer ki next action telise laga error message rayandi.
7. **Re-raise when needed:** current layer recover cheyalekapothe exception ni swallow cheyakandi.

### Quick Summary
`try` risky code ni contain chestundi, `except` expected failure ni recover chestundi, `else` successful path ni run chestundi, and `finally` cleanup ni guarantee chestundi. Ee pattern ni input validation, file operations, API calls, database code, and production applications lo apply cheyandi.
