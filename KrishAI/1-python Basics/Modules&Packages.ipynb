# Python Modules & Packages — Complete Guide

## Architecture Diagram
```
[Your Script / Notebook]
        ↓  import
[Module (.py file)]  ←→  [Package (folder with __init__.py)]
        ↓
[Built-in Modules]   [Third-party (pip install)]   [Your own .py files]
        ↓
[Python Standard Library]  →  math, os, sys, datetime, random, json ...
        ↓
[PyPI — pip install]       →  numpy, pandas, requests, langchain ...
```

## Deep Architecture Notes
- **Module** = oka `.py` file — functions, classes, variables ni reuse cheyyadam kosam
- **Package** = oka folder with `__init__.py` — related modules ni group cheyyadam
- **import** = Python interpreter aa file ni load chesi execute chestundi — once only (cached)
- **`sys.path`** = Python module konkoye locations list — ikkade lekapothe `ModuleNotFoundError`
- **`__name__`** = module ni direct run chesthe `'__main__'`, import chesthe module name avutundi
- **pip** = Python Package Installer — PyPI nunchi third-party packages download + install


## 1) Module Ante Enti?

Module = oka `.py` file — functions, classes, variables unna reusable code file. Oka sari define chesthe, eppudu kavalante import chesi vāḍochu.

````python
# Module ante basically oka .py file — import chesthe aa file lo unna anni things accessible
# Python lo built-in modules chaala unnay — extra install avasaram ledu

# math module import — mathematical operations kosam
import math

print('pi value       :', math.pi)           # pi constant
print('sqrt(16)       :', math.sqrt(16))     # square root
print('ceil(4.2)      :', math.ceil(4.2))    # round up
print('floor(4.9)     :', math.floor(4.9))   # round down
print('factorial(6)   :', math.factorial(6)) # 6! = 720
print('log(100, 10)   :', math.log(100, 10)) # log base 10
print('pow(2, 10)     :', math.pow(2, 10))   # 2^10
print('sin(90 deg)    :', round(math.sin(math.radians(90)), 2))
````

## 2) import Cheyyే Different Ways

Python lo module import cheyyadaniki 4 different syntaxes — prathi oka use case lo use chestam.

````python
# Way 1: Full module import — module.function() ga access cheyyadam
import math
print('Way 1 — import math      :', math.sqrt(25))

# Way 2: Alias tho import — long name ki short name ివ్వడం
import math as m
print('Way 2 — import math as m :', m.sqrt(25))

# Way 3: Specific function matrame import — module.prefix avasaram ledu
from math import sqrt, pi
print('Way 3 — from math import :', sqrt(25), '| pi =', pi)

# Way 4: Alias tho specific import
from math import factorial as fact
print('Way 4 — alias            :', fact(5))

# BAD practice — from math import * — cheyyadam avoid cheyyandi
# Enduku? — anni names current namespace lo vasttay, conflicts possible
````

## 4) Custom Module — Mana Sonta Module Create Cheyyadam

Mana own `.py` file ni module ga use cheyyadam possible. Ikkade `mymath.py` ane module create chesi import chestam.

````python
# mymath.py file create cheyyadam — custom module
module_code = '''
# mymath.py — Custom math module
# Idi mana sonta module — import chesthe ee functions use cheyyadam

PI = 3.14159265358979   # pi constant

def area_circle(radius):       # circle area calculate chestundi
    return PI * radius ** 2

def area_rectangle(l, w):      # rectangle area
    return l * w

def area_triangle(b, h):       # triangle area
    return 0.5 * b * h

def is_prime(n):               # prime number check
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

def primes_upto(n):            # n varaku unna prime numbers list
    return [x for x in range(2, n+1) if is_prime(x)]
'''

# File write cheyyadam
with open('mymath.py', 'w', encoding='utf-8') as f:
    f.write(module_code)
print('mymath.py created')
````

````python
# Mana custom module import cheyyadam
import mymath

print('PI               :', mymath.PI)
print('Circle area r=5  :', round(mymath.area_circle(5), 4))
print('Rectangle 4x6    :', mymath.area_rectangle(4, 6))
print('Triangle b=3 h=4 :', mymath.area_triangle(3, 4))
print('is_prime(17)     :', mymath.is_prime(17))
print('is_prime(20)     :', mymath.is_prime(20))
print('Primes upto 30   :', mymath.primes_upto(30))

# dir() tho module lo unna attributes chudadam
public = [x for x in dir(mymath) if not x.startswith('_')]
print('Module contents  :', public)
````

````python
# Specific items matrame import cheyyadam
from mymath import area_circle, is_prime, PI

print('Direct access PI :', PI)              # mymath.PI avasaram ledu
print('area_circle(7)   :', round(area_circle(7), 4))
print('is_prime(97)     :', is_prime(97))

# Module alias tho import
import mymath as mm
print('Primes 1-50      :', mm.primes_upto(50))
````

## 5) `__name__ == '__main__'` — Module vs Script

Oka `.py` file ni **directly run** chesthe `__name__` = `'__main__'`. **Import** chesthe `__name__` = module name avutundi. Idi use chesi test code ni import chessinappudu run kaakapovadam control cheyyadam.

````python
# __name__ demonstration
print('Current __name__:', __name__)   # notebook/script lo run chesthe __main__

# mymath module __name__ chudadam
import mymath
print('mymath.__name__ :', mymath.__name__)   # 'mymath' — import chesamu

# Script tho explain cheyyadam
demo_script = '''
# demo_script.py

def helper():
    return "helper function"

# Idi only direct run chessinappudu execute avutundi
# import chessinappudu skip avutundi
if __name__ == "__main__":
    print("Direct ga run chesam — testing helper()")
    print(helper())
'''
with open('demo_script.py', 'w', encoding='utf-8') as f:
    f.write(demo_script)

# Import chessinappudu if block run avvadam confirm cheyyaadaniki
import demo_script
print('demo_script imported — if block skip ayyindi')  # if block run kaaledanukokam
print('helper() result:', demo_script.helper())
````

## 6) Package — Modules ni Folder lo Organize Cheyyadam

Package = oka **folder** with `__init__.py` file. Related modules ni group cheyyadam — large projects lo idi chaala important.

```
mypackage/
    __init__.py       ← package ga mark chestundi
    calculator.py     ← calculator module
    converter.py      ← unit converter module
    utils.py          ← utility functions
```

````python
import os

# Package folder structure create cheyyadam
os.makedirs('mypackage', exist_ok=True)   # folder create

# __init__.py — package initialization file
with open('mypackage/__init__.py', 'w', encoding='utf-8') as f:
    f.write(
        '# mypackage __init__.py\n'
        '# Idi package ga mark chestundi — import chessinappudu idi first run avutundi\n'
        'from .calculator import add, subtract, multiply, divide\n'
        'from .converter  import km_to_miles, celsius_to_fahrenheit\n'
        '__version__ = "1.0.0"\n'
        '__author__  = "Sai"\n'
    )

# calculator.py module
with open('mypackage/calculator.py', 'w', encoding='utf-8') as f:
    f.write(
        '# calculator.py — basic math operations\n'
        'def add(a, b):       return a + b\n'
        'def subtract(a, b):  return a - b\n'
        'def multiply(a, b):  return a * b\n'
        'def divide(a, b):\n'
        '    if b == 0: raise ValueError("Zero tho divide cheyyaraddu!")\n'
        '    return a / b\n'
    )

# converter.py module
with open('mypackage/converter.py', 'w', encoding='utf-8') as f:
    f.write(
        '# converter.py — unit conversions\n'
        'def km_to_miles(km):            return round(km * 0.621371, 4)\n'
        'def miles_to_km(miles):         return round(miles / 0.621371, 4)\n'
        'def celsius_to_fahrenheit(c):   return round((c * 9/5) + 32, 2)\n'
        'def fahrenheit_to_celsius(f):   return round((f - 32) * 5/9, 2)\n'
        'def kg_to_pounds(kg):           return round(kg * 2.20462, 4)\n'
    )

# utils.py module
with open('mypackage/utils.py', 'w', encoding='utf-8') as f:
    f.write(
        '# utils.py — helper utilities\n'
        'def is_even(n):     return n % 2 == 0\n'
        'def clamp(v,lo,hi): return max(lo, min(v, hi))\n'
        'def flatten(lst):   return [x for sub in lst for x in sub]\n'
    )

print('Package structure created:')
for root, dirs, files in os.walk('mypackage'):
    level = root.replace('mypackage', '').count(os.sep)
    indent = '  ' * level
    print(f'{indent}{os.path.basename(root)}/')
    for file in files:
        print(f'{indent}  {file}')
````

````python
# Package import cheyyadam — different ways

# Way 1: Full module path import
from mypackage import calculator, converter, utils

print('=== Calculator ===')
print('add(10, 5)      :', calculator.add(10, 5))
print('subtract(10, 5) :', calculator.subtract(10, 5))
print('multiply(4, 6)  :', calculator.multiply(4, 6))
print('divide(15, 4)   :', calculator.divide(15, 4))

print('\n=== Converter ===')
print('100 km → miles  :', converter.km_to_miles(100))
print('37°C → F        :', converter.celsius_to_fahrenheit(37))
print('70 kg → pounds  :', converter.kg_to_pounds(70))

print('\n=== Utils ===')
print('is_even(8)      :', utils.is_even(8))
print('clamp(150,0,100):', utils.clamp(150, 0, 100))   # 150 ki 100 clamp
print('flatten         :', utils.flatten([[1,2],[3,4],[5]]))
````

````python
# Way 2: __init__.py lo define chesina shortcuts use cheyyadam
import mypackage

print('Via __init__.py shortcuts:')
print('add(3, 7)            :', mypackage.add(3, 7))             # calculator.add
print('km_to_miles(200)     :', mypackage.km_to_miles(200))      # converter
print('Package version      :', mypackage.__version__)
print('Author               :', mypackage.__author__)

# Way 3: Specific function direct import
from mypackage.calculator import divide
from mypackage.converter  import fahrenheit_to_celsius

print('divide(22, 7)        :', round(divide(22, 7), 6))
print('98.6 F → C           :', fahrenheit_to_celsius(98.6))

# Error handling demo — zero division
try:
    mypackage.divide(10, 0)
except ValueError as e:
    print('Error caught         :', e)
````

## 9) Important Built-in Modules — Quick Reference

Real projects lo daily use avye important modules ni brief ga cover chestam.

````python
# collections module — powerful data structures
from collections import Counter, defaultdict, namedtuple, deque

# Counter — elements count cheyyadam
words = ['apple', 'banana', 'apple', 'cherry', 'banana', 'apple']
counts = Counter(words)
print('Counter        :', counts)
print('Most common 2  :', counts.most_common(2))

# defaultdict — key not found aite default value
dd = defaultdict(list)     # missing key ki empty list default
dd['fruits'].append('apple')
dd['fruits'].append('mango')
dd['veggies'].append('carrot')
print('defaultdict    :', dict(dd))

# namedtuple — tuple with field names
Student = namedtuple('Student', ['name', 'age', 'marks'])
s = Student('Sai', 25, 92)
print('namedtuple     :', s)
print('Access by name :', s.name, s.marks)   # index kante readable

# deque — both ends efficient append/pop
dq = deque([1, 2, 3])
dq.appendleft(0)   # left side add
dq.append(4)       # right side add
print('deque          :', dq)
````

````python
# itertools — efficient iterators
import itertools

# combinations — C(n,r)
teams = ['Sai', 'Ramya', 'Kiran', 'Arjun']
pairs = list(itertools.combinations(teams, 2))
print(f'Pairs from {len(teams)} people ({len(pairs)} total):', pairs)

# permutations — P(n,r)
perms = list(itertools.permutations([1, 2, 3]))
print(f'Permutations of [1,2,3] ({len(perms)} total):', perms)

# chain — multiple iterables oka stream ga
combined = list(itertools.chain([1,2], [3,4], [5,6]))
print('chain result   :', combined)

# groupby — consecutive same-key elements group
data = [('A', 1), ('A', 2), ('B', 3), ('B', 4), ('A', 5)]
for key, group in itertools.groupby(data, key=lambda x: x[0]):
    print(f'  group {key}:', list(group))
````

## 10) Module Reload & Caching

Python oka module ni once matrame load chestundi — second `import` same cached version istundi. Development lo file change chesthe `importlib.reload()` use cheyyali.

````python
import importlib

# Module caching demonstrate cheyyadam
import mymath                       # first time — file load avutundi
import mymath as mm2                # second time — SAME object, no re-read
print('Same object?  :', mymath is mm2)   # True — cached

# sys.modules — loaded modules cache
print('mymath in sys.modules:', 'mymath' in __import__('sys').modules)

# Reload — file change chesthe fresh load
importlib.reload(mymath)            # file disk nunchi fresh ga load chestundi
print('Reloaded mymath successfully')

# Module attributes inspect cheyyadam
print('mymath.__file__    :', mymath.__file__)
print('mymath.__name__    :', mymath.__name__)
print('mymath.__doc__     :', str(mymath.__doc__).strip()[:40] if mymath.__doc__ else 'None')
````

## 11) Real-World Example — Project Structure

Production projects lo typical folder structure — modules + packages + main script combination.

````python
# Real project structure demonstrate cheyyadam
import os, json

structure = {
    'student_app/': {
        '__init__.py'   : '# main package',
        'main.py'       : '# entry point — python main.py',
        'models/': {
            '__init__.py': '',
            'student.py' : '# Student class',
            'course.py'  : '# Course class',
        },
        'utils/': {
            '__init__.py': '',
            'validators.py': '# input validation',
            'formatters.py': '# output formatting',
        },
        'requirements.txt': 'langchain\npandas\nfastapi',
    }
}

def show_structure(d, prefix=''):
    for name, content in d.items():
        if isinstance(content, dict):
            print(f'{prefix}📁 {name}')
            show_structure(content, prefix + '   ')
        else:
            comment = f'  # {content}' if content else ''
            print(f'{prefix}📄 {name}{comment}')

print('Typical Project Structure:')
print('='*40)
show_structure(structure)
print('='*40)
print('''
Import patterns:
  from student_app.models.student import Student
  from student_app.utils.validators import validate_marks
  from student_app import main
''')
````

## 12) Cleanup — Demo Files Remove Cheyyadam

````python
import os, shutil

# Demo files delete cheyyadam
for f in ('mymath.py', 'demo_script.py'):
    try:
        if os.path.exists(f):
            os.remove(f)
            print(f'{f} removed')
    except OSError as e:
        print(f'Could not remove {f}: {e}')

# Package folder delete cheyyadam
for folder in ('mypackage',):
    try:
        if os.path.exists(folder):
            shutil.rmtree(folder)
            print(f'{folder}/ removed')
    except OSError as e:
        print(f'Could not remove {folder}: {e}')
````
