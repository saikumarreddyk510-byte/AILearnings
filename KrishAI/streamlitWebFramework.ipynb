## Deep Architecture Notes

Step 1: User `streamlit run app.py` command run chestadu. Streamlit oka **local web server** (Tornado based) ni port 8501 meeda start chestundi.

Step 2: Browser aa server ki connect ayyi, oka **WebSocket** connection open chestundi. Ee WebSocket valle UI real-time ga update avvagalugutundi - page refresh cheyyalsina avasaram ledu.

Step 3: Server, mana `app.py` Python script ni **top nundi bottom varaku** run chestundi. Flask laaga multiple route functions ledu - **full script ye oka page**.

Step 4: Script lo `st.slider()`, `st.text_input()` laanti widgets untayi. Prathi widget oka **value** ni return chestundi (Example: slider position 25).

Step 5: Aa values tho mana logic (data filtering, ML prediction) run avutundi.

Step 6: `st.title()`, `st.dataframe()`, `st.line_chart()` laanti calls UI elements ni **order lo** collect chestayi - script lo ye order lo raastamo, page lo kuda ade order lo kanipistayi.

Step 7: Final UI browser ki WebSocket dwara push avutundi.

Step 8: **Most important:** User oka widget ni touch chesthe (slider move chesthe), Streamlit **whole script ni malli modati nundi** run chestundi - kotha widget values tho. Deenine **rerun model** antaru. Ee oka point ardham ayithe Streamlit lo 80% ardham ayinattey.

## 1. What is Streamlit Framework?

**Definition:** Streamlit ante oka **open-source Python library**, idi tho manam **pure Python code tho** data apps and machine learning web apps ni build cheyyachu - HTML, CSS, JavaScript **em raayakunda**.

**Detailed explanation:**

Normal ga oka ML model ni users ki chupinchalante ee steps kavali:

1. Backend API raayali (Flask / FastAPI).
2. Frontend raayali (HTML + CSS + JavaScript, or React).
3. Rendintini connect cheyyali (fetch calls, JSON handling, CORS settings...).
4. Rendintini deploy cheyyali.

Idi oka data scientist ki chala time waste - endukante vaari main job **model build cheyyadam**, website build cheyyadam kaadu.

**Streamlit ee problem ni solve chestundi:** Nuvvu just oka normal Python script raayi, Streamlit daanini automatic ga **web page** ga convert chestundi. Slider kavala? `st.slider()`. Table kavala? `st.dataframe()`. Chart kavala? `st.line_chart()`. Anthe - **oka line = oka UI element**.

**Kid-friendly analogy:** Flask ante oka **empty plot of land** laantidi - foundation, walls, paint anni nuvve build cheyyali (full control undi, kani time ekkuva pattandi). Streamlit ante **ready-made LEGO blocks set** laantidi - blocks (widgets) ni stack chesthe chaalu, ready-made house (web app) vachestundi. Prathi block already design chesi, paint chesi undi.

**Real-life ga eppudu use avutundi:**

- ML model ki quick **demo app** (client ki or manager ki chupinchadaniki).
- Internal **dashboards** (sales data, model metrics monitor cheyyadaniki).
- Data exploration tools (CSV upload chesi, filter chesi, charts chudadaniki).
- Prototypes - oka idea ni 30 minutes lo working app ga marchadaniki.

**Eppudu use cheyyakoodadu:** Public-facing production website (Example: e-commerce site), complex custom UI/UX design, or fine-grained login/authentication and routing kavalsinapudu. Akkada Flask / FastAPI / Django + React better.

## 2. Flask vs Streamlit - Eppudu Edi Vaadali?

| Point | Flask | Streamlit |
|---|---|---|
| Main purpose | General web apps and REST APIs | Data / ML apps and dashboards |
| Frontend | Manam HTML + CSS + JS raayali (Jinja2 templates) | **Zero frontend code** - anni Python |
| Routing | `@app.route("/home")` laaga multiple routes | Route ledu - **oka script = oka page** (multi-page optional) |
| Execution | Request vachinapudu **aa oka function** matrame run avutundi | Prathi interaction ki **full script** malli run avutundi |
| Learning curve | Medium (HTTP, templates, forms nerchukovali) | Very easy (Python thelisthe chaalu) |
| UI control | Full control (custom design possible) | Limited (Streamlit ichche design ye vastundi) |
| Best for | Production websites, APIs, backends | Prototypes, demos, internal dashboards |
| Run command | `python app.py` | `streamlit run app.py` |
| Default port | 5000 | 8501 |

**Simple rule of thumb:**

- **Backend API** kavala (mobile app or React frontend consume chestundi)? -> **Flask / FastAPI**.
- Nee **model** ni evaraina click chesi try cheyyali, and design meeda time waste cheyyakudadu? -> **Streamlit**.

**Note:** Rendu kalipi kuda vaadachu - Flask/FastAPI tho model API host chesi, Streamlit tho aa API ni call chese UI build cheyyadam production lo common pattern.

## 4. Streamlit Execution Model - MOST IMPORTANT Concept

**Detailed explanation:**

Flask lo: request vasthe, **aa oka route function** matrame run avutundi. Migatha code ni touch cheyyadu.

Streamlit lo: user **edaina** widget tho interact chesthe (button click chesina, slider move chesina, textbox lo type chesi Enter kottina), Streamlit **whole script ni line 1 nundi malli** run chestundi. Deenine **rerun** antaru.

```
User slider ni 10 -> 25 ki move chesadu
        |
        v
Streamlit: "Sare, script ni malli run chestanu"
        |
        v
app.py line 1 nundi run avutundi... kani ippudu st.slider() value 25 ni return chestundi
        |
        v
Kotha UI generate ayyi, purathanam UI place lo replace avutundi
```

**Kid-friendly analogy:** Oka **whiteboard** meeda nuvvu prathi sari full drawing ni erase chesi, kotha ga malli mothham geestunnav ani anuko. Kani nuvvu chala fast ga geestunnav, kaabatti chuse valla ki *aa oka part matrame change ayindi* ani anipistundi. Streamlit kuda alaanti pane chestundi.

**Deeni valla vachhe 3 important consequences:**

1. **Normal Python variables reset avtayi.** `counter = 0` ani script lo raasthe, prathi rerun ki adi malli 0 ye avutundi. Value ni gurthupettukovalante **`st.session_state`** vaadali (Section 10 lo detailed ga chudandi).

2. **Heavy work prathi sari malli jarugutundi.** Peddha CSV load cheyyadam or ML model train cheyyadam prathi rerun ki malli jariginte app chala slow avutundi. Deeniki solution **caching** (`@st.cache_data`, `@st.cache_resource`) - Section 11 lo chudandi.

3. **Code order = UI order.** Script lo `st.title()` ni `st.write()` kanna mundhu raasthe, page lo kuda title ye mundhu kanipistundi. Kaabatti UI layout ni control cheyyadam ante just code line order ni control cheyyadam.

**Button gurinchi special note:** `st.button()` click chesina **aa oka rerun** ki matrame `True` ni return chestundi. Next rerun lo adi malli `False` ki potundi. Idi beginners ki chala confusing - button state ni gurthupettukovali ante `st.session_state` vaadali.

**Rerun ki sambandhinchina useful functions:**

- `st.rerun()` -> manam manam ga script ni malli run cheyyamani cheppadaniki.
- `st.stop()` -> script ni akkade aapesi, taruvatha lines run avvakunda cheyyadaniki (Example: user input ivvakapothe aapesi wait cheyyadam).
- `st.form(...)` -> lopala unna widgets prathi change ki rerun cheyyakunda, **Submit** kottinapude oka sari rerun avvadaniki.

## 5. Text and Display Elements

**Detailed explanation:** Ivi page meeda text/content ni chupinchadaniki vaade basic elements. Anni `st.` tho start avtayi.

| Function | Enti chestundi |
|---|---|
| `st.title("...")` | Page ki biggest heading (HTML lo H1 laaga) |
| `st.header("...")` | Section heading (H2 laaga) |
| `st.subheader("...")` | Sub-section heading (H3 laaga) |
| `st.write(...)` | **Swiss army knife** - text, number, dataframe, chart, dict... deni ni ichchina correct ga display chestundi |
| `st.markdown("...")` | Markdown formatting tho text (bold, links, lists) |
| `st.text("...")` | Plain fixed-width text (formatting apply avvadhu) |
| `st.code("...", language="python")` | Syntax highlighting tho code block |
| `st.latex(r"...")` | Math formulas display cheyyadaniki |
| `st.divider()` | Horizontal line (section separator) |
| `st.success` / `st.info` / `st.warning` / `st.error` | Colored message boxes (green / blue / yellow / red) |
| `st.metric(label, value, delta)` | KPI number card (up/down arrow tho) |
| `st.json({...})` | Dictionary ni expandable JSON viewer laaga |

**`st.write()` enduku special?** Deeniki **magic** ani peru - nuvvu enti pass chesina, Streamlit type ni chusi automatic ga best display method ni choose chestundi. String ichchav -> text. DataFrame ichchav -> interactive table. Matplotlib figure ichchav -> chart. Dictionary ichchav -> JSON viewer.

**Magic commands:** Actually `st.write()` kuda raayakunda, just oka variable ni oka line lo raasthe (Example: `df` ani matrame), Streamlit automatic ga display chestundi - notebook laaga.

**Kid-friendly analogy:** `st.write()` ante oka **smart printer** laantidi - photo ichchina, text ichchina, table ichchina, dhaaniki tagattu correct format lo print chestundi.

````python
%%writefile streamlit_apps/01_text_elements.py
# ---------------- Architecture Diagram ----------------
# [streamlit run 01_text_elements.py]
#    |
# [Script top-to-bottom run avutundi]
#    |
# [st.title -> st.header -> st.write -> st.metric ... order lo UI elements collect avtayi]
#    |
# [Browser lo ade order lo page render avutundi]
#
# ---------------- Deep Architecture Notes ----------------
# Step 1: Streamlit ni "st" ga import chestam - idi anni UI functions ki entry point.
# Step 2: st.set_page_config() eppudu script lo MODATI streamlit command ga undali.
# Step 3: Migatha prathi st.* call oka UI element ni page ki add chestundi.
# Step 4: Code lo unna order ye page lo unna order - separate HTML/CSS avasaram ledu.

import streamlit as st  # Streamlit library ni "st" ane standard short name tho import chestunnam.
import pandas as pd  # Table/dataframe display demo kosam pandas ni import chestunnam.

st.set_page_config(  # Page-level settings ni set chestunnam; idi modati streamlit command ga undali.
    page_title="Text Elements Demo",  # Browser tab meeda kanipinche title.
    page_icon="📘",  # Browser tab meeda kanipinche icon (emoji vaadachu).
    layout="centered",  # Content ni page middle lo pedatundi ("wide" ante full width vastundi).
)

st.title("Streamlit Text Elements")  # Page ki biggest heading (H1 laaga) add chestunnam.
st.header("1. Headings")  # Section heading (H2 laaga) add chestunnam.
st.subheader("Idi oka subheader (H3 laaga)")  # Sub-section heading (H3 laaga) add chestunnam.

st.divider()  # Rendu sections madhya horizontal line geestunnam.

st.header("2. Text display methods")  # Rendo section ki heading.
st.write("st.write() - idi anni type of data ni handle chese smart function.")  # Simple string ni display chestunnam.
st.markdown("**Bold text**, *italic text*, and [oka link](https://streamlit.io)")  # Markdown formatting tho text display chestunnam.
st.text("st.text() - idi plain fixed-width text, formatting apply avvadhu.")  # Formatting lekunda plain text display chestunnam.
st.caption("st.caption() - chinna grey text, notes or footnotes ki use avutundi.")  # Chinna caption text ni display chestunnam.

st.divider()  # Inko separator line.

st.header("3. Code and Math")  # Moodo section ki heading.
st.code("def add(a, b):\n    return a + b", language="python")  # Python code ni syntax highlighting tho display chestunnam.
st.latex(r"y = mx + c")  # Math formula ni LaTeX format lo display chestunnam.

st.divider()  # Inko separator line.

st.header("4. Status message boxes")  # Naalgo section ki heading.
st.success("Success! Model successfully train ayindi.")  # Green color success box display chestunnam.
st.info("Info: Ee data 2024 nundi teesukunnam.")  # Blue color info box display chestunnam.
st.warning("Warning: Missing values unnayi, jagratha.")  # Yellow color warning box display chestunnam.
st.error("Error: File dorakaledu.")  # Red color error box display chestunnam.

st.divider()  # Inko separator line.

st.header("5. Metric cards")  # Aido section ki heading.
st.metric(  # KPI number card ni display chestunnam.
    label="Model Accuracy",  # Card meeda kanipinche label.
    value="94.5%",  # Card lo kanipinche main value.
    delta="+2.1%",  # Previous value tho polisthe entha marindo (green up arrow tho chupistundi).
)

st.divider()  # Inko separator line.

st.header("6. st.write() magic - different types")  # Aaro section ki heading.
st.write("String type:", "Hello")  # Multiple arguments ni oke line lo display chestunnam.
st.write("Number type:", 42)  # Integer value ni display chestunnam.
st.write("Dictionary type:", {"name": "Sai", "course": "Streamlit"})  # Dictionary ni JSON viewer laaga display chestunnam.

sample_df = pd.DataFrame({"Name": ["Sai", "Kumar"], "Marks": [90, 85]})  # Demo kosam chinna DataFrame create chestunnam.
st.write("DataFrame type (automatic ga interactive table avutundi):")  # DataFrame display ki mundhu oka label raastunnam.
st.write(sample_df)  # DataFrame ni pass chesthe Streamlit automatic ga table ga render chestundi.
````

## 6. Input Widgets - User nundi Data Teesukovadam

**Detailed explanation:** Widgets ante user tho interact ayye UI controls - textbox, slider, dropdown, button, checkbox laantivi. Streamlit lo prathi widget **oka value ni return chestundi**, and aa value ni manam normal Python variable lo pettukuni vaadukuntam.

```python
name = st.text_input("Nee peru enti?")   # user type chesina text ee variable lo vastundi
age  = st.slider("Nee age", 0, 100, 25)  # slider position value ee variable lo vastundi
```

Ee **return value pattern** ye Streamlit ni chala simple ga chestundi - Flask lo laaga `request.form.get(...)` raayalsina avasaram ledu.

| Widget | Enti return chestundi | Example |
|---|---|---|
| `st.text_input(label)` | User type chesina **string** | Name, email |
| `st.text_area(label)` | Multi-line **string** | Review, description |
| `st.number_input(label, min_value, max_value)` | **int / float** | Age, price |
| `st.slider(label, min, max, default)` | **int / float** (range kuda ivvachu) | Threshold, salary range |
| `st.select_slider(label, options)` | Options lo nundi **oka value** | Small/Medium/Large |
| `st.selectbox(label, options)` | Dropdown lo select chesina **oka option** | City select cheyyadam |
| `st.multiselect(label, options)` | Select chesina options **list** | Multiple features |
| `st.radio(label, options)` | Radio buttons lo select chesina **oka option** | Gender, Yes/No |
| `st.checkbox(label)` | **True / False** | Terms accept cheyyadam |
| `st.toggle(label)` | **True / False** (modern switch look) | Dark mode on/off |
| `st.button(label)` | Click chesthe **True**, lekapothe **False** | Submit, Predict |
| `st.date_input(label)` | **datetime.date** object | Booking date |
| `st.time_input(label)` | **datetime.time** object | Appointment time |
| `st.file_uploader(label)` | Upload chesina **file object** (or None) | CSV, image |
| `st.color_picker(label)` | Hex color **string** | Chart color |

**`key` parameter enduku important?** Rendu widgets ki same label unte Streamlit confuse avutundi (DuplicateWidgetID error vastundi). Appudu `key="unique_name"` ivvali. Inka, `key` ichchaka aa widget value automatic ga `st.session_state["unique_name"]` lo kuda available avutundi.

**`st.form()` enduku vaadali?** Normal ga prathi widget change ki script rerun avutundi. Kani 5 inputs unna form lo prathi input ki rerun avvadam waste. `st.form()` lopala widgets pedithe, **Submit button** kottinapude oka sari rerun avutundi.

**Kid-friendly analogy:** Widgets ante **restaurant menu order form** laantivi. Nuvvu tick chesina items (checkbox), select chesina size (dropdown), raasina name (textbox) - anni waiter (Streamlit) collect chesi kitchen (nee Python code) ki teesukelltadu.

````python
%%writefile streamlit_apps/02_widgets.py
# ---------------- Architecture Diagram ----------------
# [User widget tho interact chestadu (type/slide/click)]
#    |
# [Streamlit full script ni rerun chestundi]
#    |
# [Prathi st.<widget>() call kotha value ni return chestundi]
#    |
# [Aa values tho mana Python logic run avutundi]
#    |
# [Result ni st.write() tho display chestam]
#
# ---------------- Deep Architecture Notes ----------------
# Step 1: Prathi widget oka Python value ni return chestundi - request.form laanti extra code avasaram ledu.
# Step 2: User widget ni marchinapudu script malli run ayyi, kotha value tho result recalculate avutundi.
# Step 3: st.form() lopala unna widgets Submit kottinapude oka sari rerun ni trigger chestayi.
# Step 4: key="..." ichchi widget value ni st.session_state nundi kuda access cheyyachu.

import streamlit as st  # Streamlit library ni import chestunnam.

st.set_page_config(page_title="Widgets Demo", page_icon="🎛️")  # Page title and icon ni set chestunnam.

st.title("Streamlit Input Widgets")  # Page ki main heading add chestunnam.

st.header("1. Text inputs")  # Modati section heading.
user_name = st.text_input("Nee peru enti?", value="Sai")  # Single-line text input; default value "Sai".
user_bio = st.text_area("Nee gurinchi konni lines raayi:", height=100)  # Multi-line text input; height pixels lo.
st.write(f"Namaskaram **{user_name}**! Nuvvu {len(user_bio)} characters raasav.")  # Inputs ni vaadi oka message display chestunnam.

st.divider()  # Section separator line.

st.header("2. Number inputs and sliders")  # Rendo section heading.
user_age = st.number_input("Nee age:", min_value=1, max_value=120, value=25, step=1)  # Number box; min/max/default/step ivvocchu.
experience_years = st.slider("Experience (years):", 0, 40, 3)  # Slider: min 0, max 40, default 3.
salary_range = st.slider("Expected salary range (LPA):", 0, 100, (10, 30))  # Tuple default ivvadam valla RANGE slider avutundi.
st.write(f"Age: {user_age} | Experience: {experience_years} years | Salary range: {salary_range}")  # Anni values ni display chestunnam.

st.divider()  # Section separator line.

st.header("3. Selection widgets")  # Moodo section heading.
selected_city = st.selectbox("Nee city select cheyyi:", ["Hyderabad", "Bangalore", "Chennai", "Pune"])  # Dropdown; select chesina option ni return chestundi.
selected_skills = st.multiselect("Nee skills select cheyyi:", ["Python", "SQL", "ML", "LangChain"], default=["Python"])  # Multiple options select cheyyadaniki; list ni return chestundi.
work_mode = st.radio("Work mode:", ["Remote", "Hybrid", "Onsite"], horizontal=True)  # Radio buttons; horizontal=True valla pakka pakkana kanipistayi.
st.write(f"City: {selected_city} | Skills: {selected_skills} | Mode: {work_mode}")  # Selections ni display chestunnam.

st.divider()  # Section separator line.

st.header("4. Boolean widgets")  # Naalgo section heading.
show_details = st.checkbox("Extra details chupinchu")  # Checkbox; tick chesthe True return chestundi.
dark_mode = st.toggle("Dark mode on")  # Toggle switch; on chesthe True return chestundi.
if show_details:  # Checkbox True unte matrame lopala code run avutundi.
    st.info(f"Details: {user_name}, {user_age} years, {selected_city}. Dark mode = {dark_mode}")  # Conditional ga extra info display chestunnam.

st.divider()  # Section separator line.

st.header("5. Button - oka rerun ki matrame True")  # Aido section heading.
if st.button("Namaskaram cheppu"):  # Button click chesina RERUN lo matrame True vastundi.
    st.success(f"Namaskaram {user_name}! Meeru {selected_city} nundi.")  # Button click ayinapudu success message chupistunnam.

st.divider()  # Section separator line.

st.header("6. Form - anni inputs oke sari submit cheyyadam")  # Aaro section heading.
with st.form("job_application_form"):  # Form context start chestunnam; "job_application_form" ee form ki unique key.
    applicant_email = st.text_input("Email:")  # Form lopala unna text input - ide rerun ni trigger cheyyadu.
    expected_ctc = st.number_input("Expected CTC (LPA):", min_value=0, max_value=200, value=12)  # Form lopala number input.
    notice_period = st.selectbox("Notice period:", ["Immediate", "15 days", "30 days", "60 days"])  # Form lopala dropdown.
    form_submitted = st.form_submit_button("Application Submit cheyyi")  # Form ki mandatory submit button; click chesthe True.

if form_submitted:  # Submit button click ayinaka matrame lopala code run avutundi.
    st.success("Application submit ayindi!")  # Success message display chestunnam.
    st.json({  # Submit chesina data ni JSON viewer laaga display chestunnam.
        "name": user_name,  # Form bayata unna name value.
        "email": applicant_email,  # Form lopala unna email value.
        "expected_ctc": expected_ctc,  # Form lopala unna CTC value.
        "notice_period": notice_period,  # Form lopala unna notice period value.
    })
````

## 7. DataFrames, Tables and Charts

**Detailed explanation:** Streamlit ki pandas DataFrames tho **native support** undi - ade deenini data scientists ki favourite ga chesindi.

**Data display options:**

| Function | Enti chestundi |
|---|---|
| `st.dataframe(df)` | **Interactive** table - sort cheyyachu, scroll cheyyachu, search cheyyachu, fullscreen ki velladachu |
| `st.data_editor(df)` | **Edit cheyyagaligina** table - user cells lo values marchachu, marchina df return avutundi |
| `st.table(df)` | **Static** table - sorting ledu, chinna tables ki baaguntundi |
| `st.metric(...)` | Oka single KPI number |
| `st.json(dict)` | JSON viewer |

**Built-in charts (oka line lo chart):**

| Function | Eppudu vaadali |
|---|---|
| `st.line_chart(df)` | Time series, trends (Example: rojula vaari sales) |
| `st.bar_chart(df)` | Categories ni compare cheyyadaniki |
| `st.area_chart(df)` | Line chart + kinda fill (cumulative feel) |
| `st.scatter_chart(df)` | Rendu numeric columns madhya relation |
| `st.map(df)` | `lat` and `lon` columns unna data ni map meeda points ga |

## 10. Session State - Reruns Madhya Value ni Gurthupettukovadam

**Detailed explanation:**

Section 4 lo chusam - prathi interaction ki script full ga rerun avutundi. Ante **normal Python variables prathi sari reset avtayi**.

```python
counter = 0                          # Prathi rerun ki idi malli 0 ye
if st.button("Increase"):
    counter += 1                     # 1 ki vellutundi
st.write(counter)                    # Kani next rerun lo malli 0! ❌
```

Deeniki solution **`st.session_state`** - idi oka **dictionary** laanti object, and idi reruns madhya **survive** avutundi (browser tab open unnantha varaku).

```python
if "counter" not in st.session_state:     # Modati sari matrame initialize chestam
    st.session_state.counter = 0
if st.button("Increase"):
    st.session_state.counter += 1         # Ippudu value gurthu untundi ✅
st.write(st.session_state.counter)
```

**Rendu ways to access:**

- Attribute style: `st.session_state.counter`
- Dictionary style: `st.session_state["counter"]`

Rendu oke - ye style aina vaadachu.

**Widget `key` and session_state link:** Oka widget ki `key="user_name"` ichchinappudu, aa widget value automatic ga `st.session_state["user_name"]` lo store avutundi. Ante widget value ni script lo enni chotla nundi ayina access cheyyachu.

**Eppudu vaadali:**

- Counters, scores, click counts.
- Chat history (chatbot apps lo messages list).
- Login status (`st.session_state.logged_in = True`).
- Multi-step wizard lo current step number.
- Button click state ni gurthupettukovadam.

**Important gotcha:** Session state **prathi browser session ki separate**. Nuvvu tab close chesi malli open chesthe, anni values pothayi. Inko user vere browser lo open chesthe, vaariki vere session state untundi - so idi database kaadu, permanent storage ki `st.session_state` vaadaraadu.

**Kid-friendly analogy:** Session state ante nee **school bag** laantidi. Prathi period (rerun) ki class marutundi, kani bag lo unna books, tiffin box nee daggare untayi. Kani school nundi intiki vellipothe (tab close), bag lopala unnavi anni maayam ayipothayi.

````python
%%writefile streamlit_apps/06_session_state.py
# ---------------- Architecture Diagram ----------------
# [User button click chestadu]
#    |
# [Script full ga rerun avutundi -> normal variables reset]
#    |
# [st.session_state matram old values ni gurthupettukuntundi]
#    |
# [Counter / To-do list / Chat history laantivi correct ga pani chestayi]
#
# ---------------- Deep Architecture Notes ----------------
# Step 1: Modati run lo "key in st.session_state" check chesi default value ni set chestam.
# Step 2: Taruvatha reruns lo aa key already unnadhu, so default malli set avvadhu - value survive avutundi.
# Step 3: Widget ki key="..." ichchinappudu, aa value automatic ga session_state lo kuda vastundi.
# Step 4: Session state prathi browser session ki separate - permanent storage kaadu.

import streamlit as st  # Streamlit library ni import chestunnam.

st.set_page_config(page_title="Session State", page_icon="🧠")  # Page settings set chestunnam.

st.title("Session State - Value ni Gurthupettukovadam")  # Page main heading.

st.header("1. Session state LEKUNDA (idi work avvadhu)")  # Modati section heading - problem ni chupinchadaniki.
normal_counter = 0  # Normal Python variable - prathi rerun ki idi malli 0 ki reset avutundi.
if st.button("Normal counter penchu"):  # Button click ayithe lopala code run avutundi.
    normal_counter += 1  # Value ni 1 penchutunnam - kani ee change next rerun lo maayam avutundi.
st.write(f"Normal counter value: {normal_counter} (eppudu 0 or 1 ye untundi ❌)")  # Problem ni user ki chupistunnam.

st.divider()  # Section separator line.

st.header("2. Session state TO (idi correct ga work avutundi)")  # Rendo section heading - solution.
if "click_count" not in st.session_state:  # "click_count" key already unda ani check chestunnam.
    st.session_state.click_count = 0  # Modati run lo matrame default value 0 ni set chestunnam.

increment_col, reset_col = st.columns(2)  # Rendu buttons pakka pakkana pettadaniki rendu columns.
if increment_col.button("➕ Counter penchu"):  # Modati column lo increment button.
    st.session_state.click_count += 1  # Session state lo unna value ni 1 penchutunnam - idi survive avutundi.
if reset_col.button("🔄 Counter reset cheyyi"):  # Rendo column lo reset button.
    st.session_state.click_count = 0  # Counter ni malli 0 ki set chestunnam.

st.metric("Session counter", st.session_state.click_count)  # Current counter value ni metric card ga chupistunnam.

st.divider()  # Section separator line.

st.header("3. To-do list (list ni session state lo store cheyyadam)")  # Moodo section heading.
if "todo_items" not in st.session_state:  # Todo list key unda ani check chestunnam.
    st.session_state.todo_items = []  # Modati sari khaali list tho initialize chestunnam.

new_task = st.text_input("Kotha task raayi:", key="new_task_input")  # Task text input; key ivvadam valla session_state lo kuda vastundi.
if st.button("Task add cheyyi") and new_task.strip():  # Button click ayyi, task khaali kaakapothe matrame add chestunnam.
    st.session_state.todo_items.append(new_task.strip())  # Session state list ki kotha task ni add chestunnam.

if st.session_state.todo_items:  # List lo tasks unte matrame display chestunnam.
    for task_index, task_text in enumerate(st.session_state.todo_items, start=1):  # Prathi task ni number tho loop chestunnam.
        st.write(f"{task_index}. {task_text}")  # Task ni number tho display chestunnam.
    if st.button("Anni tasks clear cheyyi"):  # Anni tasks ni tholaginche button.
        st.session_state.todo_items = []  # List ni khaali chestunnam.
else:  # Tasks emi lekapothe.
    st.info("Inka tasks emi ledu. Paina oka task add cheyyi.")  # User ki guidance message.

st.divider()  # Section separator line.

st.header("4. Widget key automatic ga session state lo vastundi")  # Naalgo section heading.
st.text_input("Nee peru (key='user_name'):", key="user_name")  # key ichchina text input.
st.write("st.session_state['user_name'] value:", st.session_state.get("user_name", ""))  # Session state nundi ade value ni chadhuvutunnam.

st.divider()  # Section separator line.

st.header("5. Full session state content")  # Aido section heading.
st.json({key: str(value) for key, value in st.session_state.items()})  # Session state lo unna anni keys/values ni JSON ga chupistunnam.
````

## 11. Caching - App ni Fast Ga Cheyyadam

**Detailed explanation:**

Rerun model valla vachhe peddha problem: **heavy work prathi sari malli jarugutundi**. Example: 500 MB CSV load cheyyadam, or ML model train cheyyadam. User oka slider move chesthe, ee panulu anni malli modati nundi start avthe app unusable ga slow avutundi.

**Caching solution:** Function result ni **gurthupettukoni**, same inputs malli vasthe, function ni run cheyyakunda **store chesina result ni** ichesthundi.

**Rendu decorators unnayi - vaati madhya difference chala important:**

| Decorator | Deniki | Ela store chestundi |
|---|---|---|
| `@st.cache_data` | **Data** ki - DataFrames, lists, dicts, API responses, numbers | Result ni **copy** chesi store chestundi (serialize) |
| `@st.cache_resource` | **Resources** ki - ML models, DB connections, tokenizers | Actual **oke object** ni share chestundi (copy ledu) |

```python
@st.cache_data                       # DataFrame return chese function ki
def load_data(path):
    return pd.read_csv(path)         # Idi oka sari matrame run avutundi

@st.cache_resource                   # Model return chese function ki
def load_model():
    return joblib.load("model.pkl")  # Model memory lo oka sari matrame load avutundi
```

**Enduku ee difference?** DataFrame ni oke object ga share chesthe, oka user modify chesthe migatha valandariki kuda change avutundi - danger. Kaabatti `cache_data` prathi sari copy istundi. Kani ML model ni copy cheyyadam waste (memory ekkuva, slow) - and daanini evaru modify cheyyaru, kaabatti `cache_resource` oke object ni share chestundi.

**Ela pani chestundi (internally):** Streamlit function name + arguments ni kalipi oka **hash key** create chestundi. Same key malli vasthe, cache lo unna result ni istundi. Arguments marithe (path marithe), kotha key avutundi, so function malli run avutundi.

**Useful parameters:**

```python
@st.cache_data(ttl=3600)          # 1 hour taruvatha cache expire avutundi (time-to-live seconds lo)
@st.cache_data(max_entries=10)    # Maximum 10 different results matrame store chestundi
@st.cache_data(show_spinner="Data load avutundi...")   # Load avutunnapudu custom spinner message
```

Cache ni manually clear cheyyadaniki: `st.cache_data.clear()` or terminal lo `streamlit cache clear`.

**Kid-friendly analogy:** Nuvvu oka maths sum ni chala kashtapadi solve chesav. Answer ni oka **notebook lo raasukunnav**. Next time ade sum vasthe, malli solve cheyyakunda notebook lo chusi answer cheppestav - ade caching. Kani sum lo numbers marithe (arguments marithe), malli kotha ga solve cheyyali.

**Common mistake:** `@st.cache_data` ni widgets unna function meeda pettaraadu (`st.slider()` laantivi lopala unte). Cached function lo UI elements raayaru - andhulo pure computation matrame undali.

````python
%%writefile streamlit_apps/07_caching.py
# ---------------- Architecture Diagram ----------------
# [User widget marchadu -> rerun]
#    |
# [Cached function call avutundi]
#    |
# [Cache lo same arguments unna result unda? ]
#    |-- Undi  -> stored result ni instant ga icchestundi (function run avvadhu)
#    |-- Ledu  -> function run ayyi, result ni cache lo pettukuntundi
#
# ---------------- Deep Architecture Notes ----------------
# Step 1: @st.cache_data data (DataFrame/list/dict) ki - prathi sari oka COPY ni istundi.
# Step 2: @st.cache_resource models/connections ki - andarikee OKE object ni share chestundi.
# Step 3: Cache key = function name + arguments; arguments marithe function malli run avutundi.
# Step 4: Cached function lopala st.slider laanti widgets raayakudadu - pure computation matrame undali.

import streamlit as st  # Streamlit library ni import chestunnam.
import pandas as pd  # Data handling kosam pandas import chestunnam.
import numpy as np  # Random data generate cheyyadaniki numpy import chestunnam.
import time  # Slow work ni simulate cheyyadaniki time module import chestunnam.

st.set_page_config(page_title="Caching Demo", page_icon="⚡")  # Page settings set chestunnam.

st.title("Caching - App ni Fast Ga Cheyyadam")  # Page main heading.


@st.cache_data(show_spinner="Data load avutundi... (idi slow)")  # Ee function result ni cache lo store cheyyamani cheptunnam.
def load_big_dataset(number_of_rows):  # Peddha dataset ni load chese function (slow work ni simulate chestundi).
    time.sleep(2)  # 2 seconds delay - real ga file chadavadam/API call laanti slow pani ni simulate chestunnam.
    np.random.seed(0)  # Random values consistent ga undadaniki seed set chestunnam.
    return pd.DataFrame({  # Sample DataFrame ni create chesi return chestunnam.
        "Product": np.random.choice(["Laptop", "Mobile", "Tablet"], number_of_rows),  # Random product names column.
        "Price": np.random.randint(5000, 90000, number_of_rows),  # Random price values column.
        "Units": np.random.randint(1, 50, number_of_rows),  # Random units sold column.
    })


@st.cache_resource  # ML model laanti heavy RESOURCE ki ee decorator vaadali - copy cheyyakunda oke object ni share chestundi.
def create_expensive_resource():  # Oka heavy resource (model/DB connection) ni create chese function.
    time.sleep(2)  # Resource create avvadaniki time pattundi ani simulate chestunnam.
    return {"resource_name": "FakeModel", "created_at": time.strftime("%H:%M:%S")}  # Resource ni dictionary ga return chestunnam.


st.header("1. @st.cache_data demo")  # Modati section heading.
row_count = st.slider("Enni rows kavali?", 100, 1000, 300, step=100)  # Rows count ni select chese slider.

start_time = time.time()  # Function call ki mundhu time ni note chestunnam.
dataset = load_big_dataset(row_count)  # Cached function ni call chestunnam - modati sari slow, taruvatha instant.
elapsed_seconds = time.time() - start_time  # Function ki entha time pattindo calculate chestunnam.

st.write(f"Load time: **{elapsed_seconds:.2f} seconds**")  # Entha time pattindo display chestunnam.
st.caption("Modati sari ~2 seconds pattundi. Ade slider value ki malli vasthe 0.00 seconds - ade caching power.")  # Explanation caption.
st.dataframe(dataset.head(), width="stretch")  # Data modati konni rows ni chupistunnam.

st.divider()  # Section separator line.

st.header("2. @st.cache_resource demo")  # Rendo section heading.
expensive_resource = create_expensive_resource()  # Cached resource ni teesukuntunnam - oka sari matrame create avutundi.
st.json(expensive_resource)  # Resource details ni JSON ga chupistunnam.
st.caption("created_at time MAARADU - ante resource oka sari matrame create ayindi, prathi rerun ki kaadu.")  # Important observation.

st.divider()  # Section separator line.

st.header("3. Cache ni clear cheyyadam")  # Moodo section heading.
clear_data_col, clear_resource_col = st.columns(2)  # Rendu buttons ki rendu columns.
if clear_data_col.button("Data cache clear cheyyi"):  # Data cache ni clear chese button.
    st.cache_data.clear()  # Anni @st.cache_data results ni tholagistunnam.
    st.success("Data cache clear ayindi. Ippudu malli slow avutundi.")  # Confirmation message.
if clear_resource_col.button("Resource cache clear cheyyi"):  # Resource cache ni clear chese button.
    st.cache_resource.clear()  # Anni @st.cache_resource objects ni tholagistunnam.
    st.success("Resource cache clear ayindi. Kotha resource create avutundi.")  # Confirmation message.
````

````python
%%writefile streamlit_apps/08_building_web_app.py
# ---------------- Architecture Diagram ----------------
# [streamlit run 08_building_web_app.py]
#    |
# [set_page_config -> page title/icon/layout]
#    |
# [@st.cache_data load_employee_data() -> DataFrame (oka sari matrame generate avutundi)]
#    |
# [Sidebar filters: department / experience / salary]
#    |
# [DataFrame filter avutundi]
#    |
# [KPI metrics (4 columns) -> Tabs (Table | Charts | Insights) -> Download button]
#    |
# [Browser lo full dashboard kanipistundi]
#
# ---------------- Deep Architecture Notes ----------------
# Step 1: set_page_config eppudu modati streamlit command ga undali - lekapothe error vastundi.
# Step 2: Data generation heavy pani - @st.cache_data valla prathi rerun ki malli jaragadu.
# Step 3: Sidebar widgets values ni return chestayi; user vaatini marchinapudu full script rerun avutundi.
# Step 4: Filtering pandas boolean masks tho jarugutundi - pure Python, Streamlit specific em ledu.
# Step 5: Metrics, tabs, charts anni Python objects nundi direct ga render avtayi - HTML avasaram ledu.
# Step 6: download_button tho user filtered result ni CSV ga teesukovachu.

import streamlit as st  # Streamlit library ni import chestunnam - anni UI elements deeni nundi vastayi.
import pandas as pd  # Employee data ni DataFrame ga handle cheyyadaniki pandas import chestunnam.
import numpy as np  # Sample random data generate cheyyadaniki numpy import chestunnam.

st.set_page_config(  # Page level settings - idi script lo MODATI streamlit command ga undali.
    page_title="Employee Data Explorer",  # Browser tab meeda kanipinche title.
    page_icon="👨‍💼",  # Browser tab meeda kanipinche icon.
    layout="wide",  # Content ni full screen width lo pedutundi.
    initial_sidebar_state="expanded",  # Sidebar ni default ga open ga chupistundi.
)


@st.cache_data  # Ee function result ni cache lo pettamani cheptunnam - prathi rerun ki malli generate avvadhu.
def load_employee_data(total_employees=300):  # Sample employee data ni create chese function.
    np.random.seed(7)  # Random values prathi sari same ga raavadaniki seed fix chestunnam.
    departments = ["Engineering", "Sales", "HR", "Marketing", "Finance"]  # Available department names list.
    cities = ["Hyderabad", "Bangalore", "Chennai", "Pune", "Mumbai"]  # Available city names list.
    experience_values = np.random.randint(0, 21, total_employees)  # Prathi employee ki 0-20 years experience.
    return pd.DataFrame({  # Anni columns ni kalipi oka DataFrame ga return chestunnam.
        "EmployeeID": [f"EMP{index:04d}" for index in range(1, total_employees + 1)],  # EMP0001 laanti unique IDs generate chestunnam.
        "Department": np.random.choice(departments, total_employees),  # Prathi employee ki random department assign chestunnam.
        "City": np.random.choice(cities, total_employees),  # Prathi employee ki random city assign chestunnam.
        "Experience": experience_values,  # Paina generate chesina experience values.
        "Salary": (experience_values * 1.2 + np.random.uniform(3, 8, total_employees)).round(1),  # Experience base ga salary (LPA) calculate chestunnam.
        "Rating": np.random.choice([1, 2, 3, 4, 5], total_employees, p=[0.05, 0.10, 0.30, 0.35, 0.20]),  # Performance rating - probabilities tho realistic ga.
    })


employee_data = load_employee_data()  # Cached function ni call chesi employee data ni teesukuntunnam.

if "filter_change_count" not in st.session_state:  # Session state lo counter key unda ani check chestunnam.
    st.session_state.filter_change_count = 0  # Modati run lo counter ni 0 tho initialize chestunnam.

with st.sidebar:  # Ee block lopala unna anni widgets sidebar lo kanipistayi.
    st.header("🔍 Filters")  # Sidebar ki heading add chestunnam.
    selected_departments = st.multiselect(  # Departments select chese multiselect widget.
        "Departments:",  # Widget label.
        options=sorted(employee_data["Department"].unique()),  # Data lo unna unique departments ne options ga istunnam.
        default=sorted(employee_data["Department"].unique()),  # Default ga anni departments selected.
    )
    experience_range = st.slider(  # Experience range ni select chese range slider.
        "Experience (years):",  # Widget label.
        min_value=int(employee_data["Experience"].min()),  # Data lo unna minimum experience.
        max_value=int(employee_data["Experience"].max()),  # Data lo unna maximum experience.
        value=(0, 20),  # Tuple ivvadam valla RANGE slider avutundi.
    )
    minimum_rating = st.select_slider(  # Minimum rating ni select chese slider.
        "Minimum rating:",  # Widget label.
        options=[1, 2, 3, 4, 5],  # Available rating values.
        value=1,  # Default ga anni ratings allow chestunnam.
    )
    st.divider()  # Sidebar lo separator line.
    st.caption(f"Filters applied: {st.session_state.filter_change_count} times")  # Enni sarlu filters run ayyayo chupistunnam.

st.session_state.filter_change_count += 1  # Prathi rerun ki counter ni penchutunnam.

filtered_employees = employee_data[  # Anni filter conditions ni kalipi data ni filter chestunnam.
    (employee_data["Department"].isin(selected_departments))  # Department select chesina list lo unnada ani check.
    & (employee_data["Experience"].between(experience_range[0], experience_range[1]))  # Experience selected range lo unda ani check.
    & (employee_data["Rating"] >= minimum_rating)  # Rating minimum kanna ekkuva unda ani check.
]

st.title("👨‍💼 Employee Data Explorer")  # Main area lo page title.
st.write("Sidebar lo filters marchi, results instant ga update avvadam chudu.")  # Chinna instruction text.

if filtered_employees.empty:  # Filters valla data emi migalaledha ani check chestunnam.
    st.warning("Ee filters ki matching employees evaru leru. Filters ni marchi try cheyyi.")  # User ki warning message.
    st.stop()  # Script ni ikkade aapesthunnam - taruvatha code run avvadu.

kpi_count, kpi_salary, kpi_experience, kpi_rating = st.columns(4)  # Naalugu KPI cards ki naalugu columns create chestunnam.
kpi_count.metric("Employees", len(filtered_employees))  # Filtered employees count ni metric ga chupistunnam.
kpi_salary.metric("Avg Salary (LPA)", f"{filtered_employees['Salary'].mean():.1f}")  # Average salary ni metric ga chupistunnam.
kpi_experience.metric("Avg Experience", f"{filtered_employees['Experience'].mean():.1f} yrs")  # Average experience ni chupistunnam.
kpi_rating.metric("Avg Rating", f"{filtered_employees['Rating'].mean():.2f}")  # Average rating ni chupistunnam.

table_tab, charts_tab, insights_tab = st.tabs(["📄 Table", "📈 Charts", "💡 Insights"])  # Moodu tabs create chestunnam.

with table_tab:  # Modati tab content.
    st.subheader("Filtered employees")  # Tab lo oka subheading.
    st.dataframe(filtered_employees, width="stretch", hide_index=True)  # Filtered data ni interactive table ga chupistunnam.
    st.download_button(  # Filtered data ni CSV ga download chese button.
        label="⬇️ Filtered data CSV download cheyyi",  # Button meeda kanipinche text.
        data=filtered_employees.to_csv(index=False),  # DataFrame ni CSV string ga convert chestunnam.
        file_name="filtered_employees.csv",  # Download ayye file peru.
        mime="text/csv",  # Browser ki file type ni cheptunnam.
    )

with charts_tab:  # Rendo tab content.
    chart_left, chart_right = st.columns(2)  # Rendu charts pakka pakkana pettadaniki rendu columns.
    with chart_left:  # Ednam column lo modati chart.
        st.subheader("Department wise count")  # Chart ki heading.
        st.bar_chart(filtered_employees["Department"].value_counts())  # Prathi department lo enni mandi unnaro bar chart ga.
    with chart_right:  # Kudi column lo rendo chart.
        st.subheader("City wise average salary")  # Chart ki heading.
        st.bar_chart(filtered_employees.groupby("City")["Salary"].mean())  # Prathi city average salary ni bar chart ga.
    st.subheader("Experience vs Average Salary")  # Moodo chart ki heading.
    st.line_chart(filtered_employees.groupby("Experience")["Salary"].mean())  # Experience penchithe salary ela marutundo line chart ga.

with insights_tab:  # Moodo tab content.
    st.subheader("Top 5 highest paid employees")  # Insight heading.
    st.dataframe(  # Top earners ni table ga chupistunnam.
        filtered_employees.nlargest(5, "Salary")[["EmployeeID", "Department", "City", "Salary", "Rating"]],  # Salary base ga top 5 rows and important columns.
        width="stretch",  # Full container width lo table (kotha API; purathanam width="stretch" deprecated).
        hide_index=True,  # Index column ni hide chestunnam.
    )
    st.subheader("Department summary")  # Rendo insight heading.
    department_summary = filtered_employees.groupby("Department").agg(  # Prathi department ki summary statistics calculate chestunnam.
        Employees=("EmployeeID", "count"),  # Aa department lo employees count.
        AvgSalary=("Salary", "mean"),  # Aa department average salary.
        AvgRating=("Rating", "mean"),  # Aa department average rating.
    ).round(2)  # Values ni 2 decimals ki round chestunnam.
    st.dataframe(department_summary, width="stretch")  # Summary table ni display chestunnam.
    with st.expander("Ee app ela pani chestundo chudali?"):  # Explanation ni expander lopala pedutunnam.
        st.markdown(  # Markdown format lo explanation raastunnam.
            "1. Sidebar lo filter marchagane **full script rerun** avutundi.\n"  # Modati step explanation.
            "2. `@st.cache_data` valla data malli generate avvadhu - andhuke fast ga untundi.\n"  # Rendo step explanation.
            "3. Filtered DataFrame nundi metrics, charts, tables anni recalculate avtayi."  # Moodo step explanation.
        )
````

## 13. Example of ML App With Streamlit Web App (Course Video 81)

**Detailed explanation:** Ippudu asalu **Streamlit enduku famous ayindo** aa use case - oka **ML model ni web app ga** convert cheyyadam. Idi course lo unna *Example Of ML App With Streamlit Web App* video ki matching example.

**Ee app enti chestundi:** Classic **Iris flower classification**. User sidebar lo 4 measurements (sepal length, sepal width, petal length, petal width) ni sliders tho set chestadu, and app **Random Forest** model tho aa flower ye species (Setosa / Versicolor / Virginica) o predict chestundi.

**ML app architecture (idi ye ML app ki ayina same):**

```
[User inputs (sliders/textboxes)]
        |
        v
[Input ni model expect chese SHAPE ki convert cheyyadam (DataFrame / numpy array)]
        |
        v
[Cached model.predict(input) and model.predict_proba(input)]
        |
        v
[Prediction + confidence ni display cheyyadam]
```

**Chala important 3 rules:**

1. **Model ni `@st.cache_resource` tho load cheyyali.** Lekapothe user prathi sari slider move chesinapudu model malli train avutundi (or malli disk nundi load avutundi) - app chala slow avutundi.

2. **Input shape model expect chesina laage undali.** Model 4 features tho train ayithe, prediction ki kuda **ade 4 features, ade order lo** ivvali. Column names kuda match avvali (sklearn DataFrame tho train chesthe).

3. **Just prediction chaalu kaadu - confidence kuda chupinchali.** `predict_proba()` tho prathi class ki probability vastundi. User ki *idi Setosa* ani cheppadam kanna, *95% confidence tho Setosa* ani cheppadam chala better.

**Production lo model ni ela handle cheyyali:** Ee example lo simplicity kosam app lopale model train chestunnam. Real project lo model ni **mundhe train chesi**, file ga save chesi, app lo just load cheyyali:

```python
# Training script lo (oka sari matrame run avutundi)
import joblib
joblib.dump(trained_model, "iris_model.pkl")

# Streamlit app lo
@st.cache_resource
def load_model():
    return joblib.load("iris_model.pkl")
```

**Kid-friendly analogy:** Model ante oka **experienced doctor** laantidi. `@st.cache_resource` ante aa doctor ni hospital lo **permanent ga** kurchobettadam - prathi patient ki kotha doctor ni hire cheyyadam (malli train cheyyadam) waste kadha?

Kinda unna cell lo modatiga model logic ni **notebook lo ne** run chesi chuddam (output tho), taruvatha ade logic ni Streamlit app ga raddam.

````python
from sklearn.datasets import load_iris  # Iris dataset ni load cheyyadaniki sklearn nundi function ni import chestunnam.
from sklearn.ensemble import RandomForestClassifier  # Classification kosam Random Forest model ni import chestunnam.
from sklearn.model_selection import train_test_split  # Data ni train/test ga split cheyyadaniki import chestunnam.
from sklearn.metrics import accuracy_score  # Model accuracy ni measure cheyyadaniki import chestunnam.
import pandas as pd  # Data ni DataFrame ga handle cheyyadaniki pandas import chestunnam.

iris_dataset = load_iris()  # Iris dataset ni load chestunnam - 150 flowers, 4 measurements, 3 species.

features_data = pd.DataFrame(iris_dataset.data, columns=iris_dataset.feature_names)  # Measurements ni DataFrame ga marustunnam.
target_labels = iris_dataset.target  # Prathi flower ki matching species number (0, 1, or 2).

print("Features shape:", features_data.shape)  # Enni rows, enni columns unnayo print chestunnam.
print("Feature names:", iris_dataset.feature_names)  # Naalugu measurement column names ni print chestunnam.
print("Target names:", list(iris_dataset.target_names))  # Moodu species names ni print chestunnam.
print()  # Khaali line - output ni chadavadaniki easy ga.
print(features_data.head())  # Modati 5 rows data ni print chestunnam.

train_features, test_features, train_labels, test_labels = train_test_split(  # Data ni train and test ga divide chestunnam.
    features_data,  # Input features (4 measurements).
    target_labels,  # Output labels (species).
    test_size=0.2,  # 20% data ni testing ki pakkana pedutunnam.
    random_state=42,  # Prathi sari same split raavadaniki seed fix chestunnam.
    stratify=target_labels,  # Prathi species ni train/test rendintilo proportionate ga unchutunnam.
)

iris_model = RandomForestClassifier(n_estimators=100, random_state=42)  # 100 decision trees unna Random Forest model create chestunnam.
iris_model.fit(train_features, train_labels)  # Training data tho model ni train chestunnam.

test_predictions = iris_model.predict(test_features)  # Test data meeda predictions cheyistunnam.
model_accuracy = accuracy_score(test_labels, test_predictions)  # Correct predictions percentage ni calculate chestunnam.

print()  # Khaali line separator.
print(f"Test accuracy: {model_accuracy:.2%}")  # Model accuracy ni percentage format lo print chestunnam.

sample_input = pd.DataFrame(  # Oka sample flower measurements ni model expect chese format lo create chestunnam.
    [[5.1, 3.5, 1.4, 0.2]],  # Sepal length, sepal width, petal length, petal width values.
    columns=iris_dataset.feature_names,  # Column names ni training data laage unchutunnam - idi chala important.
)

predicted_class = iris_model.predict(sample_input)[0]  # Sample input ki species number ni predict chestunnam.
predicted_probabilities = iris_model.predict_proba(sample_input)[0]  # Prathi species ki confidence probability ni teesukuntunnam.

print()  # Khaali line separator.
print("Sample input:", [5.1, 3.5, 1.4, 0.2])  # Ye input ichchamo print chestunnam.
print("Predicted species:", iris_dataset.target_names[predicted_class])  # Predict ayina species peru ni print chestunnam.
print("Confidence per class:", dict(zip(iris_dataset.target_names, predicted_probabilities.round(3))))  # Prathi species confidence ni print chestunnam.
````

````python
%%writefile streamlit_apps/09_ml_app.py
# ---------------- Architecture Diagram ----------------
# [User sidebar lo 4 sliders tho flower measurements set chestadu]
#    |
# [Inputs -> pandas DataFrame (model expect chese shape and column names tho)]
#    |
# [@st.cache_resource tho train ayina RandomForest model (oka sari matrame train avutundi)]
#    |
# [model.predict() -> species | model.predict_proba() -> confidence]
#    |
# [Prediction card + probability bar chart + dataset explorer display avtayi]
#
# ---------------- Deep Architecture Notes ----------------
# Step 1: @st.cache_resource valla model oka sari matrame train avutundi - prathi slider move ki kaadu.
# Step 2: @st.cache_data valla dataset load kuda oka sari matrame jarugutundi.
# Step 3: Input ni EXACT ga training features order and names tho DataFrame ga ivvali - lekapothe wrong predictions.
# Step 4: predict() class ni istundi; predict_proba() prathi class confidence ni istundi - rendu chupinchadam best practice.
# Step 5: Real project lo model ni mundhe train chesi joblib tho save chesi, ikkada just load cheyyali.

import streamlit as st  # Streamlit library ni import chestunnam.
import pandas as pd  # Input ni DataFrame ga marchadaniki pandas import chestunnam.
from sklearn.datasets import load_iris  # Iris dataset ni load chese function ni import chestunnam.
from sklearn.ensemble import RandomForestClassifier  # Random Forest classification model ni import chestunnam.
from sklearn.model_selection import train_test_split  # Train/test split cheyyadaniki import chestunnam.
from sklearn.metrics import accuracy_score  # Accuracy calculate cheyyadaniki import chestunnam.

st.set_page_config(  # Page level settings - modati streamlit command.
    page_title="Iris Species Predictor",  # Browser tab title.
    page_icon="🌸",  # Browser tab icon.
    layout="wide",  # Full width layout.
)


@st.cache_data  # Dataset ni oka sari matrame load cheyyadaniki data cache decorator.
def load_iris_dataset():  # Iris data ni load chese function.
    iris_bunch = load_iris()  # sklearn nundi iris dataset ni teesukuntunnam.
    feature_frame = pd.DataFrame(iris_bunch.data, columns=iris_bunch.feature_names)  # Measurements ni DataFrame ga marustunnam.
    return feature_frame, iris_bunch.target, list(iris_bunch.target_names), iris_bunch.feature_names  # Anni pieces ni return chestunnam.


@st.cache_resource  # Model oka HEAVY RESOURCE - copy cheyyakunda oke object ni share cheyyadaniki ee decorator.
def train_iris_model(feature_frame, target_values):  # Model ni train chese function.
    train_x, test_x, train_y, test_y = train_test_split(  # Data ni train and test ga split chestunnam.
        feature_frame, target_values, test_size=0.2, random_state=42, stratify=target_values  # 80/20 split, stratified.
    )
    model = RandomForestClassifier(n_estimators=100, random_state=42)  # 100 trees unna Random Forest ni create chestunnam.
    model.fit(train_x, train_y)  # Training data tho model ni train chestunnam.
    test_accuracy = accuracy_score(test_y, model.predict(test_x))  # Test data meeda accuracy ni calculate chestunnam.
    return model, test_accuracy  # Train ayina model and daani accuracy ni return chestunnam.


features_frame, target_values, species_names, feature_names = load_iris_dataset()  # Cached function nundi dataset ni teesukuntunnam.
trained_model, model_accuracy = train_iris_model(features_frame, target_values)  # Cached function nundi model ni teesukuntunnam.

st.title("🌸 Iris Flower Species Predictor")  # Page main heading.
st.write("Sidebar lo flower measurements set cheyyi - model instant ga species ni predict chestundi.")  # Chinna instruction.

with st.sidebar:  # Ee block lopala widgets sidebar lo kanipistayi.
    st.header("🔧 Flower Measurements")  # Sidebar heading.
    st.caption("Prathi slider ni move chesinapudu prediction automatic ga update avutundi.")  # Rerun behaviour ni explain chese caption.
    user_measurements = {}  # User ichchina values ni store cheyyadaniki khaali dictionary.
    for feature_name in feature_names:  # Prathi feature (4 measurements) ki loop chestunnam.
        user_measurements[feature_name] = st.slider(  # Aa feature ki oka slider create chestunnam.
            feature_name,  # Slider label - feature peru.
            min_value=float(features_frame[feature_name].min()),  # Dataset lo unna minimum value ni slider min ga pedutunnam.
            max_value=float(features_frame[feature_name].max()),  # Dataset lo unna maximum value ni slider max ga pedutunnam.
            value=float(features_frame[feature_name].mean().round(1)),  # Default ga average value ni set chestunnam.
            step=0.1,  # Slider ni 0.1 steps lo move cheyyadaniki.
        )
    st.divider()  # Sidebar lo separator line.
    st.metric("Model test accuracy", f"{model_accuracy:.1%}")  # Model accuracy ni sidebar lo metric ga chupistunnam.

input_frame = pd.DataFrame([user_measurements])  # User values ni oka single-row DataFrame ga marustunnam - model ki ide shape kavali.

predicted_class_index = trained_model.predict(input_frame)[0]  # Model tho species number ni predict chestunnam.
class_probabilities = trained_model.predict_proba(input_frame)[0]  # Prathi species ki confidence probability ni teesukuntunnam.
predicted_species_name = species_names[predicted_class_index]  # Species number ni species peru ga convert chestunnam.
prediction_confidence = class_probabilities[predicted_class_index]  # Predict chesina class ki unna confidence value.

prediction_column, probability_column = st.columns([1, 1])  # Prediction and probabilities ki rendu equal columns.

with prediction_column:  # Ednam column content.
    st.subheader("Prediction")  # Section subheading.
    st.success(f"Predicted species: **{predicted_species_name.upper()}**")  # Predicted species ni green success box lo chupistunnam.
    st.metric("Confidence", f"{prediction_confidence:.1%}")  # Confidence ni percentage metric ga chupistunnam.
    st.write("Nuvvu ichchina measurements:")  # Input values ki oka label.
    st.dataframe(input_frame, width="stretch", hide_index=True)  # User inputs ni table ga chupistunnam.

with probability_column:  # Kudi column content.
    st.subheader("Confidence per species")  # Section subheading.
    probability_frame = pd.DataFrame({  # Prathi species probability ni DataFrame ga marustunnam - chart kosam.
        "Species": species_names,  # Moodu species names column.
        "Probability": class_probabilities,  # Vaati matching probabilities column.
    }).set_index("Species")  # Species ni index ga pedutunnam - bar chart X-axis ki.
    st.bar_chart(probability_frame)  # Probabilities ni bar chart ga display chestunnam.

st.divider()  # Section separator line.

feature_tab, data_tab, help_tab = st.tabs(["🌟 Feature Importance", "📊 Training Data", "❓ How it works"])  # Moodu tabs create chestunnam.

with feature_tab:  # Modati tab content.
    st.write("Model prediction cheyyadaniki ye measurement entha important o:")  # Feature importance ni explain chese line.
    importance_frame = pd.DataFrame({  # Feature importance values ni DataFrame ga marustunnam.
        "Feature": feature_names,  # Naalugu feature names.
        "Importance": trained_model.feature_importances_,  # Model nerchukunna importance scores.
    }).set_index("Feature").sort_values("Importance", ascending=False)  # Importance base ga sort chestunnam.
    st.bar_chart(importance_frame)  # Importance ni bar chart ga display chestunnam.

with data_tab:  # Rendo tab content.
    display_frame = features_frame.copy()  # Original data ni marchakunda undadaniki copy teesukuntunnam.
    display_frame["Species"] = [species_names[label] for label in target_values]  # Species peru column ni add chestunnam.
    selected_species = st.multiselect(  # Ye species chudalo select chese widget.
        "Species filter:",  # Widget label.
        options=species_names,  # Moodu species options.
        default=species_names,  # Default ga anni species selected.
    )
    st.dataframe(  # Filtered training data ni table ga chupistunnam.
        display_frame[display_frame["Species"].isin(selected_species)],  # Select chesina species rows ni matrame chupistunnam.
        width="stretch",  # Full width table.
        hide_index=True,  # Index column ni hide chestunnam.
    )

with help_tab:  # Moodo tab content.
    st.markdown(  # App ela pani chestundo markdown lo explain chestunnam.
        "1. **Sliders** nundi 4 measurements collect avtayi.\n"  # Modati step.
        "2. Avi oka **single row DataFrame** ga marutayi (model expect chese shape).\n"  # Rendo step.
        "3. **RandomForest** model `predict()` tho species ni, `predict_proba()` tho confidence ni istundi.\n"  # Moodo step.
        "4. Model `@st.cache_resource` valla **oka sari matrame** train avutundi - slider move chesina malli train avvadhu.\n"  # Naalgo step.
        "5. Prathi slider change **full script rerun** ni trigger chestundi, kani cached parts malli run avvavu."  # Aido step.
    )
````

## 16. Common Mistakes and Gotchas

| # | Mistake | Enti avutundi | Correct way |
|---|---|---|---|
| 1 | `python app.py` tho run cheyyadam | UI raadhu, *missing ScriptRunContext* warnings | `streamlit run app.py` |
| 2 | `st.set_page_config()` ni middle lo pettadam | `StreamlitSetPageConfigMustBeFirstCommandError` | Script lo **modati** streamlit command ga pettali |
| 3 | Normal variable tho counter maintain cheyyadam | Prathi rerun ki reset avutundi | `st.session_state` vaadali |
| 4 | Prathi rerun ki model train cheyyadam | App chala slow avutundi | `@st.cache_resource` vaadali |
| 5 | Rendu widgets ki same label | `DuplicateWidgetID` error | `key="unique_name"` ivvali |
| 6 | `if st.button(...)` lopala important state pettadam | Next rerun lo maayam avutundi | Button lopala `st.session_state` ni update cheyyali |
| 7 | `st.file_uploader()` result ni None check cheyyakunda vaadadam | `AttributeError` vastundi | `if uploaded_file is not None:` check cheyyali |
| 8 | Cached function lopala `st.slider()` laanti widgets raayadam | Weird caching behaviour, warnings | Cached functions lo **pure computation** matrame undali |
| 9 | Peddha DataFrame (lakshala rows) ni direct ga display cheyyadam | Browser hang avutundi | `df.head(1000)` laaga limit cheyyali |
| 10 | API keys ni code lo hardcode cheyyadam | GitHub lo key leak avutundi | `st.secrets` vaadali, `.streamlit/secrets.toml` ni gitignore lo pettali |
| 11 | Prediction input ni wrong column order tho ivvadam | Model wrong prediction istundi (error raadhu!) | Training features **exact order and names** tho DataFrame ivvali |
| 12 | Session state ni database laaga vaadadam | Tab close chesthe data poddhu | Real DB (SQLite/Postgres) or file lo save cheyyali |

**Debugging tips:**

- Error vasthe browser lo full **traceback** kanipistundi - adi terminal lo kuda vastundi.
- `st.write(variable)` ni print statement laaga vaadi values ni check cheyyachu.
- `st.session_state` motham chudalante: `st.write(st.session_state)`.
- Cache valla weird behaviour anipisthe: `streamlit cache clear` run cheyyi.

## Final Summary - 8 Points

1. **Streamlit ante** pure Python tho web apps build chese library - HTML/CSS/JS avasaram ledu.
2. **`streamlit run app.py`** tho matrame run cheyyali - `python app.py` tho kaadu.
3. **Rerun model:** prathi interaction ki full script top-to-bottom malli run avutundi. Ee oka point Streamlit lo asalu core.

6. **Caching** (`@st.cache_data` data ki, `@st.cache_resource` models ki) valla app fast ga untundi.
7. **ML app pattern:** cached model + user inputs ni correct shape lo pettadam + `predict()` and `predict_proba()` ni display cheyyadam.
8. **Deployment:** GitHub + `requirements.txt` + Streamlit Community Cloud - konni nimishalalo app live avutundi.
