## 1. What is Flask Framework?

![Flask WSGI Jinja2 Definition](assets/flask-wsgi-jinja2-definition.png)

**Definition:** Flask ante **Web framework which is created with the Python Programming Language**. Idi tho manam websites, APIs, and web applications ni Python code tho build cheyyachu.

**Detailed explanation:** Normal ga oka website build cheyyalante, HTTP requests handle cheyyadam, routing (which URL goes to which function), HTML rendering, form data process cheyyadam - anni manual ga low-level code raayali. Flask ee anni common tasks ki ready-made tools istundi, so manam business logic meeda easy ga focus pettavachu.

**Two most important building blocks:**

1. **WSGI (Web Server Gateway Interface)** -> Protocol/standard, idi Web Server and Web App (Flask) madhya communication ni define chestundi.
2. **Jinja2 Template Engine** -> Dynamic HTML pages generate cheyyadaniki Flask use chese templating engine.

**Kid-friendly analogy:** Flask ante oka **ready-made toy kitchen set** laantidi anukondi. Nuvvu food (website logic) cook cheyyalante, prathi tool (stove, knife, plate) ni scratch nundi build cheyyalsina avasaram ledu - kitchen set (Flask) already anni tools tho vastundi; nuvvu direct ga cooking (business logic) start cheyyachu.

**Why Flask popular?** Simple, lightweight, flexible, and beginner-friendly kabatti chala mandi web developers Flask tho start chestaru.

## 2. WSGI - Web Server Gateway Interface

**Detailed explanation:** WSGI ante oka **protocol/standard** - idi **Web Server** (Example: Apache, IIS, or cloud hosting laanti AWS EC2, Azure App) and **Web App** (Example: manam Flask tho raasina Python code) madhya "common language" laaga pani chestundi.

Web Server ki Python code ela run cheyyalo direct ga teliyadu, and Web App (Flask) ki raw HTTP requests ela handle cheyyalo teliyadu. WSGI madhyalo oka **bridge/translator** laaga pani chesi, requests ni Web Server nundi Web App ki, and responses ni Web App nundi Web Server ki pass chestundi.

**Kid-friendly analogy:** Nuvvu (Web Server) oka foreign country nundi vachina guest (Client request) tho matladali, kani neeku vaari language raadu. Web App (Flask) ki neeku kavalsina answer telusu, kani daaniki kuda guest language raadu. Madhyalo oka **translator (WSGI)** unte, guest question ni Flask ki artham ayye language lo cheppi, Flask answer ni malli guest ki artham ayye language lo cheptadu.

**Simple meaning:** WSGI valla, different web servers (Apache, IIS, Gunicorn, etc.) meeda oke Flask code ni run cheyyachu - Flask code marchalsina avasaram ledu, WSGI translator laaga help chestundi.

## 3. Web Server vs Web App Architecture

![Web Server And Web App Architecture](assets/flask-web-server-web-app-architecture.png)

**Detailed explanation of the diagram:**

1. **Multiple Requests:** Client browsers nundi different requests (Req1 -> /home, Req2, Req3) anni **Web Server** ki vellutayi.
2. **Web Server:** Ee requests ni receive chesi handle chestundi. Idi different hosting environments meeda run avvachu:

   - **Apache** -> Popular open-source web server software.
   - **IIS (Internet Information Services)** -> Microsoft's web server software (mostly Windows).
3. **WSGI Protocol:** Web Server, Web App madhya bidirectional communication (yellow arrows lo chupinchinattu - request oka direction lo, response inko direction lo) WSGI protocol dwara jarugutundi.
4. **Web App:** Idi manam raasina actual Flask code; ikkada routes define chestam:
   - /home -> Home page route.
   - /API -> API endpoints route.
   - /functionalities -> Business logic/features route.
5. **Flask Framework:** Web App ni build cheyyadaniki manam **Flask Framework** ni use chestunnam (diagram lo red arrow tho chupinchinattu).

**Kid-friendly analogy:** Idi oka **restaurant** laantidi anukondi. Customers (Req1, Req2, Req3) different tables nundi orders pampistaru. **Waiter (Web Server)** orders ni teesukuni kitchen (Web App) ki pass chestadu. Waiter and kitchen madhya oka fixed communication style (WSGI) untundi, so order correct ga ardham avutundi. Kitchen (Web App) different sections (/home = starters, /API = main course, /functionalities = desserts) ga divide ayi undachu, and ee mottham kitchen setup **Flask Framework** (kitchen equipment brand) tho build ayindi.

**Simple meaning:** Web Server "traffic handle cheyye person", Web App "actual logic run cheyye code", WSGI "rendinti madhya common language".

## 4. Jinja2 Template Engine Overview

**Detailed explanation:** Jinja2 ante Flask default ga use chese **templating engine**. Idi tho manam HTML files lo Python variables, loops, conditions ni embed cheyyachu, so static HTML kaakunda **dynamic HTML** (data based ga change ayye pages) generate cheyyachu.

Jinja2 syntax:

- `{{ variable_name }}` -> Python variable value ni HTML lo print cheyyadaniki.
- `{% if condition %} ... {% endif %}` -> Conditional logic HTML lo add cheyyadaniki.
- `{% for item in items %} ... {% endfor %}` -> Loop HTML lo add cheyyadaniki (Example: list of products display cheyyadam).

**Kid-friendly analogy:** Jinja2 ante oka **fill-in-the-blanks greeting card** laantidi anukondi. Card meeda "Dear ___, Happy Birthday! You are turning ___ years old" ani print chesi unte, prathi friend ki vere vere name and age fill chesi ivvochu - card design (HTML) same ga untundi, kani blanks (`{{ name }}`, `{{ age }}`) lo values change avtayi.

**Why important?** Ee section tarvatha vachche code examples anni ("Integrating HTML", "Building Dynamic URL") ee Jinja2 concept meeda based ayi untayi.

## 5. Introduction to Flask Framework (Installation and First App)

**Installation:** Flask ni install cheyyadaniki terminal lo `pip install flask` run cheyyali.

**First app structure:** Flask app create cheyyalante, minimum ga:

1. `Flask` class ni import cheyyali.
2. `Flask(__name__)` tho oka app object create cheyyali.
3. `@app.route("/path")` decorator tho route define cheyyali.
4. Aa route ki matching function raayali, adi response return chestundi.
5. `app.run()` tho server ni start cheyyali (real project lo).

**Note:** Ee notebook lo `app.run()` use cheyyakunda, Flask's built-in **test client** (`app.test_client()`) use chestunnam - idi real server start cheyyakunda, requests ni simulate chesi test cheyyadaniki use chestaru (production/testing best practice).

````python
from flask import Flask  # Flask class ni import chestunnam; ide manam app build cheyyadaniki core class.

app = Flask(__name__)  # Flask application object create chestunnam; __name__ tho current module ni identify chestundi.

@app.route("/home")  # "/home" URL ki ee function ni map chestunnam.
def home_page():  # Home route ki matching view function define chestunnam.
    return "Welcome to Flask Home Page"  # Simple text response return chestunnam.

test_client = app.test_client()  # Real server start cheyyakunda requests test cheyyadaniki test client create chestunnam.
response = test_client.get("/home")  # "/home" route ki GET request pampistunnam.

print(f"Status code: {response.status_code}")  # HTTP status code print chestunnam (200 = success).
print(f"Response body: {response.get_data(as_text=True)}")  # Response content ni text ga print chestunnam.
````

### First Flask app code line-by-line explanation

1. `Flask(__name__)` -> Flask ki current file/module details ni pass chestundi, so templates and static files ni correct ga locate cheyyagalguthundi.
2. `@app.route("/home")` -> "idi /home URL ki matching route" ani Flask ki cheptundi (decorator pattern).
3. `def home_page():` -> Ee function ni Flask automatic ga call chestundi, prathi sari evaraina /home ki request pampinapudu.
4. `return "..."` -> Ee string ne response ga client ki pampistundi.
5. `app.test_client()` -> Real network server lekunda, in-memory ga requests ni simulate cheyyadaniki special testing object.
6. `test_client.get("/home")` -> /home route ki GET HTTP request pampistundi, and response ni return chestundi.
7. `response.status_code` -> 200 vaste request successful ani ardham.

**Simple meaning:** Idi Flask app ki "Hello World" laanti first step - oka route, oka function, oka response.

## 6. Understanding Simple Flask App Skeleton

**Detailed explanation:** Real-world Flask project lo, konni standard folders and files untayi:

```text
my_flask_project/
|-- app.py              # Main application file (routes, logic)
|-- templates/           # HTML files ikkada untayi (Jinja2 use chestundi)
|   |-- home.html
|   |-- greet.html
|-- static/              # CSS, JavaScript, images ikkada untayi
|   |-- style.css
|-- requirements.txt     # Project ki kavalsina packages list
```

- **`app.py`** -> Flask object create chesi, routes and functions define chese main file.
- **`templates/`** -> `render_template()` function ee folder nundi HTML files ni automatic ga load chestundi.
- **`static/`** -> CSS, JS, images laanti files ikkada pedatamu; browser lo `/static/filename` tho access cheyyachu.
- **`requirements.txt`** -> `pip install -r requirements.txt` tho anni dependencies okesari install cheyyadaniki.

**Kid-friendly analogy:** Idi oka **school bag** laantidi anukondi. `app.py` ante nee **timetable** (ee period emi cheyyalo cheptundi). `templates/` folder ante nee **notebooks** (emi raayalo, chupinchalo). `static/` folder ante nee **pencil box** (colors, tools). Anni kaligi oka organized bag (Flask project) create chestayi.

````python
from flask import Flask  # Flask class import chestunnam.

app = Flask(__name__)  # App object create chestunnam; templates/ and static/ folders automatic ga detect avtayi.

@app.route("/")  # Root URL ("/") ki route define chestunnam.
def index_page():  # Root route ki matching function.
    return "This is the root route of our Flask skeleton"  # Simple response return chestunnam.

@app.route("/about")  # "/about" URL ki inko route add chestunnam.
def about_page():  # About route ki matching function.
    return "This is the about page of our Flask skeleton"  # About page response return chestunnam.

test_client = app.test_client()  # Test client create chestunnam.

root_response = test_client.get("/")  # Root route ni test chestunnam.
about_response = test_client.get("/about")  # About route ni test chestunnam.

print(f"Root route status: {root_response.status_code}, body: {root_response.get_data(as_text=True)}")  # Root route result print chestunnam.
print(f"About route status: {about_response.status_code}, body: {about_response.get_data(as_text=True)}")  # About route result print chestunnam.
````

### App skeleton code line-by-line explanation

1. Same Flask app lo multiple routes (`/` and `/about`) define cheyyachu - prathi route ki separate function.
2. Route names URL structure ni represent chestayi; `/about` ante "About Us" page laantidi anukovachu.
3. Test client tho rendu routes ni verify chestunnam, real server run cheyyakunda.
4. Real project lo, ee functions `return "text"` kaakunda `render_template("home.html")` laanti HTML files ni return chestayi (next section lo chuddam).

**Simple meaning:** Flask app skeleton ante - app object + routes + functions; ivi anni kalisi oka basic working application ni build chestayi.

## 7. Integrating HTML with Flask Web App

**Detailed explanation:** Manam text return cheyyadam kaakunda, real websites lo **HTML pages** return cheyyali. Flask ki idi `render_template()` function tho cheyyachu:

1. HTML file ni `templates/` folder lo pedatamu.
2. `render_template("filename.html", variable=value)` call chesthe, Flask aa HTML file ni load chesi, Jinja2 tho variables ni fill chesi, final HTML ni return chestundi.

Ee example kosam manam already `templates/home.html` and `templates/greet.html` files create chesi unnam:

- `home.html` -> Static welcome message.
- `greet.html` -> `{{ user_name }}` and `{{ course_name }}` ane Jinja2 variables tho dynamic content.

**Kid-friendly analogy:** `render_template()` ante oka **printing press** laantidi anukondi. Nuvvu design (HTML template) oka sari create chestavu, kani prathi print (request) ki different name/details (variables) fill chesi print cheyyachu - design mare avasaram ledu, content matrame maruthundi.

````python
from flask import Flask, render_template  # Flask and render_template ni import chestunnam.

app = Flask(__name__)  # App object create chestunnam; default ga "templates" folder ni automatic ga chustundi.

@app.route("/home")  # "/home" route ki static HTML page render chestunnam.
def home_page():  # Home page function.
    return render_template("home.html")  # templates/home.html file ni load chesi HTML ga return chestunnam.

@app.route("/greet")  # "/greet" route ki dynamic HTML page render chestunnam.
def greet_page():  # Greet page function.
    return render_template(
        "greet.html",  # templates/greet.html file ni load chestunnam.
        user_name="Sai",  # {{ user_name }} placeholder ki "Sai" value pass chestunnam.
        course_name="Python and Flask",  # {{ course_name }} placeholder ki course name pass chestunnam.
    )

test_client = app.test_client()  # Test client create chestunnam.

home_response = test_client.get("/home")  # "/home" route ni test chestunnam.
greet_response = test_client.get("/greet")  # "/greet" route ni test chestunnam.

print("---- Home Page HTML ----")  # Home page output ni identify cheyyadaniki heading print chestunnam.
print(home_response.get_data(as_text=True))  # Rendered home.html content print chestunnam.

print("\n---- Greet Page HTML ----")  # Greet page output ni identify cheyyadaniki heading print chestunnam.
print(greet_response.get_data(as_text=True))  # Rendered greet.html content (with filled variables) print chestunnam.
````

### HTML integration code line-by-line explanation

1. `render_template("home.html")` -> `templates/home.html` file ni as-is load chesi return chestundi (no dynamic values needed).
2. `render_template("greet.html", user_name="Sai", course_name="Python and Flask")` -> `greet.html` file lo unna `{{ user_name }}` and `{{ course_name }}` placeholders ni ee values tho replace chestundi.
3. Prathi request ki different `user_name` pass chesthe, same template different content tho render avtundi.
4. Output console lo full rendered HTML ga kanipistundi, real browser lo idi formatted page ga kanipistundi.

**Simple meaning:** `render_template()` valla, HTML design and Python logic ni separate ga maintain cheyyachu - templates folder lo design, `app.py` lo logic.

## 8. Working with HTTP Verbs GET and POST

**Detailed explanation:** HTTP requests different "verbs" (methods) tho pampavachu:

- **GET** -> Server nundi data **fetch** (retrieve) cheyyadaniki use chestam (Example: page open cheyyadam).
- **POST** -> Server ki new data **submit** cheyyadaniki use chestam (Example: form fill chesi submit cheyyadam).

Flask lo, `@app.route("/path", methods=["GET", "POST"])` tho oka route ni rendu verbs ki kuda handle cheyyachu, and `request.method` tho current request e verb tho vachindo check cheyyachu.

**Kid-friendly analogy:** **GET** ante library nundi oka book ni **chadavadaniki teesukovadam** (data fetch, kani book content change avvadu). **POST** ante library ki nee own oka kotha book ni **donate cheyyadam** (new data add cheyyadam/submit cheyyadam).

````python
from flask import Flask, request  # Flask and request object ni import chestunnam; request tho incoming data ni access chestam.

app = Flask(__name__)  # App object create chestunnam.

@app.route("/submit", methods=["GET", "POST"])  # "/submit" route ni GET and POST rendintiki kuda handle cheyyadaniki configure chestunnam.
def submit_form():  # Form submission ni handle chese function.
    if request.method == "POST":  # Current request POST aa ani check chestunnam.
        submitted_name = request.form.get("student_name", "Unknown")  # Form data nundi "student_name" field ni fetch chestunnam.
        return f"POST received. Student name: {submitted_name}"  # Submitted data tho response return chestunnam.
    return "This is a GET request. Please submit the form using POST."  # GET request ki simple message return chestunnam.

test_client = app.test_client()  # Test client create chestunnam.

get_response = test_client.get("/submit")  # GET request ni simulate chestunnam.
post_response = test_client.post("/submit", data={"student_name": "Sai"})  # POST request tho form data pampistunnam.

print(f"GET response: {get_response.get_data(as_text=True)}")  # GET response print chestunnam.
print(f"POST response: {post_response.get_data(as_text=True)}")  # POST response print chestunnam.
````

### GET and POST code line-by-line explanation

1. `methods=["GET", "POST"]` -> Ee route rendu verbs (GET, POST) ni accept chestundi ani Flask ki cheptundi.
2. `request.method` -> Current incoming request e HTTP verb tho vachindo chupistundi (`"GET"` or `"POST"`).
3. `request.form.get("student_name", "Unknown")` -> POST request tho vachina form data nundi `"student_name"` field value ni fetch chestundi; field lekapothe default ga `"Unknown"` istundi.
4. `test_client.get("/submit")` -> GET request simulate chestundi -> `request.method` `"GET"` avtundi -> else branch run avtundi.
5. `test_client.post("/submit", data={...})` -> POST request tho form data pampistundi -> if branch run avtundi.

**Simple meaning:** Same route, different HTTP verb batti different behavior ivvachu - idi web forms ki chala common pattern.

## 9. Building Dynamic URL, Variables Rule and Jinja2 Template Engine

**Detailed explanation:** "Dynamic URL" ante URL lo konni parts fixed kaakunda, **variable** ga undadam. Example: `/student/1`, `/student/2` - ikkada `1`, `2` anevi dynamic values.

Flask lo ee dynamic parts ni **URL converters** tho define chestamu:

- `<int:student_id>` -> Integer values matrame accept chestundi.
- `<string:student_name>` -> Text values accept chestundi.

Ee dynamic values ni function parameters ga receive chesi, Jinja2 template ki pass cheyyachu, so page content URL batti change avutundi.

**Kid-friendly analogy:** Idi oka **mailbox with a number** laantidi anukondi. Mailbox address (`/student/<id>`) same structure lo untundi, kani number (`1`, `2`, `3`) batti different person ki letter (response) veltundi - same route "template", different data prathi request ki.

````python
from flask import Flask, render_template  # Flask and render_template ni import chestunnam.

app = Flask(__name__)  # App object create chestunnam.

students_data = {  # Student id ni key ga, student details ni value ga store chese sample data.
    1: "Sai",  # Student id 1 ki matching name.
    2: "Kumar",  # Student id 2 ki matching name.
}

@app.route("/student/<int:student_id>")  # "<int:student_id>" ane dynamic URL segment ni define chestunnam.
def student_details(student_id):  # URL nundi vachina student_id ni function parameter ga receive chestunnam.
    student_name = students_data.get(student_id, "Unknown Student")  # Matching student name ni fetch chestunnam.
    return render_template(  # Jinja2 template ki dynamic values pass chesi render chestunnam.
        "greet.html",  # Existing greet.html template ni reuse chestunnam.
        user_name=student_name,  # URL nundi vachina dynamic student name.
        course_name=f"Details for student ID {student_id}",  # Dynamic course_name field lo ID ni kuda chupistunnam.
    )

test_client = app.test_client()  # Test client create chestunnam.

response_student_one = test_client.get("/student/1")  # Student id 1 kosam dynamic route ni test chestunnam.
response_student_two = test_client.get("/student/2")  # Student id 2 kosam dynamic route ni test chestunnam.

print("---- /student/1 ----")  # First response ni identify cheyyadaniki heading print chestunnam.
print(response_student_one.get_data(as_text=True))  # Student 1 details print chestunnam.

print("\n---- /student/2 ----")  # Second response ni identify cheyyadaniki heading print chestunnam.
print(response_student_two.get_data(as_text=True))  # Student 2 details print chestunnam.
````

### Dynamic URL code line-by-line explanation

1. `<int:student_id>` -> Ee URL segment ni Flask automatic ga integer ga convert chesi, `student_id` parameter ki pass chestundi.
2. `students_data.get(student_id, "Unknown Student")` -> Dictionary nundi matching student name fetch chestundi; dorakakapothe default value istundi.
3. Same function, same template (`greet.html`), kani URL batti (`/student/1` vs `/student/2`) different data render avtundi.
4. Idi real-world lo profile pages, product pages, order details laanti features ki base pattern.

**Simple meaning:** Dynamic URLs + Jinja2 combine ayithe, oke code tho velaadi different pages (prathi student/product/order ki) generate cheyyachu.

## 10. Working with REST API's and HTTP Verbs PUT and DELETE

**Detailed explanation:** REST API ante, HTTP verbs ni use chesi data meeda **CRUD** operations (Create, Read, Update, Delete) cheyyadaniki oka standard style:

- **GET** -> Data ni **Read** (fetch) cheyyadaniki.
- **POST** -> Kotha data ni **Create** cheyyadaniki.
- **PUT** -> Existing data ni **Update** (modify) cheyyadaniki.
- **DELETE** -> Data ni **Delete** (remove) cheyyadaniki.

`jsonify()` function tho Python dictionary ni JSON response ga convert chesi client ki pampistamu - REST APIs mostly JSON format lone data exchange chestayi.

**Kid-friendly analogy:** Idi oka **toy inventory register** laantidi anukondi. **GET** ante register chusi list cheppadam ("ee toys unnayi"). **POST** ante kotha toy ni register lo add cheyyadam. **PUT** ante already unna toy details ni **update** cheyyadam (Example: quantity marchadam). **DELETE** ante register nundi oka toy entry ni **teesestundi**.

````python
from flask import Flask, jsonify, request  # Flask, jsonify (JSON response kosam), request (incoming data kosam) import chestunnam.

app = Flask(__name__)  # App object create chestunnam.

toy_inventory = {  # In-memory "database" laaga pani chese simple dictionary.
    1: {"toy_name": "Robot", "quantity": 5},  # Toy id 1 details.
    2: {"toy_name": "Car", "quantity": 10},  # Toy id 2 details.
}

@app.route("/toys", methods=["GET"])  # GET verb tho anni toys ni list cheyyadaniki route.
def get_all_toys():  # Anni toys ni return chese function.
    return jsonify(toy_inventory)  # Full inventory dictionary ni JSON ga return chestunnam.

@app.route("/toys/<int:toy_id>", methods=["PUT"])  # PUT verb tho existing toy ni update cheyyadaniki route.
def update_toy(toy_id):  # Existing toy details ni update chese function.
    if toy_id not in toy_inventory:  # Toy id exist ayye leda ani check chestunnam.
        return jsonify({"error": "Toy not found"}), 404  # Toy dorakakapothe 404 error return chestunnam.
    updated_quantity = request.json.get("quantity")  # Request body nundi kotha quantity value fetch chestunnam.
    toy_inventory[toy_id]["quantity"] = updated_quantity  # Existing toy quantity ni update chestunnam.
    return jsonify(toy_inventory[toy_id])  # Updated toy details ni return chestunnam.

@app.route("/toys/<int:toy_id>", methods=["DELETE"])  # DELETE verb tho toy ni remove cheyyadaniki route.
def delete_toy(toy_id):  # Toy entry ni delete chese function.
    if toy_id not in toy_inventory:  # Toy id exist ayye leda ani check chestunnam.
        return jsonify({"error": "Toy not found"}), 404  # Toy dorakakapothe 404 error return chestunnam.
    removed_toy = toy_inventory.pop(toy_id)  # Toy entry ni dictionary nundi teesesi, teesesina details ni store chestunnam.
    return jsonify({"message": "Toy deleted", "deleted_toy": removed_toy})  # Delete confirmation ni return chestunnam.

test_client = app.test_client()  # Test client create chestunnam.

get_response = test_client.get("/toys")  # Anni toys ni fetch chesi test chestunnam.
print(f"GET /toys -> {get_response.get_json()}")  # Initial inventory print chestunnam.

put_response = test_client.put("/toys/1", json={"quantity": 20})  # Toy id 1 quantity ni 20 ki update chestunnam.
print(f"PUT /toys/1 -> {put_response.get_json()}")  # Updated toy details print chestunnam.

delete_response = test_client.delete("/toys/2")  # Toy id 2 ni delete chestunnam.
print(f"DELETE /toys/2 -> {delete_response.get_json()}")  # Delete confirmation print chestunnam.

final_response = test_client.get("/toys")  # Final inventory ni malli fetch chestunnam.
print(f"GET /toys (after changes) -> {final_response.get_json()}")  # Updated final inventory print chestunnam.
````

### REST API code line-by-line explanation

1. `jsonify(toy_inventory)` -> Python dictionary ni proper JSON HTTP response ga convert chestundi.
2. `methods=["PUT"]` -> Ee route matrame PUT requests ni accept chestundi.
3. `request.json.get("quantity")` -> Client pampina JSON body nundi `"quantity"` value ni fetch chestundi.
4. `toy_inventory.pop(toy_id)` -> Dictionary nundi entry ni teesesi, teesesina value ni return chestundi (DELETE operation).
5. `test_client.put(...)` and `test_client.delete(...)` -> PUT and DELETE HTTP requests ni simulate chestayi, real server lekunda.
6. Final `GET /toys` call tho, PUT (update) and DELETE (remove) changes reflect ayinatlu confirm chestunnam.

**Simple meaning:** REST API ante - resources (toys, students, products) meeda GET/POST/PUT/DELETE verbs tho standard operations cheyyadam.

- **WSGI:** Web Server mariyu Web App madhya communication protocol.
- **Web Server:** Requests handle chese layer (AWS EC2, Azure App, Apache, IIS laanti hosting environments meeda run avtundi).
- **Web App:** Manam raasina actual Flask logic; routes (`/home`, `/api`, `/functionalities`) ikkada define chestamu.
- **Jinja2:** Dynamic HTML generate cheyyadaniki Flask default templating engine.
- **Flask App Skeleton:** `app.py` + `templates/` + `static/` combination.
- **HTML Integration:** `render_template()` tho HTML files ni serve cheyyadam.
- **GET vs POST:** GET data fetch cheyyadaniki, POST data submit cheyyadaniki.
- **Dynamic URL:** `<int:id>` laanti converters tho URL lo variable parts define cheyyadam.
- **REST API (PUT/DELETE):** Existing data update (PUT) mariyu delete (DELETE) cheyyadaniki HTTP verbs.
