## 1. Class and Object

**Concept:** Class ante plan/blueprint. Object ante aa class nundi create ayina actual item.

**Why use it?** Same structure unna multiple objects create cheyyadaniki class use chestam. Example: many students ki same fields name, age, course untayi.

### `self` ante enti?

`self` ante **current object reference**. Class nundi object create ayyaka, aa object own data ni access cheyyadaniki Python lo `self` use chestam.

Example: `student_one = Student("Sai", 24, "Python")` ani create chesthe, `self` ante ikkada `student_one` object ni point chestundi.

`self.name = name` ante external ga vachina `name` value ni current object lo `name` attribute ga save cheyyadam.

**Important:** `self` keyword fixed kaadu, but Python convention. Manam vere name use cheyyachu, kani industry lo almost always `self` use chestaru.

### `def __init__(...)` ante enti?

`__init__` ante Python lo **constructor method**. Object create ayye time lo automatic ga run avtundi.

`def __init__(self, name, age, course):` ante Student object create chesinappudu `name`, `age`, `course` values receive chesi aa object lo save cheyyadaniki method.

Example: `Student("Sai", 24, "Python")` ani call chesthe Python automatic ga `__init__` ni call chestundi. Appudu `name = "Sai"`, `age = 24`, `course = "Python"` avtayi.

`__init__` lo double underscores untayi kabatti idi special/magic method. Manam direct ga call cheyyalsina avasaram ledu; object create chesinappude Python call chestundi.

**Simple meaning:** `__init__` object ki starting/initial values set cheyyadaniki use chestam.


```python
class Student:  # Student ane class/blueprint create chestunnam.
    def __init__(self, name, age, course):  # __init__ constructor; object create ayye time lo automatic ga run ayyi starting values set chestundi.
        self.name = name  # self.name ante current object lo name attribute; right side name parameter value ni store chestunnam.
        self.age = age  # self.age ante current object lo age attribute; right side age parameter value ni store chestunnam.
        self.course = course  # self.course ante current object lo course attribute; right side course parameter value ni store chestunnam.

    def show_details(self):  # Student details display cheyyadaniki method create chestunnam.
        print(f"Name: {self.name}, Age: {self.age}, Course: {self.course}")  # Object data ni readable format lo print chestunnam.

student_one = Student("Sai", 24, "Python")  # Student class nundi first object create chestunnam.
student_two = Student("Kumar", 25, "Machine Learning")  # Student class nundi second object create chestunnam.

student_one.show_details()  # First object details display chestunnam.
student_two.show_details()  # Second object details display chestunnam.
```

## 2. Instance Variables and Instance Methods

**Concept:** Instance variable ante object-specific data. Instance method ante object data meeda work chese function.

**Why use it?** Prati object ki own values untayi. Example: oka employee salary vere, inkoka employee salary vere.


```python
class Employee:  # Employee blueprint define chestunnam.
    def __init__(self, employee_name, monthly_salary):  # Employee object create ayye time lo name and salary set chestunnam.
        self.employee_name = employee_name  # Employee name ni instance variable ga store chestunnam.
        self.monthly_salary = monthly_salary  # Salary ni instance variable ga store chestunnam.

    def yearly_salary(self):  # Yearly salary calculate cheyyadaniki instance method.
        return self.monthly_salary * 12  # Monthly salary ni 12 tho multiply chesi annual salary return chestunnam.

employee = Employee("Ravi", 50000)  # Employee object create chestunnam.
print(employee.employee_name)  # Object lo stored employee name print chestunnam.
print(employee.yearly_salary())  # Method call chesi yearly salary print chestunnam.
```

## 3. Encapsulation

**Concept:** Encapsulation ante data ni direct access cheyyakunda controlled methods/properties through access cheyyadam.

**Why use it?** Invalid data object lo store avvakunda protect cheyyadaniki. Example: bank balance negative ga direct ga set cheyyakudadhu.


```python
class BankAccount:  # Bank account blueprint create chestunnam.
    def __init__(self, owner_name, opening_balance):  # Account create ayye time lo owner and balance set chestunnam.
        self.owner_name = owner_name  # Owner name public attribute ga store chestunnam.
        self.__balance = opening_balance  # Balance ni private attribute ga store chestunnam.

    def deposit(self, amount):  # Money add cheyyadaniki deposit method.
        if amount <= 0:  # Deposit amount valid aa kadha ani check chestunnam.
            raise ValueError("Deposit amount positive undali")  # Invalid amount vaste clear error istunnam.
        self.__balance += amount  # Valid amount ni private balance ki add chestunnam.

    def withdraw(self, amount):  # Money remove cheyyadaniki withdraw method.
        if amount > self.__balance:  # Withdraw amount available balance kanna ekkuva aa ani check chestunnam.
            raise ValueError("Insufficient balance")  # Balance saripokapothe clear error istunnam.
        self.__balance -= amount  # Valid withdraw amount ni balance nundi subtract chestunnam.

    def get_balance(self):  # Private balance ni safe ga read cheyyadaniki getter method.
        return self.__balance  # Current balance return chestunnam.

account = BankAccount("Sai", 1000)  # BankAccount object create chestunnam.
account.deposit(500)  # Account lo 500 deposit chestunnam.
account.withdraw(300)  # Account nundi 300 withdraw chestunnam.
print(account.get_balance())  # Final balance ni getter through print chestunnam.
```

## 4. Inheritance

**Concept:** Inheritance ante oka class properties and methods ni inkoka class reuse cheyyadam.

**Why use it?** Common code duplicate cheyyakunda parent class lo pettukoni child classes lo reuse cheyyadaniki.


```python
class Vehicle:  # Parent class Vehicle create chestunnam.
    def __init__(self, brand):  # Vehicle brand initialize cheyyadaniki constructor.
        self.brand = brand  # Brand ni object lo store chestunnam.

    def start(self):  # Common start behavior define chestunnam.
        print(f"{self.brand} vehicle started")  # Vehicle start message print chestunnam.

class Car(Vehicle):  # Car class Vehicle nundi inherit chestundi.
    def __init__(self, brand, fuel_type):  # Car ki brand plus fuel type initialize chestunnam.
        super().__init__(brand)  # Parent Vehicle constructor call chesi brand set chestunnam.
        self.fuel_type = fuel_type  # Car-specific fuel type store chestunnam.

    def show_fuel_type(self):  # Car fuel type display cheyyadaniki method.
        print(f"Fuel type: {self.fuel_type}")  # Fuel type print chestunnam.

car = Car("Toyota", "Petrol")  # Car object create chestunnam.
car.start()  # Parent class nundi inherited method call chestunnam.
car.show_fuel_type()  # Child class own method call chestunnam.
```

## 5. Polymorphism

**Concept:** Polymorphism ante same method name different classes lo different behavior ivvadam.

**Why use it?** Same interface use chesi different object types ni flexible ga handle cheyyadaniki.


```python
class Dog:  # Dog class create chestunnam.
    def speak(self):  # Dog sound behavior define chestunnam.
        return "Dog barks"  # Dog-specific sound return chestunnam.

class Cat:  # Cat class create chestunnam.
    def speak(self):  # Cat sound behavior define chestunnam.
        return "Cat meows"  # Cat-specific sound return chestunnam.

animals = [Dog(), Cat()]  # Different class objects ni same list lo store chestunnam.

for animal in animals:  # List lo unna every animal object meeda loop chestunnam.
    print(animal.speak())  # Same speak method call chestunnam but output object type batti marutundi.
```

## 6. Abstraction

**Concept:** Abstraction ante implementation details hide chesi required behavior/interface matrame expose cheyyadam.

**Why use it?** Child classes compulsory ga certain methods implement cheyyali ani enforce cheyyadaniki.


```python
from abc import ABC  # Abstract base class create cheyyadaniki ABC import chestunnam.
from abc import abstractmethod  # Abstract methods define cheyyadaniki abstractmethod import chestunnam.

class Payment(ABC):  # Abstract parent class Payment create chestunnam.
    @abstractmethod  # Ee method child class lo compulsory implement avvali ani mark chestunnam.
    def pay(self, amount):  # Payment interface define chestunnam.
        pass  # Abstract method ki body required ledu because child class implement chestundi.

class UpiPayment(Payment):  # UPI payment class Payment nundi inherit chestundi.
    def pay(self, amount):  # Abstract pay method ni actual ga implement chestunnam.
        print(f"Paid {amount} using UPI")  # UPI payment message print chestunnam.

payment = UpiPayment()  # Concrete child class object create chestunnam.
payment.pay(250)  # Payment interface through amount pay chestunnam.
```

## 7. Class Variables, Class Methods, and Static Methods

**Concept:** Class variable class level lo share avtundi. Class method class data meeda work chestundi. Static method class/object data require cheyyakunda utility function laga work chestundi.

**Why use it?** Shared data and utility logic ni class inside organized ga keep cheyyadaniki.


```python
class Course:  # Course blueprint create chestunnam.
    platform_name = "AILearnings"  # Class variable create chestunnam; all objects ki common value.

    def __init__(self, course_name):  # Course object create ayye time lo course name set chestunnam.
        self.course_name = course_name  # Course name ni instance variable ga store chestunnam.

    @classmethod  # Method class level data access chestundi ani decorator use chestunnam.
    def change_platform_name(cls, new_platform_name):  # Class variable update cheyyadaniki class method.
        cls.platform_name = new_platform_name  # Shared platform name ni update chestunnam.

    @staticmethod  # Object/class state required leni utility method ani decorator use chestunnam.
    def is_valid_duration(hours):  # Course duration valid aa kadha ani check chestunnam.
        return hours > 0  # Hours positive unte valid ani True return chestunnam.

course = Course("Python OOP")  # Course object create chestunnam.
print(course.platform_name)  # Object through class variable read chestunnam.
Course.change_platform_name("KrishAI")  # Class method through shared platform name change chestunnam.
print(course.platform_name)  # Updated class variable value print chestunnam.
print(Course.is_valid_duration(10))  # Static method call chesi duration valid aa kadha check chestunnam.
```

## 8. Method Overriding

**Concept:** Child class parent method ni same name tho re-define chesthe method overriding antaru.

**Why use it?** Parent behavior generic ga unte, child class ki custom behavior ivvadani ki overriding use chestam.


```python
class Notification:  # Parent notification class create chestunnam.
    def send(self):  # Generic send behavior define chestunnam.
        print("Sending notification")  # Generic message print chestunnam.

class EmailNotification(Notification):  # EmailNotification parent nundi inherit chestundi.
    def send(self):  # Parent send method ni override chestunnam.
        print("Sending email notification")  # Email-specific message print chestunnam.

notification = EmailNotification()  # Child class object create chestunnam.
notification.send()  # Overridden child method execute avtundi.
```

## 9. Composition

**Concept:** Composition ante oka class object ni inkoka class inside use cheyyadam. Has-a relationship ani cheptaru.

**Why use it?** Inheritance kakunda small reusable objects ni combine chesi bigger object create cheyyadaniki.


```python
class Engine:  # Engine class create chestunnam.
    def start_engine(self):  # Engine start behavior define chestunnam.
        print("Engine started")  # Engine start message print chestunnam.

class Bike:  # Bike class create chestunnam.
    def __init__(self):  # Bike object create ayye time lo setup chestunnam.
        self.engine = Engine()  # Bike has-a Engine relationship create chestunnam.

    def start_bike(self):  # Bike start behavior define chestunnam.
        self.engine.start_engine()  # Bike inside unna engine object method call chestunnam.
        print("Bike started")  # Bike start message print chestunnam.

bike = Bike()  # Bike object create chestunnam.
bike.start_bike()  # Bike start method call chestunnam.
```

## 10. Mini Project - Library Book System

**Concepts used:** Class, object, encapsulation, methods, list of objects.

**Why this example?** Real-world app lo OOP concepts ela combine avtayo easy ga understand avvadani ki.


```python
class Book:  # Book blueprint create chestunnam.
    def __init__(self, title, author):  # Book object create ayye time lo title and author set chestunnam.
        self.title = title  # Book title ni store chestunnam.
        self.author = author  # Book author ni store chestunnam.
        self.is_available = True  # Book initially available ani mark chestunnam.

    def borrow(self):  # Book borrow cheyyadaniki method.
        if not self.is_available:  # Book already borrowed aa ani check chestunnam.
            print(f"{self.title} currently not available")  # Not available message print chestunnam.
            return  # Further logic stop chestunnam.
        self.is_available = False  # Book borrowed kabatti available status False chestunnam.
        print(f"You borrowed {self.title}")  # Borrow success message print chestunnam.

    def return_book(self):  # Book return cheyyadaniki method.
        self.is_available = True  # Book available status malli True chestunnam.
        print(f"You returned {self.title}")  # Return success message print chestunnam.

class Library:  # Library blueprint create chestunnam.
    def __init__(self):  # Library object create ayye time lo setup chestunnam.
        self.books = []  # Library lo books store cheyyadaniki empty list create chestunnam.

    def add_book(self, book):  # New book add cheyyadaniki method.
        self.books.append(book)  # Book object ni library list lo add chestunnam.

    def show_available_books(self):  # Available books display cheyyadaniki method.
        for book in self.books:  # Library lo unna every book meeda loop chestunnam.
            if book.is_available:  # Book available aa kadha check chestunnam.
                print(f"{book.title} by {book.author}")  # Available book details print chestunnam.

library = Library()  # Library object create chestunnam.
python_book = Book("Python Basics", "Krish")  # First book object create chestunnam.
ml_book = Book("Machine Learning", "Sai")  # Second book object create chestunnam.

library.add_book(python_book)  # First book ni library lo add chestunnam.
library.add_book(ml_book)  # Second book ni library lo add chestunnam.

library.show_available_books()  # Current available books display chestunnam.
python_book.borrow()  # Python book borrow chestunnam.
library.show_available_books()  # Borrow taruvata available books display chestunnam.
python_book.return_book()  # Python book return chestunnam.
library.show_available_books()  # Return taruvata available books display chestunnam.
```

## 11. Classes and Objects Practice Questions With Solutions

**Concept:** Practice questions ante topic ni apply cheyyadaniki small tasks.

**Why use it?** Class/object syntax chadavadam okati, but own ga object create cheyyadam practice chesthe concept strong avtundi.

**Question:** `Laptop` class create cheyyandi. Brand, RAM, price store cheyyandi. Discount apply chesi final price calculate cheyyandi.


```python
class Laptop:  # Laptop ane class/blueprint create chestunnam.
    def __init__(self, brand, ram_gb, price):  # Object create ayye time lo brand, RAM, price set chestunnam.
        self.brand = brand  # Current laptop object lo brand value store chestunnam.
        self.ram_gb = ram_gb  # Current laptop object lo RAM size store chestunnam.
        self.price = price  # Current laptop object lo original price store chestunnam.

    def apply_discount(self, discount_percentage):  # Discount apply cheyyadaniki method create chestunnam.
        discount_amount = self.price * discount_percentage / 100  # Discount amount calculate chestunnam.
        final_price = self.price - discount_amount  # Original price nundi discount subtract chestunnam.
        return final_price  # Final discounted price return chestunnam.

laptop = Laptop("Dell", 16, 70000)  # Laptop class nundi object create chestunnam.
print(laptop.brand)  # Laptop brand print chestunnam.
print(laptop.apply_discount(10))  # 10 percent discount taruvata final price print chestunnam.
```

## 12. Magic Methods In Python

**Concept:** Magic methods ante double underscore methods, like `__init__`, `__str__`, `__len__`. Python internally automatic ga call chestundi.

**Why use it?** Built-in behavior ni custom class ki ivvadani ki. Example: object print cheste readable text ravali ante `__str__` use chestam.

**Common magic methods:**

- `__init__`: Object create ayye time lo run avtundi.
- `__str__`: `print(object)` call chesinappudu readable output istundi.
- `__len__`: `len(object)` call chesinappudu length return chestundi.


```python
class ShoppingCart:  # ShoppingCart class create chestunnam.
    def __init__(self):  # Cart object create ayye time lo automatic ga run avtundi.
        self.items = []  # Cart lo items store cheyyadaniki empty list create chestunnam.

    def add_item(self, item_name):  # Cart lo item add cheyyadaniki normal method.
        self.items.append(item_name)  # Item name ni items list lo add chestunnam.

    def __len__(self):  # len(cart) call chesinappudu Python ee method ni automatic ga call chestundi.
        return len(self.items)  # Cart lo enni items unnayo return chestunnam.

    def __str__(self):  # print(cart) call chesinappudu Python ee method ni automatic ga call chestundi.
        return f"ShoppingCart items: {self.items}"  # Cart object ni readable text ga return chestunnam.

cart = ShoppingCart()  # ShoppingCart object create chestunnam.
cart.add_item("Laptop")  # Cart lo Laptop add chestunnam.
cart.add_item("Mouse")  # Cart lo Mouse add chestunnam.
print(len(cart))  # __len__ method automatic ga run ayi item count print avtundi.
print(cart)  # __str__ method automatic ga run ayi readable cart details print avtayi.
```

## 13. Operator Overloading In Python

**Concept:** Operator overloading ante `+`, `-`, `*`, `==` laanti operators ki custom class objects meeda own behavior define cheyyadam.

**Why use it?** Objects ni natural ga add/compare cheyyadaniki. Example: two money objects add chesthe total money object ravali.

**Important:** Operator overloading magic methods tho chestam. Example: `+` kosam `__add__` method.


```python
class Money:  # Money amount represent cheyyadaniki class create chestunnam.
    def __init__(self, rupees):  # Money object create ayye time lo rupees set chestunnam.
        self.rupees = rupees  # Rupees value ni current object lo store chestunnam.

    def __add__(self, other):  # money_one + money_two call chesinappudu ee method run avtundi.
        total_rupees = self.rupees + other.rupees  # Two money objects rupees values add chestunnam.
        return Money(total_rupees)  # Total amount tho new Money object return chestunnam.

    def __str__(self):  # Money object print chesinappudu readable format ivvadaniki.
        return f"Rs.{self.rupees}"  # Rupees ni string format lo return chestunnam.

wallet_amount = Money(500)  # First Money object create chestunnam.
bank_amount = Money(1500)  # Second Money object create chestunnam.
total_amount = wallet_amount + bank_amount  # __add__ automatic ga call ayi total Money object create avtundi.
print(total_amount)  # __str__ automatic ga call ayi total amount print avtundi.
```

## 14. Custom Exception Handling

**Concept:** Custom exception ante manam own error class create cheyyadam.

**Why use it?** Business rules clear ga explain cheyyadaniki. Example: age less than 18 unte `AgeTooSmallError` ani specific error ivvachu.

**Flow:** Condition check -> Invalid case unte custom exception raise -> `try except` lo handle -> User-friendly message print.


```python
class AgeTooSmallError(Exception):  # Custom exception class create chestunnam.
    pass  # Extra logic required ledu kabatti pass use chestunnam.

def register_voter(age):  # Voter registration validation function create chestunnam.
    if age < 18:  # Age 18 kanna takkuva aa ani check chestunnam.
        raise AgeTooSmallError("Voting ki age minimum 18 undali")  # Custom exception raise chestunnam.
    return "Registration successful"  # Age valid unte success message return chestunnam.

try:  # Risky code ni try block lo run chestunnam.
    result = register_voter(16)  # Invalid age tho function call chestunnam.
    print(result)  # Error rakapothe result print chestunnam.
except AgeTooSmallError as error:  # Custom exception catch chestunnam.
    print(error)  # User-friendly custom error message print chestunnam.
```

## 15. Complete OOPS Practice Question With Solution

**Question:** Student result system create cheyyandi. Student name, marks store cheyyali. Marks invalid unte custom exception raise cheyyali. `+` operator use chesi two subject marks combine cheyyali. Final grade show cheyyali.

**Concepts used:** Class, object, encapsulation, magic method, operator overloading, custom exception, method.

**Why this question?** Image lo unna OOP concepts ni one complete example lo combine cheyyadaniki.


```python
class InvalidMarksError(Exception):  # Invalid marks kosam custom exception create chestunnam.
    pass  # Custom message raise time lo istam kabatti body empty ga pettam.

class SubjectMarks:  # Single subject marks represent cheyyadaniki class create chestunnam.
    def __init__(self, subject_name, marks):  # Subject object create ayye time lo subject and marks set chestunnam.
        if marks < 0 or marks > 100:  # Marks 0 to 100 madhya unnaya ani validate chestunnam.
            raise InvalidMarksError("Marks 0 nundi 100 madhya undali")  # Invalid marks unte custom error raise chestunnam.
        self.subject_name = subject_name  # Subject name ni object lo store chestunnam.
        self.marks = marks  # Valid marks ni object lo store chestunnam.

    def __add__(self, other):  # Two SubjectMarks objects add chesinappudu run avtundi.
        return self.marks + other.marks  # Two subjects marks total return chestunnam.

class StudentResult:  # Student result represent cheyyadaniki class create chestunnam.
    def __init__(self, student_name):  # StudentResult object create ayye time lo name set chestunnam.
        self.student_name = student_name  # Student name ni current object lo store chestunnam.
        self.subjects = []  # Subject marks store cheyyadaniki empty list create chestunnam.

    def add_subject(self, subject_marks):  # Student ki subject marks add cheyyadaniki method.
        self.subjects.append(subject_marks)  # SubjectMarks object ni subjects list lo add chestunnam.

    def calculate_average(self):  # Average marks calculate cheyyadaniki method.
        total_marks = sum(subject.marks for subject in self.subjects)  # All subject marks total calculate chestunnam.
        return total_marks / len(self.subjects)  # Total ni subject count tho divide chesi average return chestunnam.

    def get_grade(self):  # Average basis meeda grade decide cheyyadaniki method.
        average_marks = self.calculate_average()  # Average marks calculate chestunnam.
        if average_marks >= 90:  # Average 90 or above aa ani check chestunnam.
            return "A"  # Excellent performance ki A grade return chestunnam.
        if average_marks >= 75:  # Average 75 or above aa ani check chestunnam.
            return "B"  # Good performance ki B grade return chestunnam.
        if average_marks >= 60:  # Average 60 or above aa ani check chestunnam.
            return "C"  # Average performance ki C grade return chestunnam.
        return "D"  # 60 kanna takkuva unte D grade return chestunnam.

math_marks = SubjectMarks("Math", 85)  # Math subject marks object create chestunnam.
science_marks = SubjectMarks("Science", 90)  # Science subject marks object create chestunnam.
combined_marks = math_marks + science_marks  # __add__ use chesi two subject marks combine chestunnam.
print(combined_marks)  # Combined marks print chestunnam.

student_result = StudentResult("Sai")  # StudentResult object create chestunnam.
student_result.add_subject(math_marks)  # Math marks ni student result lo add chestunnam.
student_result.add_subject(science_marks)  # Science marks ni student result lo add chestunnam.
print(student_result.calculate_average())  # Average marks print chestunnam.
print(student_result.get_grade())  # Final grade print chestunnam.
```
