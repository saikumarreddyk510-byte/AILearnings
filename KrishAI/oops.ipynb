## 1. Class and Object

**Concept:** Class ante plan/blueprint. Object ante aa class nundi create ayina actual item.

**Why use it?** Same structure unna multiple objects create cheyyadaniki class use chestam. Example: many students ki same fields name, age, course untayi.

### `self` ante enti?

`self` ante **current object reference**. Class nundi object create ayyaka, aa object own data ni access cheyyadaniki Python lo `self` use chestam.

Example: `student_one = Student("Sai", 24, "Python")` ani create chesthe, `self` ante ikkada `student_one` object ni point chestundi.

`self.name = name` ante external ga vachina `name` value ni current object lo `name` attribute ga save cheyyadam.

**Important:** `self` keyword fixed kaadu, but Python convention. Manam vere name use cheyyachu, kani industry lo almost always `self` use chestaru.

### `def __init__(...)` ante enti?

`__init__` ante Python lo **constructor method**. Object create ayye time lo automatic ga run avtundi.

`def __init__(self, name, age, course):` ante Student object create chesinappudu `name`, `age`, `course` values receive chesi aa object lo save cheyyadaniki method.

Example: `Student("Sai", 24, "Python")` ani call chesthe Python automatic ga `__init__` ni call chestundi. Appudu `name = "Sai"`, `age = 24`, `course = "Python"` avtayi.

`__init__` lo double underscores untayi kabatti idi special/magic method. Manam direct ga call cheyyalsina avasaram ledu; object create chesinappude Python call chestundi.

**Simple meaning:** `__init__` object ki starting/initial values set cheyyadaniki use chestam.


````python
class Student:  # Student ane class/blueprint create chestunnam.
    def __init__(self, name, age, course):  # __init__ constructor; object create ayye time lo automatic ga run ayyi starting values set chestundi.
        self.name = name  # self.name ante current object lo name attribute; right side name parameter value ni store chestunnam.
        self.age = age  # self.age ante current object lo age attribute; right side age parameter value ni store chestunnam.
        self.course = course  # self.course ante current object lo course attribute; right side course parameter value ni store chestunnam.

    def show_details(self):  # Student details display cheyyadaniki method create chestunnam.
        print(f"Name: {self.name}, Age: {self.age}, Course: {self.course}")  # Object data ni readable format lo print chestunnam.

student_one = Student("Sai", 24, "Python")  # Student class nundi first object create chestunnam.
student_two = Student("Kumar", 25, "Machine Learning")  # Student class nundi second object create chestunnam.

student_one.show_details()  # First object details display chestunnam.
student_two.show_details()  # Second object details display chestunnam.
````

## 2. Instance Variables and Instance Methods

**Concept:** Instance variable ante object-specific data. Instance method ante object data meeda work chese function.

**Why use it?** Prati object ki own values untayi. Example: oka employee salary vere, inkoka employee salary vere.


````python
class Employee:  # Employee blueprint define chestunnam.
    def __init__(self, employee_name, monthly_salary):  # Employee object create ayye time lo name and salary set chestunnam.
        self.employee_name = employee_name  # Employee name ni instance variable ga store chestunnam.
        self.monthly_salary = monthly_salary  # Salary ni instance variable ga store chestunnam.

    def yearly_salary(self):  # Yearly salary calculate cheyyadaniki instance method.
        return self.monthly_salary * 12  # Monthly salary ni 12 tho multiply chesi annual salary return chestunnam.

employee = Employee("Ravi", 50000)  # Employee object create chestunnam.
print(employee.employee_name)  # Object lo stored employee name print chestunnam.
print(employee.yearly_salary())  # Method call chesi yearly salary print chestunnam.
````

## 3. Encapsulation

**Concept:** Encapsulation ante data ni direct access cheyyakunda controlled methods/properties through access cheyyadam.

**Why use it?** Invalid data object lo store avvakunda protect cheyyadaniki. Example: bank balance negative ga direct ga set cheyyakudadhu.


````python
class BankAccount:  # Bank account blueprint create chestunnam.
    def __init__(self, owner_name, opening_balance):  # Account create ayye time lo owner and balance set chestunnam.
        self.owner_name = owner_name  # Owner name public attribute ga store chestunnam.
        self.__balance = opening_balance  # Balance ni private attribute ga store chestunnam.

    def deposit(self, amount):  # Money add cheyyadaniki deposit method.
        if amount <= 0:  # Deposit amount valid aa kadha ani check chestunnam.
            raise ValueError("Deposit amount positive undali")  # Invalid amount vaste clear error istunnam.
        self.__balance += amount  # Valid amount ni private balance ki add chestunnam.

    def withdraw(self, amount):  # Money remove cheyyadaniki withdraw method.
        if amount > self.__balance:  # Withdraw amount available balance kanna ekkuva aa ani check chestunnam.
            raise ValueError("Insufficient balance")  # Balance saripokapothe clear error istunnam.
        self.__balance -= amount  # Valid withdraw amount ni balance nundi subtract chestunnam.

    def get_balance(self):  # Private balance ni safe ga read cheyyadaniki getter method.
        return self.__balance  # Current balance return chestunnam.

account = BankAccount("Sai", 1000)  # BankAccount object create chestunnam.
account.deposit(500)  # Account lo 500 deposit chestunnam.
account.withdraw(300)  # Account nundi 300 withdraw chestunnam.
print(account.get_balance())  # Final balance ni getter through print chestunnam.
````

## 4. Inheritance

**Concept:** Inheritance ante oka class properties and methods ni inkoka class reuse cheyyadam.

**Why use it?** Common code duplicate cheyyakunda parent class lo pettukoni child classes lo reuse cheyyadaniki.


````python
class Vehicle:  # Parent class Vehicle create chestunnam.
    def __init__(self, brand):  # Vehicle brand initialize cheyyadaniki constructor.
        self.brand = brand  # Brand ni object lo store chestunnam.

    def start(self):  # Common start behavior define chestunnam.
        print(f"{self.brand} vehicle started")  # Vehicle start message print chestunnam.

class Car(Vehicle):  # Car class Vehicle nundi inherit chestundi.
    def __init__(self, brand, fuel_type):  # Car ki brand plus fuel type initialize chestunnam.
        super().__init__(brand)  # Parent Vehicle constructor call chesi brand set chestunnam.
        self.fuel_type = fuel_type  # Car-specific fuel type store chestunnam.

    def show_fuel_type(self):  # Car fuel type display cheyyadaniki method.
        print(f"Fuel type: {self.fuel_type}")  # Fuel type print chestunnam.

car = Car("Toyota", "Petrol")  # Car object create chestunnam.
car.start()  # Parent class nundi inherited method call chestunnam.
car.show_fuel_type()  # Child class own method call chestunnam.
````

## 5. Polymorphism

**Concept:** Polymorphism ante same method name different classes lo different behavior ivvadam.

**Why use it?** Same interface use chesi different object types ni flexible ga handle cheyyadaniki.


````python
class Dog:  # Dog class create chestunnam.
    def speak(self):  # Dog sound behavior define chestunnam.
        return "Dog barks"  # Dog-specific sound return chestunnam.

class Cat:  # Cat class create chestunnam.
    def speak(self):  # Cat sound behavior define chestunnam.
        return "Cat meows"  # Cat-specific sound return chestunnam.

animals = [Dog(), Cat()]  # Different class objects ni same list lo store chestunnam.

for animal in animals:  # List lo unna every animal object meeda loop chestunnam.
    print(animal.speak())  # Same speak method call chestunnam but output object type batti marutundi.
````

## 6. Abstraction

**Concept:** Abstraction ante implementation details hide chesi required behavior/interface matrame expose cheyyadam.

**Why use it?** Child classes compulsory ga certain methods implement cheyyali ani enforce cheyyadaniki.


````python
from abc import ABC  # Abstract base class create cheyyadaniki ABC import chestunnam.
from abc import abstractmethod  # Abstract methods define cheyyadaniki abstractmethod import chestunnam.

class Payment(ABC):  # Abstract parent class Payment create chestunnam.
    @abstractmethod  # Ee method child class lo compulsory implement avvali ani mark chestunnam.
    def pay(self, amount):  # Payment interface define chestunnam.
        pass  # Abstract method ki body required ledu because child class implement chestundi.

class UpiPayment(Payment):  # UPI payment class Payment nundi inherit chestundi.
    def pay(self, amount):  # Abstract pay method ni actual ga implement chestunnam.
        print(f"Paid {amount} using UPI")  # UPI payment message print chestunnam.

payment = UpiPayment()  # Concrete child class object create chestunnam.
payment.pay(250)  # Payment interface through amount pay chestunnam.
````

## 7. Class Variables, Class Methods, and Static Methods

**Concept:** Class variable class level lo share avtundi. Class method class data meeda work chestundi. Static method class/object data require cheyyakunda utility function laga work chestundi.

**Why use it?** Shared data and utility logic ni class inside organized ga keep cheyyadaniki.


````python
class Course:  # Course blueprint create chestunnam.
    platform_name = "AILearnings"  # Class variable create chestunnam; all objects ki common value.

    def __init__(self, course_name):  # Course object create ayye time lo course name set chestunnam.
        self.course_name = course_name  # Course name ni instance variable ga store chestunnam.

    @classmethod  # Method class level data access chestundi ani decorator use chestunnam.
    def change_platform_name(cls, new_platform_name):  # Class variable update cheyyadaniki class method.
        cls.platform_name = new_platform_name  # Shared platform name ni update chestunnam.

    @staticmethod  # Object/class state required leni utility method ani decorator use chestunnam.
    def is_valid_duration(hours):  # Course duration valid aa kadha ani check chestunnam.
        return hours > 0  # Hours positive unte valid ani True return chestunnam.

course = Course("Python OOP")  # Course object create chestunnam.
print(course.platform_name)  # Object through class variable read chestunnam.
Course.change_platform_name("KrishAI")  # Class method through shared platform name change chestunnam.
print(course.platform_name)  # Updated class variable value print chestunnam.
print(Course.is_valid_duration(10))  # Static method call chesi duration valid aa kadha check chestunnam.
````

## 8. Method Overriding

**Concept:** Child class parent method ni same name tho re-define chesthe method overriding antaru.

**Why use it?** Parent behavior generic ga unte, child class ki custom behavior ivvadani ki overriding use chestam.


````python
class Notification:  # Parent notification class create chestunnam.
    def send(self):  # Generic send behavior define chestunnam.
        print("Sending notification")  # Generic message print chestunnam.

class EmailNotification(Notification):  # EmailNotification parent nundi inherit chestundi.
    def send(self):  # Parent send method ni override chestunnam.
        print("Sending email notification")  # Email-specific message print chestunnam.

notification = EmailNotification()  # Child class object create chestunnam.
notification.send()  # Overridden child method execute avtundi.
````

## 9. Composition

**Concept:** Composition ante oka class object ni inkoka class inside use cheyyadam. Has-a relationship ani cheptaru.

**Why use it?** Inheritance kakunda small reusable objects ni combine chesi bigger object create cheyyadaniki.


````python
class Engine:  # Engine class create chestunnam.
    def start_engine(self):  # Engine start behavior define chestunnam.
        print("Engine started")  # Engine start message print chestunnam.

class Bike:  # Bike class create chestunnam.
    def __init__(self):  # Bike object create ayye time lo setup chestunnam.
        self.engine = Engine()  # Bike has-a Engine relationship create chestunnam.

    def start_bike(self):  # Bike start behavior define chestunnam.
        self.engine.start_engine()  # Bike inside unna engine object method call chestunnam.
        print("Bike started")  # Bike start message print chestunnam.

bike = Bike()  # Bike object create chestunnam.
bike.start_bike()  # Bike start method call chestunnam.
````

## 10. Mini Project - Library Book System

**Concepts used:** Class, object, encapsulation, methods, list of objects.

**Why this example?** Real-world app lo OOP concepts ela combine avtayo easy ga understand avvadani ki.


````python
class Book:  # Book blueprint create chestunnam.
    def __init__(self, title, author):  # Book object create ayye time lo title and author set chestunnam.
        self.title = title  # Book title ni store chestunnam.
        self.author = author  # Book author ni store chestunnam.
        self.is_available = True  # Book initially available ani mark chestunnam.

    def borrow(self):  # Book borrow cheyyadaniki method.
        if not self.is_available:  # Book already borrowed aa ani check chestunnam.
            print(f"{self.title} currently not available")  # Not available message print chestunnam.
            return  # Further logic stop chestunnam.
        self.is_available = False  # Book borrowed kabatti available status False chestunnam.
        print(f"You borrowed {self.title}")  # Borrow success message print chestunnam.

    def return_book(self):  # Book return cheyyadaniki method.
        self.is_available = True  # Book available status malli True chestunnam.
        print(f"You returned {self.title}")  # Return success message print chestunnam.

class Library:  # Library blueprint create chestunnam.
    def __init__(self):  # Library object create ayye time lo setup chestunnam.
        self.books = []  # Library lo books store cheyyadaniki empty list create chestunnam.

    def add_book(self, book):  # New book add cheyyadaniki method.
        self.books.append(book)  # Book object ni library list lo add chestunnam.

    def show_available_books(self):  # Available books display cheyyadaniki method.
        for book in self.books:  # Library lo unna every book meeda loop chestunnam.
            if book.is_available:  # Book available aa kadha check chestunnam.
                print(f"{book.title} by {book.author}")  # Available book details print chestunnam.

library = Library()  # Library object create chestunnam.
python_book = Book("Python Basics", "Krish")  # First book object create chestunnam.
ml_book = Book("Machine Learning", "Sai")  # Second book object create chestunnam.

library.add_book(python_book)  # First book ni library lo add chestunnam.
library.add_book(ml_book)  # Second book ni library lo add chestunnam.

library.show_available_books()  # Current available books display chestunnam.
python_book.borrow()  # Python book borrow chestunnam.
library.show_available_books()  # Borrow taruvata available books display chestunnam.
python_book.return_book()  # Python book return chestunnam.
library.show_available_books()  # Return taruvata available books display chestunnam.
````

## 12. Magic Methods In Python

**Concept:** Magic methods ante double underscore methods, like `__init__`, `__str__`, `__len__`. Python internally automatic ga call chestundi.

**Why use it?** Built-in behavior ni custom class ki ivvadani ki. Example: object print cheste readable text ravali ante `__str__` use chestam.

**Common magic methods:**

- `__init__`: Object create ayye time lo run avtundi.
- `__str__`: `print(object)` call chesinappudu readable output istundi.
- `__len__`: `len(object)` call chesinappudu length return chestundi.


````python
class ShoppingCart:  # ShoppingCart class create chestunnam.
    def __init__(self):  # Cart object create ayye time lo automatic ga run avtundi.
        self.items = []  # Cart lo items store cheyyadaniki empty list create chestunnam.

    def add_item(self, item_name):  # Cart lo item add cheyyadaniki normal method.
        self.items.append(item_name)  # Item name ni items list lo add chestunnam.

    def __len__(self):  # len(cart) call chesinappudu Python ee method ni automatic ga call chestundi.
        return len(self.items)  # Cart lo enni items unnayo return chestunnam.

    def __str__(self):  # print(cart) call chesinappudu Python ee method ni automatic ga call chestundi.
        return f"ShoppingCart items: {self.items}"  # Cart object ni readable text ga return chestunnam.

cart = ShoppingCart()  # ShoppingCart object create chestunnam.
cart.add_item("Laptop")  # Cart lo Laptop add chestunnam.
cart.add_item("Mouse")  # Cart lo Mouse add chestunnam.
print(len(cart))  # __len__ method automatic ga run ayi item count print avtundi.
print(cart)  # __str__ method automatic ga run ayi readable cart details print avtayi.
````

## 13. Operator Overloading In Python

**Concept:** Operator overloading ante `+`, `-`, `*`, `==` laanti operators ki custom class objects meeda own behavior define cheyyadam.

**Why use it?** Objects ni natural ga add/compare cheyyadaniki. Example: two money objects add chesthe total money object ravali.

**Important:** Operator overloading magic methods tho chestam. Example: `+` kosam `__add__` method.


````python
class Money:  # Money amount represent cheyyadaniki class create chestunnam.
    def __init__(self, rupees):  # Money object create ayye time lo rupees set chestunnam.
        self.rupees = rupees  # Rupees value ni current object lo store chestunnam.

    def __add__(self, other):  # money_one + money_two call chesinappudu ee method run avtundi.
        total_rupees = self.rupees + other.rupees  # Two money objects rupees values add chestunnam.
        return Money(total_rupees)  # Total amount tho new Money object return chestunnam.

    def __str__(self):  # Money object print chesinappudu readable format ivvadaniki.
        return f"Rs.{self.rupees}"  # Rupees ni string format lo return chestunnam.

wallet_amount = Money(500)  # First Money object create chestunnam.
bank_amount = Money(1500)  # Second Money object create chestunnam.
total_amount = wallet_amount + bank_amount  # __add__ automatic ga call ayi total Money object create avtundi.
print(total_amount)  # __str__ automatic ga call ayi total amount print avtundi.
````

## 14. Custom Exception Handling

**Concept:** Custom exception ante manam own error class create cheyyadam.

**Why use it?** Business rules clear ga explain cheyyadaniki. Example: age less than 18 unte `AgeTooSmallError` ani specific error ivvachu.

**Flow:** Condition check -> Invalid case unte custom exception raise -> `try except` lo handle -> User-friendly message print.


````python
class AgeTooSmallError(Exception):  # Custom exception class create chestunnam.
    pass  # Extra logic required ledu kabatti pass use chestunnam.

def register_voter(age):  # Voter registration validation function create chestunnam.
    if age < 18:  # Age 18 kanna takkuva aa ani check chestunnam.
        raise AgeTooSmallError("Voting ki age minimum 18 undali")  # Custom exception raise chestunnam.
    return "Registration successful"  # Age valid unte success message return chestunnam.

try:  # Risky code ni try block lo run chestunnam.
    result = register_voter(16)  # Invalid age tho function call chestunnam.
    print(result)  # Error rakapothe result print chestunnam.
except AgeTooSmallError as error:  # Custom exception catch chestunnam.
    print(error)  # User-friendly custom error message print chestunnam.
````

## 15. Complete OOPS Practice Question With Solution

**Question:** Student result system create cheyyandi. Student name, marks store cheyyali. Marks invalid unte custom exception raise cheyyali. `+` operator use chesi two subject marks combine cheyyali. Final grade show cheyyali.

**Concepts used:** Class, object, encapsulation, magic method, operator overloading, custom exception, method.

**Why this question?** Image lo unna OOP concepts ni one complete example lo combine cheyyadaniki.


````python
class InvalidMarksError(Exception):  # Invalid marks kosam custom exception create chestunnam.
    pass  # Custom message raise time lo istam kabatti body empty ga pettam.

class SubjectMarks:  # Single subject marks represent cheyyadaniki class create chestunnam.
    def __init__(self, subject_name, marks):  # Subject object create ayye time lo subject and marks set chestunnam.
        if marks < 0 or marks > 100:  # Marks 0 to 100 madhya unnaya ani validate chestunnam.
            raise InvalidMarksError("Marks 0 nundi 100 madhya undali")  # Invalid marks unte custom error raise chestunnam.
        self.subject_name = subject_name  # Subject name ni object lo store chestunnam.
        self.marks = marks  # Valid marks ni object lo store chestunnam.

    def __add__(self, other):  # Two SubjectMarks objects add chesinappudu run avtundi.
        return self.marks + other.marks  # Two subjects marks total return chestunnam.

class StudentResult:  # Student result represent cheyyadaniki class create chestunnam.
    def __init__(self, student_name):  # StudentResult object create ayye time lo name set chestunnam.
        self.student_name = student_name  # Student name ni current object lo store chestunnam.
        self.subjects = []  # Subject marks store cheyyadaniki empty list create chestunnam.

    def add_subject(self, subject_marks):  # Student ki subject marks add cheyyadaniki method.
        self.subjects.append(subject_marks)  # SubjectMarks object ni subjects list lo add chestunnam.

    def calculate_average(self):  # Average marks calculate cheyyadaniki method.
        total_marks = sum(subject.marks for subject in self.subjects)  # All subject marks total calculate chestunnam.
        return total_marks / len(self.subjects)  # Total ni subject count tho divide chesi average return chestunnam.

    def get_grade(self):  # Average basis meeda grade decide cheyyadaniki method.
        average_marks = self.calculate_average()  # Average marks calculate chestunnam.
        if average_marks >= 90:  # Average 90 or above aa ani check chestunnam.
            return "A"  # Excellent performance ki A grade return chestunnam.
        if average_marks >= 75:  # Average 75 or above aa ani check chestunnam.
            return "B"  # Good performance ki B grade return chestunnam.
        if average_marks >= 60:  # Average 60 or above aa ani check chestunnam.
            return "C"  # Average performance ki C grade return chestunnam.
        return "D"  # 60 kanna takkuva unte D grade return chestunnam.

math_marks = SubjectMarks("Math", 85)  # Math subject marks object create chestunnam.
science_marks = SubjectMarks("Science", 90)  # Science subject marks object create chestunnam.
combined_marks = math_marks + science_marks  # __add__ use chesi two subject marks combine chestunnam.
print(combined_marks)  # Combined marks print chestunnam.

student_result = StudentResult("Sai")  # StudentResult object create chestunnam.
student_result.add_subject(math_marks)  # Math marks ni student result lo add chestunnam.
student_result.add_subject(science_marks)  # Science marks ni student result lo add chestunnam.
print(student_result.calculate_average())  # Average marks print chestunnam.
print(student_result.get_grade())  # Final grade print chestunnam.
````

## 12.1 Iterators in Python

**Iterator ante enti?** Iterator ante values ni one-by-one ga provide chese object.

Iterator ki two important methods untayi:

- `__iter__()` -> iterator object ni return chestundi.
- `__next__()` -> next value ni return chestundi.

`next()` call chesina prathi sari next value vastundi. Values complete ayyaka Python `StopIteration` exception raise chestundi.

**Why use it?** Entire data ni memory lo one time load cheyyakunda, required values ni one-by-one process cheyyadaniki iterator useful.

````python
class CountIterator:  # Custom iterator class create chestunnam.
    def __init__(self, maximum_value):  # Iterator limit ni object lo store chestunnam.
        self.maximum_value = maximum_value  # Maximum value ni instance variable ga save chestunnam.
        self.current_value = 1  # First value 1 nundi start chestunnam.

    def __iter__(self):  # Iterable object ni iterator ga return cheyyadaniki method.
        return self  # Current object itself iterator kabatti self return chestunnam.

    def __next__(self):  # Next value provide cheyyadaniki method.
        if self.current_value > self.maximum_value:  # Current value limit cross chesinda ani check chestunnam.
            raise StopIteration  # Values complete ayinappudu iteration stop chestunnam.
        returned_value = self.current_value  # Current value ni temporary variable lo store chestunnam.
        self.current_value += 1  # Next call kosam current value ni one increment chestunnam.
        return returned_value  # Current value ni caller ki return chestunnam.

number_iterator = CountIterator(3)  # 1 nundi 3 varaku values provide chese iterator create chestunnam.
for number in number_iterator:  # Iterator values ni one-by-one read chestunnam.
    print(number)  # Prathi value ni display chestunnam.
````

### Iterator code line-by-line explanation

1. `CountIterator` -> iteration rules define chese class.
2. `maximum_value` -> enni values ivvalo decide chestundi.
3. `current_value` -> next ga return cheyyalsina value track chestundi.
4. `__iter__` -> object ni iterator ga identify cheyyadaniki.
5. `__next__` -> one next value return cheyyadaniki.
6. `StopIteration` -> values complete ayinappudu loop ki stop signal.
7. `for` loop -> internally `iter()` and repeated `next()` calls chestundi.

**Simple meaning:** Iterator memory-friendly ga values ni one-by-one istundi.

## 12.2 Generators With Practical Implementation

**Generator ante enti?** Generator ante `yield` use chesi values ni one-by-one produce chese special function.

Normal function `return` chesina taruvata complete ga end avtundi. Generator `yield` daggara pause avtundi; next value request vachinappudu akkadi nundi resume avtundi.

**Why use it?** Large files, API records, logs, or infinite sequences ni memory efficient ga process cheyyadaniki generators use chestam.

```text
Generator call -> paused generator object
next() -> one value -> pause
next() -> next value -> pause
```

````python
def square_generator(maximum_value):  # Squares one-by-one provide chese generator function.
    for number in range(1, maximum_value + 1):  # 1 nundi maximum value varaku loop chestunnam.
        yield number * number  # Square value return chesi function state ni pause chestunnam.

squares = square_generator(3)  # Generator function call chesi generator object create chestunnam.
print(next(squares))  # First square 1 ni request chestunnam.
print(next(squares))  # Next square 4 ni request chestunnam.
print(next(squares))  # Next square 9 ni request chestunnam.

for square in square_generator(3):  # Generator values ni loop tho one-by-one read chestunnam.
    print(square)  # Current generated square ni display chestunnam.
````

### Generator code line-by-line explanation

1. `def square_generator` -> generator function define chestunnam.
2. `range(...)` -> required numbers sequence create chestundi.
3. `yield` -> value return chesi function state ni save chestundi; function complete ga destroy kaadu.
4. `squares = ...` -> immediate ga all squares list create kaadu; generator object create avtundi.
5. `next(squares)` -> next available value request chestundi.
6. `for` loop -> generator values ni automatic ga consume chestundi.

**Iterator vs Generator:** Generator iterator create cheyyadaniki easy syntax. Manam manually `__iter__` and `__next__` rayalsina avasaram takkuva.

## 12.3 Function Copy, Closures and Decorators

### Function Copy

Python lo functions kuda objects. Oka function ni another variable ki assign chesthe function reference copy avtundi.

```python
new_name = original_function
```

Ikkada function execute avvadu; same function ni point chese second name create avtundi. Call cheyyali ante `new_name()` use chestam.

### Closure

Closure ante inner function outer function variables ni remember cheskovadam, outer function execution complete ayina taruvata kuda.

Use case: configuration, counters, private state.

### Decorator

Decorator ante existing function code ni modify cheyyakunda extra behavior add chese function.

```text
Original function -> Decorator wrapper -> Extra behavior + Original behavior
```

Examples: logging, timing, authentication, validation.

````python
def greet(person_name):  # Person name tho greeting return chese original function.
    return f"Hello {person_name}"  # Greeting string create chesi return chestunnam.

copied_greet = greet  # Function ni execute cheyyakunda same function reference copy chestunnam.
print(copied_greet("Sai"))  # Copied function reference tho original behavior call chestunnam.

def create_multiplier(multiplier):  # Multiplier value capture chese outer function.
    def multiply(number):  # Outer variable ni use chese inner function.
        return number * multiplier  # Remember chesina multiplier tho number multiply chestunnam.
    return multiply  # Inner function ni return chestunnam; idi closure create chestundi.

double = create_multiplier(2)  # multiplier=2 ni closure remember cheskuntundi.
print(double(5))  # Closure 5 ni 2 tho multiply chesi 10 return chestundi.

def log_call(function):  # Original function ni receive chese decorator define chestunnam.
    def wrapper(*args, **kwargs):  # Any positional/keyword arguments accept chese wrapper.
        print("Function start")  # Original function mundu logging chestunnam.
        result = function(*args, **kwargs)  # Original function ni same arguments tho call chestunnam.
        print("Function end")  # Original function taruvata logging chestunnam.
        return result  # Original result ni caller ki return chestunnam.
    return wrapper  # Decorated wrapper function ni return chestunnam.

@log_call  # greet_user function ki log_call decorator apply chestunnam.
def greet_user(user_name):  # User greeting function define chestunnam.
    return f"Welcome {user_name}"  # User ki welcome message return chestunnam.

print(greet_user("Sai"))  # Wrapper logging tho decorated function execute chestunnam.
````

### Function copy, closure, decorator line-by-line explanation

1. `copied_greet = greet` -> function definition copy kaadu; same function object ki another reference.
2. `copied_greet("Sai")` -> second name use chesi original function execute chestunnam.
3. `create_multiplier(2)` -> outer function multiplier value capture chestundi.
4. `multiply` -> inner function outer `multiplier` ni remember chestundi.
5. `double(5)` -> remembered value 2 use chesi 10 return chestundi.
6. `log_call(function)` -> decorator original function ni input ga teesukuntundi.
7. `wrapper(*args, **kwargs)` -> different function arguments ni flexible ga accept chestundi.
8. `function(*args, **kwargs)` -> original function behavior preserve chestundi.
9. `@log_call` -> `greet_user = log_call(greet_user)` ki equivalent.
10. `greet_user("Sai")` -> wrapper first run ayi logging + original greeting execute chestundi.

## Section 14: Python Multi Threading and Multi Processing

![Multi Threading And Multi Processing](assets/multithreading-multiprocessing-topics.png)

**Process ante enti?** Process ante independent ga run ayye oka program instance; daaniki own memory space untundi.

**Thread ante enti?** Thread ante process lopala run ayye chinna execution unit; same process lo unna threads memory ni share chestayi.

**Multithreading vs Multiprocessing:**

- Multithreading -> I/O bound tasks ki manchidi (file read/write, network calls, web scraping). Python GIL (Global Interpreter Lock) valla CPU bound tasks ki full parallel speed ivvadu.
- Multiprocessing -> CPU bound tasks ki manchidi (heavy calculations, image processing). Prathi process ki separate memory and separate Python interpreter untundi kabatti true parallel execution jarugutundi.

Ee section lo cover chese topics:

1. What is process and threads
2. Multithreading practical implementation
3. Multiprocessing practical implementation
4. Thread pool executor and process pool
5. Web scraping usecase with multithreading
6. Real world usecase implementation with multiprocessing

````python
import threading  # Threading module ni import chestunnam.
import time  # Delay simulate cheyyadaniki time module import chestunnam.

def print_numbers():  # Numbers print chese task; I/O bound simulate cheyyadaniki sleep vadutunnam.
    for number in range(1, 4):  # 1 nundi 3 varaku loop chestunnam.
        time.sleep(0.5)  # I/O wait ni simulate cheyyadaniki chinna delay pedutunnam.
        print(f"Number: {number}")  # Current number print chestunnam.

def print_letters():  # Letters print chese task; idi kuda I/O bound simulate chestundi.
    for letter in "abc":  # a, b, c letters meeda loop chestunnam.
        time.sleep(0.5)  # Same delay pedutunnam, so anni threads parallel ga run avtunnattu chupinchali.
        print(f"Letter: {letter}")  # Current letter print chestunnam.

thread_one = threading.Thread(target=print_numbers)  # print_numbers ni run chese thread create chestunnam.
thread_two = threading.Thread(target=print_letters)  # print_letters ni run chese thread create chestunnam.

start_time = time.time()  # Total execution time measure cheyyadaniki start timestamp store chestunnam.
thread_one.start()  # First thread ni start chestunnam.
thread_two.start()  # Second thread ni start chestunnam; rendu almost parallel ga run avtayi.

thread_one.join()  # Main program thread_one complete ayye varaku wait chestundi.
thread_two.join()  # Main program thread_two complete ayye varaku wait chestundi.

print(f"Total time: {time.time() - start_time:.2f} seconds")  # Rendu threads parallel run ayina valla total time takkuva vastundi.
````

### Multithreading code line-by-line explanation

1. `threading.Thread(target=print_numbers)` -> `print_numbers` function ni separate thread lo run cheyyadaniki thread object create chestundi.
2. `.start()` -> thread execution ni actual ga begin chestundi.
3. Rendu threads `.start()` ayina taruvata almost same time lo run avtayi (I/O wait sepatlo CPU vere thread ki switch avtundi).
4. `.join()` -> main program aa thread complete ayye varaku aagi wait chestundi.
5. Sequential ga (one by one) run chesthe ~3 seconds pattundi, kani threading valla parallel ga ~1.5 seconds lo complete avtundi.

**Why useful?** I/O bound operations lo (file, network) wait time meeda program block avvakunda vere pani cheyyadaniki threading help chestundi.

### 14.1 Multiprocessing Practical Implementation

**CPU bound task example:** Large number ni square cheyyadam laanti heavy calculation. Idi multiple CPU cores meeda true parallel ga run cheyyalante multiprocessing use chestam.

````python
import multiprocessing  # Multiprocessing module ni import chestunnam.

def compute_square(number):  # CPU bound heavy calculation ni simulate chese function.
    result = number * number  # Number ni square chestunnam.
    print(f"Square of {number} is {result}")  # Result ni print chestunnam.

if __name__ == "__main__":  # Windows lo multiprocessing safe ga run avvadaniki ee guard avasaram.
    numbers_to_process = [2, 4, 6, 8]  # Parallel ga process cheyyalsina numbers list.
    processes = []  # Anni process objects store cheyyadaniki empty list create chestunnam.

    for number in numbers_to_process:  # Prathi number ki separate process create chestunnam.
        process = multiprocessing.Process(target=compute_square, args=(number,))  # New process object create chestunnam.
        processes.append(process)  # Process ni list lo add chestunnam.
        process.start()  # Process execution ni start chestunnam.

    for process in processes:  # Anni processes complete ayye varaku wait cheyyadaniki loop chestunnam.
        process.join()  # Prathi process complete ayye varaku main program wait chestundi.

    print("All processes completed")  # Anni processes complete ayina taruvata final message print chestunnam.
````

### Multiprocessing code line-by-line explanation

1. `multiprocessing.Process(target=..., args=(...))` -> new independent process create chestundi; idi separate memory space tho run avtundi.
2. `if __name__ == "__main__":` -> Windows OS lo process spawn cheyyalante ee guard tappanisari (compulsory), lekapothe infinite process creation error vastundi.
3. `.start()` -> process execution ni begin chestundi; multiple processes true parallel ga (different CPU cores meeda) run avvachu.
4. `.join()` -> main program aa process complete ayye varaku wait chestundi.
5. Threads GIL valla true CPU parallelism ivvavu, kani processes prathi okkati separate Python interpreter kabatti true parallel CPU work chestayi.

**Simple meaning:** CPU heavy work ki multiprocessing, I/O wait heavy work ki multithreading select cheyyali.

### 14.2 Thread Pool Executor and Process Pool

Manually prathi task ki thread/process create cheyyakunda, `concurrent.futures` module tho ready-made pool of workers ni reuse cheyyachu. Idi code ni simple ga and efficient ga chestundi.

````python
from concurrent.futures import ThreadPoolExecutor  # Thread pool ni easy ga manage cheyye class import chestunnam.

def fetch_square(number):  # Simple task function; real world lo idi network call kuda avvachu.
    return number * number  # Number square ni return chestunnam.

numbers_list = [1, 2, 3, 4, 5]  # Process cheyyalsina numbers list.

with ThreadPoolExecutor(max_workers=3) as executor:  # Maximum 3 threads tho pool create chestunnam.
    results = list(executor.map(fetch_square, numbers_list))  # Anni numbers ni pool threads ki distribute chestunnam.

print(results)  # Anni squares ni order tho print chestunnam.
````

### Thread pool code line-by-line explanation

3. `with ... as executor:` -> block complete ayyaka pool automatic ga clean up avtundi (threads close avtayi).
4. `ProcessPoolExecutor` kuda idi same vidhamga use chestam, kani adi CPU bound tasks ki processes create chestundi.

**Why useful?** Manual thread/process management kanna pool executor easy ga scale avtundi and resources ni efficient ga reuse chestundi.

### 14.3 Real World Usecases: Web Scraping and Multiprocessing

**Multithreading usecase:** Web scraping lo multiple URLs ki network requests pampinapudu, prathi request response kosam wait cheyyali. Ee wait time lo threading valla several requests parallel ga fire cheyyachu, so total time takkuva avtundi.

**Multiprocessing usecase:** Large dataset meeda heavy CPU calculations (Example: image resizing, data transformation) chesetappudu, dataset ni chunks ga split chesi prathi chunk ni separate process ki assign cheyyadam valla total execution time takkuva avtundi.

```python
import multiprocessing  # Multiprocessing module import chestunnam.

def calculate_group_sum(number_group, result_queue):  # Oka group sum calculate chesi queue lo pettae function.
    group_sum = sum(number_group)  # Current group total sum calculate chestunnam.
    result_queue.put(group_sum)  # Result ni shared queue lo pettutunnam, so main process access cheyyagalgutundi.

if __name__ == "__main__":  # Multiprocessing safety guard.
    all_numbers = list(range(1, 21))  # 1 nundi 20 varaku numbers list create chestunnam.
    number_groups = [all_numbers[i:i + 5] for i in range(0, 20, 5)]  # Numbers ni 4 groups ga split chestunnam.
    result_queue = multiprocessing.Queue()  # Processes madhya results share cheyyadaniki queue create chestunnam.
    processes = []  # Process objects store cheyyadaniki list.

    for group in number_groups:  # Prathi group ki separate process create chestunnam.
        process = multiprocessing.Process(target=calculate_group_sum, args=(group, result_queue))  # Process create chestunnam.
        processes.append(process)  # Process list lo add chestunnam.
        process.start()  # Process start chestunnam.

    for process in processes:  # Anni processes complete ayye varaku wait chestunnam.
        process.join()  # Prathi process ni join chestunnam.

    total_sum = 0  # Final total sum ni store cheyyadaniki variable.
    while not result_queue.empty():  # Queue lo values unnanni varaku loop chestunnam.
        total_sum += result_queue.get()  # Queue nundi prathi group sum ni teesi total ki add chestunnam.

    print(f"Total sum: {total_sum}")  # Final combined sum ni print chestunnam.
```

**Simple meaning:** Multiple processes work independent ga chesi, `Queue` use chesi results ni safe ga main process ki pampistayi.

### 15.1 Key Concepts in Python Memory Management

**Detailed explanation:** Python program run avuthunapudu, prathi variable, list, object anni RAM (computer memory) lo store avtayi. Ee memory ni evaru manage chestaru ante — programmer manual ga chestada? Ledu. - **Reference counting:** Prathi object ki enni references (pointers) unnayo count chestundi; count zero ayithe object automatic ga deallocate avtundi.

**Kid-friendly analogy:** Idi oka **toy library** laantidi anukondi. ### 15.2 Memory Allocation and Deallocation

**Detailed explanation:** "Allocation" ante oka new object create chesinapudu, aa object ki RAM lo konchem space (memory) reserve cheyyadam. "Deallocation" ante aa object ki inka avasaram lekapoyinapudu, aa reserve chesina space ni free chesi vere objects ki tirigi use cheyyadaniki ivvadam.

Example: `sample_list = [1, 2, 3, 4, 5]` ani raayagane, Python:

1. RAM lo list kosam konchem space **allocate** chestundi (list ki entha size kavalo calculate chesi).
2. Values (1, 2, 3, 4, 5) ni aa space lo store chestundi.
3. `sample_list` variable ni aa memory address ki point cheyyadaniki set chestundi.

Taruvatha, `del sample_list` chesthe leda variable ni vere daaniki reassign chesthe, aa list ki reference ledu kabatti Python aa memory ni **deallocate** (free) chestundi.

**Kid-friendly analogy:** Idi oka **hostel room booking** laantidi anukondi. Nuvvu hostel lo join ayinapudu (object create chesinapudu), warden (Python memory manager) neeku oka room (memory space) **allocate** chestadu. Nuvvu hostel nundi leave (`del`) teeskunapudu, aa room ni **deallocate** chesi, next student ki (next object ki) aa room ni ready chestadu. Warden prathi room ni track chestune untadu — evaru unnaru, evaru ledu ani.

**Why important?** Ee process automatic ga jarigithe, manam "out of memory" errors ekkuva avvakunda, and memory efficient ga reuse avutundi.

````python
import sys  # Object size chudataniki sys module import chestunnam.

print("Step 1: Object create cheyyadam (allocation)")  # Allocation step ni indicate chestunnam.
sample_list = [1, 2, 3, 4, 5]  # New list object create chestunnam; idi automatic ga heap meeda allocate avtundi.
print(f"Size in bytes: {sys.getsizeof(sample_list)}")  # List object ki allocate ayina memory size print chestunnam.

print("\nStep 2: Object ni delete cheyyadam (deallocation)")  # Deallocation step ni indicate chestunnam.
del sample_list  # Variable ni delete chestunnam kabatti reference count zero avtundi.
print("sample_list deleted. Memory ippudu free chesi vere object kosam ready ga undi.")  # Deallocation complete ani confirm chestunnam.
````

### Memory allocation and deallocation code line-by-line explanation

1. `sample_list = [1, 2, 3, 4, 5]` -> ee line run ayye moment, Python list object kosam heap meeda memory **allocate** chestundi.
2. `sys.getsizeof(sample_list)` -> aa object ki entha memory (bytes) allocate ayindo chupistundi.
3. `del sample_list` -> variable name ni remove chestunnam; ippudu aa list object ki evaru point cheyyatledu (reference count zero).
4. Reference count zero ayina ventane, Python automatic ga aa memory ni **deallocate** chesi operating system ki return chestundi (leda internal reuse kosam pool lo pedutundi).

**Simple meaning:** Object create ayye prathi sari allocation jarugutundi; object ki avasaram lekapothe deallocation jarugutundi — idantha Python automatic ga handle chestundi.

### 15.3 Reference Counting

**Detailed explanation:** Reference counting ante prathi object ki, "ippudu enni variables/places nundi ee object ni point chestunnaro" ane oka counter maintain cheyyadam. Ee counter ni Python internal ga automatic ga update chestundi:

- Kotha variable oka object ni point cheyinapudu -> count **+1** avtundi.
- Oka variable delete ayinapudu leda vere object ki reassign ayinapudu -> count **-1** avtundi.
- Count **zero** ki vachinapudu -> aa object ki evaru avasaram ledu ani, Python ventane aa object ni memory nundi teesestundi.

**Kid-friendly analogy:** Idi class lo unna oka **shared pencil box** laantidi anukondi. Prathi sari oka student aa pencil box ni teeskunapudu, teacher oka **tally mark** (count) vestundi. Inko student kuda teeskunte, malli oka tally mark add avtundi. Oka student pencil box ni venakki icheste, oka tally mark teesestundi. Anni students box ni venakki ichesesariki (count = zero), teacher aa pencil box ni store room lo pettestundi (memory free) — because ika evarikee daani avasaram ledu.

**Why important?** Idi Python's default and fastest memory cleanup mechanism — most objects (simple variables, lists, dicts) ee reference counting valla e clean avtayi, additional garbage collector avasaram lekunda.

````python
import sys  # Reference count chudataniki sys module import chestunnam.

pencil_box = ["pencil", "eraser", "sharpener"]  # Shared object create chestunnam (pencil box).
print(f"Initial reference count: {sys.getrefcount(pencil_box)}")  # Starting reference count print chestunnam.

student_one = pencil_box  # First student box ni teeskunnadu -> rendo reference create ayindi.
print(f"After student_one takes it: {sys.getrefcount(pencil_box)}")  # Count perigina print chestunnam.

student_two = pencil_box  # Second student kuda box ni teeskunnadu -> mudo reference create ayindi.
print(f"After student_two takes it: {sys.getrefcount(pencil_box)}")  # Count inka perigina print chestunnam.

student_one = None  # First student box ni venakki icchadu -> reference count taggindi.
print(f"After student_one gives it back: {sys.getrefcount(pencil_box)}")  # Count taggina print chestunnam.

student_two = None  # Second student kuda box ni venakki icchadu -> reference count inka taggindi.
print(f"After student_two gives it back: {sys.getrefcount(pencil_box)}")  # Count original level ki vachina print chestunnam.
````

### Reference counting code line-by-line explanation

1. `pencil_box = [...]` -> new object create avtundi, reference count 1 tho start avtundi (module-level variable valla base count konchem ekkuva ga kuda kanipinchachu).
2. `student_one = pencil_box` -> kotha variable same object ni point chestundi, so reference count **+1** avtundi. Kotha copy create avvadu — same object ki rendo name matrame.
3. `student_two = pencil_box` -> inko variable kuda same object ni point chestundi, reference count malli **+1** avtundi.
4. `student_one = None` -> aa variable ippudu object ni point cheyyatledu, so reference count **-1** avtundi.
5. `student_two = None` -> chivari reference kuda remove avtundi; ippudu evaru point cheyyatledu kabatti object automatic ga deallocate avtundi.

**Simple meaning:** Reference count ante "prathi object ni entha mandi vaadutunnaru" ane simple counter — count zero ayithe object safe ga tiseyavachu.

### 15.4 Garbage Collection

**Detailed explanation:** Reference counting fast ga work chestundi, kani daaniki oka pedda problem undi — **circular references**. Example: object A, object B ni point chestundi; object B malli object A ni point chestundi. Ippudu rendu objects ki kuda "nenu use avutunnanu" ane reference count kaneesam 1 ga untundi, kaani real ga vere ekkadi nundi (main program nundi) evaru access cheyyatledu. Ilanti "unreachable, kani reference count zero kaadu" ane objects ni **cyclic garbage** antaru.

Ee problem solve cheyyadaniki Python's **garbage collector** periodically run avtundi. Idi "reachability" ni check chestundi — main program nundi (root objects nundi) start chesi, evi evi objects ki actually access undo trace chestundi. Evi unreachable ga unnayo (circular references included) vaatini clean chestundi.

**Kid-friendly analogy:** Idi oka **classroom lo iddaru pillalu okari cheyi okaru pattukunna game** laantidi anukondi — A, B cheyi pattukuni unnadu, B malli A cheyi pattukuni unnadu. School iddarini intiki pampesina (main program ki access ledu ayina), vaalliddaru okari cheyi okaru vadalakunda classroom lone undipotunnaru (reference count zero kaadu). Ee situation ni fix cheyyadaniki **principal (garbage collector)** periodic ga classroom ki vachi, "evaru bayati nundi (main program nundi) accessible kaaru" ani check chesi, ilanti hand-holding pairs ni veru chesi intiki pampistadu (memory free chestadu).

**Why important?** Idi lekapothe, circular references unna programs lo memory slow ga full ayipotundi (memory leak), ekkuva sepu run ayye applications (web servers, long scripts) lo idi pedda problem.

````python
class Student:  # Circular reference create cheyyadaniki simple class define chestunnam.
    def __init__(self, name):  # Student object create ayye time lo name set chestunnam.
        self.name = name  # Student name ni store chestunnam.
        self.best_friend = None  # Best friend ni point cheyyadaniki placeholder.

student_a = Student("A")  # First student object create chestunnam.
student_b = Student("B")  # Second student object create chestunnam.

student_a.best_friend = student_b  # A, B ni point chestundi (handshake step 1).
student_b.best_friend = student_a  # B malli A ni point chestundi (handshake step 2) -> circular reference create ayindi.

del student_a  # Variable name ni delete chestunnam, kani student_b nundi ippatiki A ki reference undi.
del student_b  # Rendo variable name ni kuda delete chestunnam; ippudu evaru direct ga point cheyyatledu, kani rendu objects okadaniki okati point chestune unnayi.

print("Both variable names deleted, kani circular reference valla objects inka memory lo unnayi (until gc runs).")  # Current situation ni explain chestunnam.
````

### Garbage collection code line-by-line explanation

1. `student_a.best_friend = student_b` and `student_b.best_friend = student_a` -> rendu objects okadaniki okati point chestunnayi, idi classic circular reference.
2. `del student_a` / `del student_b` -> variable names ni matrame remove chestunnam; kani rendu objects okadaniki okati point chestune unnayi kabatti reference count zero kaadu.
3. Plain reference counting valla ee rendu objects clean avvavu — endukante evarikee direct ga access ledu, kani internal ga okarito okaru "reference" chesukuntunnaru.
4. Ilanti unreachable-but-referenced objects ni **garbage collector** periodic ga background lo detect chesi, safely deallocate chestundi (next code cell lo `gc.collect()` tho idi confirm cheddam).

### 15.5 The `gc` Module

**Detailed explanation:** Python's built-in `gc` module tho, garbage collector ela behave avutundo manam control cheyyachu and inspect cheyyachu:

- `gc.collect()` -> garbage collector ni **manual ga** immediately trigger chestundi; endaina circular references unte clean chestundi and enni objects collect ayyayo return chestundi.
- `gc.isenabled()` -> automatic gc enable ga unda ledha ani check chestundi (default: enabled).
- `gc.disable()` / `gc.enable()` -> automatic gc ni poorthiga off/on cheyyachu (usually performance-critical sections lo temporary ga disable chesi, taruvatha enable chestaru).
- `gc.get_stats()` -> garbage collector enni generations lo enni objects track chestundo statistics ga istundi.

**Kid-friendly analogy:** `gc` module ante oka **cleaning helper button** laantidi anukondi. Normal ga school (Python) tanaka tanu periodic ga cleaning (garbage collection) chestune untundi. Kani nuvvu "ippude ventane clean cheyandi" ani cheppalante, oka special button (`gc.collect()`) nokkithe, cleaning helper ventane vachi, hand-holding (circular reference) pillalandarini veru chesi intiki pampistadu, and enthamandini pampinchado neeku cheptadu.

**Why important?** Normal applications lo manam `gc.collect()` manual ga call cheyyalsina avasaram ledu (Python automatic ga chesukuntundi), kani memory debugging/profiling chesetappudu idi chala useful.

````python
import gc  # Garbage collector module ni import chestunnam.

class Node:  # Circular reference create cheyyadaniki simple class define chestunnam.
    def __init__(self, name):  # Node object create ayye time lo name set chestunnam.
        self.name = name  # Node name ni store chestunnam.
        self.reference = None  # Vere node ni point cheyyadaniki placeholder.

first_node = Node("first")  # First node object create chestunnam.
second_node = Node("second")  # Second node object create chestunnam.

first_node.reference = second_node  # First node, second node ni point chestundi.
second_node.reference = first_node  # Second node, malli first node ni point chestundi -> circular reference create ayindi.

del first_node  # Variable name ni delete chestunnam, kani circular reference valla objects deallocate avvakapovachu.
del second_node  # Rendo variable name ni kuda delete chestunnam.

print(f"Is gc enabled: {gc.isenabled()}")  # Automatic garbage collection enable ga unda ani check chestunnam.

collected_objects_count = gc.collect()  # Garbage collector ni manual ga run chesi circular references clean chestunnam.
print(f"Objects collected by gc: {collected_objects_count}")  # gc enni unreachable objects free chesindo print chestunnam.

generation_stats = gc.get_stats()  # Garbage collector generations statistics ni fetch chestunnam.
print(f"Number of gc generations tracked: {len(generation_stats)}")  # Enni generations track chestundo print chestunnam.
````

### `gc` module code line-by-line explanation

1. `first_node.reference = second_node` and `second_node.reference = first_node` -> rendu objects okadaniki okati point chestunnayi (circular reference), just like previous example.
2. `del first_node` / `del second_node` -> variable names ni matrame remove chestunnam; reference count zero kaadu (endukante objects okatokati inka point chestunnayi), so plain reference counting valla clean avvadu.
3. `gc.isenabled()` -> Python default ga automatic gc periodic ga background lo run avutunda ani cheptundi (default: enabled).
4. `gc.collect()` -> garbage collector ni manual ga trigger chestundi; idi circular references ni detect chesi safely deallocate chestundi, and enni objects collect ayyayo return chestundi.
5. `gc.get_stats()` -> internal ga gc moodu "generations" (young, middle, old objects) ga divide chesi track chestundi ani chupistundi — konni objects thakkuva sarlu check chesthe chalu (long-living objects), konni objects frequent ga check cheyyali (short-living objects).

**Simple meaning:** Reference counting fast and simple, kani circular references ki `gc` module backup ga panichestundi — manam daani behavior ni control and inspect kuda cheyyachu.

### 15.6 Memory Management Best Practices

1. **Unnecessary references avoid cheyyi:** Object ni inka avasaram lekapothe `del` cheyyadam or `None` assign cheyyadam valla reference count taggi, memory fast ga free avutundi.
2. **Circular references careful ga handle cheyyi:** Possible ayithe circular references design cheyyakudadu; tappanisarayithe `weakref` module use cheyyandi (weak reference reference count ni penchadu).
3. **Large objects ki context managers vadandi:** Files, database connections laanti vati ni `with` statement tho vadithe, block complete ayyaka automatic ga close/release avtayi.
4. **Generators use cheyyandi large data kosam:** Entire large dataset ni list ga load cheyyakunda, generators (`yield`) tho okkasariki okka item matrame memory loki teesukondi.
5. **Profiling tools use cheyyandi:** `sys.getsizeof()`, `tracemalloc`, `gc.get_stats()` laanti tools tho memory usage ni monitor cheyyandi, memory leaks ni early ga guruthinchadaniki.

**Kid-friendly analogy:** Idi oka **chinna room ni clean ga pettukovadam** laantidi anukondi. Nuvvu prathi roju vadanidhi (unnecessary references) bayataki pampiste (delete cheste), room eppudu khaaliga untundi. Rendu toys okadaniki okati tied chesi (circular reference) pettakunda unte, easy ga clean cheyavachu. Pedda boxes (large objects) ni vaadina taruvatha moosestey (`with` statement/context managers), room messy kaadu. Mottham toys okesari bayataki teeyakunda, okkokkati ga teesukunte (generators) room eppudu overcrowded kaadu.

**Simple meaning:** Python automatic ga memory manage chesina, manam manchi coding practices follow chesthe application fast ga and memory-efficient ga untundi.

````python
def read_large_range_bad():  # Anti-pattern: entire range ni list ga memory lo load chestundi.
    return [number for number in range(1000000)]  # 1 million numbers anni okesari memory lo untayi.

def read_large_range_good():  # Best practice: generator use chesi lazy ga values produce chestundi.
    for number in range(1000000):  # 1 million numbers meeda loop chestunnam.
        yield number  # One-by-one value ni yield chestunnam; anni okesari memory lo undavu.

import sys  # Memory size compare cheyyadaniki sys module import chestunnam.

large_list = read_large_range_bad()  # Anti-pattern function call chesi full list create chestunnam.
large_generator = read_large_range_good()  # Best-practice function call chesi generator object create chestunnam.

print(f"List size in bytes: {sys.getsizeof(large_list)}")  # Full list ki entha memory allocate ayindo print chestunnam.
print(f"Generator size in bytes: {sys.getsizeof(large_generator)}")  # Generator object size chinnaga untundi ani print chestunnam.
````
