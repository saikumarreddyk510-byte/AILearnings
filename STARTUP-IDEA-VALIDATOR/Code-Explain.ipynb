# Startup Idea Validator Notebook

This notebook builds a startup research assistant using Groq LLM and ddgs search.

- Install required packages
- Load environment variables
- Initialize Groq client
- Perform a web search
- Generate research analysis for a startup idea

````python
# Install required packages (run once)
%pip install -q groq python-dotenv ddgs wikipedia requests langgraph crewai
````

````python
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
if not groq_api_key:
    raise EnvironmentError(
        "GROQ_API_KEY is not set. Add it to your .env file and rerun the first cell."
    )

try:
    from groq import Groq
except ModuleNotFoundError as e:
    raise ModuleNotFoundError(
        "The groq package is not installed. Run the install cell above and restart the kernel."
    ) from e

# Initialize Groq client with API key from environment
client = Groq(api_key=groq_api_key)

print("✅ Groq client initialized successfully!")
print(f"API Key loaded: {bool(groq_api_key)}")
````

````python
def ask_llm(prompt):
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": "You are a helpful AI teacher."},
            {"role": "user", "content": prompt}
        ],
        temperature=0.3
    )
    return response.choices[0].message.content

# Test the function
print(ask_llm("Explain AI Agent in simple english telugu mixed."))
````

````python
def calculator(expression):
    try:
        return eval(expression)
    except Exception as e:
        return f"Error: {e}"

def simple_agent(user_input):
    if "calculate" in user_input.lower():
        expression = user_input.lower().replace("calculate", "").strip()
        result = calculator(expression)
        return f"The calculated result is: {result}"
    else:
        return ask_llm(user_input)

print(simple_agent("calculate 25 * 4"))
print(simple_agent("What is an AI agent?"))
````

````python
import warnings

warnings.filterwarnings(
    "ignore",
    message=r"This package \(`duckduckgo_search`\) has been renamed to `ddgs`! Use `pip install ddgs` instead\.",
)

try:
    from ddgs import DDGS
except ModuleNotFoundError as e:
    raise ModuleNotFoundError(
        "The ddgs package is not installed. Run the install cell above and restart the kernel."
    ) from e

def web_search(query, max_results=3):
    results = []
    with DDGS() as ddgs:
        search_results = ddgs.text(query, max_results=max_results)
        for r in search_results:
            results.append({
                "title": r.get("title", "No title"),
                "href": r.get("href", ""),
                "body": r.get("body", "No summary available")
            })
    return results

query = "latest trends in AI startups"
search_results = web_search(query)

for item in search_results:
    print(item.get("title"))
    print(item.get("body"))
    print("-" * 50)
````

````python
def startup_research_agent(idea):
    search_data = web_search(f"{idea} market trends competitors", max_results=5)

    context = "\n\n".join([
        f"Title: {item['title']}\nSummary: {item['body']}\nURL: {item['href']}"
        for item in search_data
    ])

    prompt = f"""
You are a startup research analyst.

Startup idea:
{idea}

Web research data:
{context}

Give:
1. Market opportunity
2. Possible competitors
3. Target audience
4. Risks
5. Final quick opinion
"""

    return ask_llm(prompt)

idea = "AI tutor for Telugu speaking students"
print(startup_research_agent(idea))
````
