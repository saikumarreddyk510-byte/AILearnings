## Architecture: Startup Idea Validator Pipeline

```
[User Input - Startup Idea]
        ↓
[Groq LLM - llama-3.1-8b-instant]
        ↓
[CrewAI - Multi-Agent Analysis]
  ├── Research Agent (web search via DuckDuckGo)
  ├── Analysis Agent (market evaluation)
  └── Report Agent (final recommendations)
        ↓
[Comprehensive Validation Report]
  - Market viability
  - Competition analysis
  - Revenue potential
  - Implementation roadmap
  - Risk assessment
```

````python
import os
from dotenv import load_dotenv
from groq import Groq

# Load environment variables
load_dotenv()

# Initialize client
client = Groq(api_key=os.getenv("GROQ_API_KEY"))

def ask_llm(prompt):
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": "You are a helpful AI teacher."},
            {"role": "user", "content": prompt}
        ],
        temperature=0.3
    )
    return response.choices[0].message.content

# Test the function
print(ask_llm("Explain AI Agent in simple english telugu mixed."))
````

````python
def calculator(expression):
    try:
        return eval(expression)
    except Exception as e:
        return f"Error: {e}"

def simple_agent(user_input):
    if "calculate" in user_input.lower():
        expression = user_input.lower().replace("calculate", "").strip()
        result = calculator(expression)
        return f"The calculated result is: {result}"
    else:
        return ask_llm(user_input)

print(simple_agent("calculate 25 * 4"))
print(simple_agent("What is an AI agent?"))
````

````python
from duckduckgo_search import DDGS

def web_search(query, max_results=3):
    results = []
    with DDGS() as ddgs:
        for r in ddgs.text(query, max_results=max_results):
            results.append({
                "title": r.get("title"),
                "href": r.get("href"),
                "body": r.get("body")
            })
    return results

query = "latest trends in AI startups"
search_results = web_search(query)

for item in search_results:
    print(item["title"])
    print(item["body"])
    print("-" * 50)
````
