````python
import os
from dotenv import load_dotenv
from groq import Groq

# Load environment variables from .env file
load_dotenv()

# Initialize Groq client with API key from environment
client = Groq(api_key=os.getenv("GROQ_API_KEY"))

print("✅ Groq client initialized successfully!")
print(f"API Key loaded: {bool(os.getenv('GROQ_API_KEY'))}")
````

## Architecture: Startup Idea Validator Pipeline

```
[User Input - Startup Idea]
        ↓
[Groq LLM - llama-3.1-8b-instant]
        ↓
[CrewAI - Multi-Agent Analysis]
  ├── Research Agent (web search via DuckDuckGo)
  ├── Analysis Agent (market evaluation)
  └── Report Agent (final recommendations)
        ↓
[Comprehensive Validation Report]
  - Market viability
  - Competition analysis
  - Revenue potential
  - Implementation roadmap
  - Risk assessment
```

````python
# Test simple chat completion with Groq
chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "What are 3 key factors for a successful startup idea?"
        }
    ],
    model="llama-3.1-8b-instant",
    temperature=0.7,
    max_tokens=500
)

print("Groq Response:")
print(chat_completion.choices[0].message.content)
````

````python
def ask_llm(prompt: str, client: Groq) -> str:
    """
    Ask Groq LLM a question and get response.
    
    Args:
        prompt: Question or prompt for the LLM
        client: Groq client instance
        
    Returns:
        String response from the LLM
    """
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": "You are a helpful AI teacher."},
            {"role": "user", "content": prompt}
        ],
        temperature=0.3
    )
    return response.choices[0].message.content

# Test the function
print(ask_llm("Explain AI Agent in simple words.", client))
````
