**confusion_matrix(actual, predicted):**
```
[[TN  FP]
 [FN  TP]]

TN = Actually NO,  Predicted NO  ?
TP = Actually YES, Predicted YES ?
FP = Actually NO,  Predicted YES ? (False Alarm)
FN = Actually YES, Predicted NO  ? (Missed)
```

**classification_report:**
```
              precision  recall  f1-score
0 (dissatisfied)  0.xx    0.xx    0.xx
1 (satisfied)     0.xx    0.xx    0.xx
accuracy                           0.xx
```

- **Precision:** Predicted positive lo enni correct ? "spam filter false alarm entha?"
- **Recall:** Actual positive lo enni catch chesam ? "satisfied passengers ni miss chesama?"
- **F1-score:** Precision and Recall balance ? harmonic mean

**Test vs Train compare chestam:**
- Test report = real performance
- Train report = training performance
- Train >> Test ? Overfit (Decision Tree common problem) ?

````python
print(confusion_matrix(y_test, test_predictions))
print(classification_report(y_test, test_predictions))

print(confusion_matrix(y_train, train_predictions))
print(classification_report(y_train, train_predictions))
````

| Cell | Type | Meaning |
|------|------|---------|

| **31** top-right | **FP** ? | Actually 0, Predicted 1 ? dissatisfied ni satisfied ani wrong ga chesam |


#### Metrics Explanation:

**Precision (0.67 for class 0):**
> Model "dissatisfied" ani cheppinappudu, 67% correct
> Formula: TN / (TN + FP) = 58 / (58+31) = 0.65

**Recall (0.65 for class 0):**
> Actual dissatisfied lo 65% ni correctly catch chesam
> Formula: TN / (TN + FN) = 58 / (58+28) = 0.67

**F1-Score (0.66):**
> Precision + Recall balance chese single score
> 0.85+ ? Good | 0.65 ? Average | 0.50 ? Random guessing

## Cell 22: Random Forest ? Train Cheyyatam

```python
RF = RandomForestClassifier(n_estimators=99, class_weight="balanced")
RF.fit(X_train, y_train)
```

**RandomForestClassifier(n_estimators=99):**
- `n_estimators=99` ? 99 Decision Trees build chesthundi
- Prathi tree different bootstrap sample meeda train avutundi (Bagging)
- 99 trees vote chesi final prediction chesthundi

**Why 99? Not 100?**
> Convention ledu ? odd number unte tie-breaking avoid avutundi in voting. 100 trees lo 50-50 split possible, 99 lo impossible.

**class_weight="balanced":**
- Same as Decision Tree ? class imbalance handle chesthundi

**RF.fit(X_train, y_train):**
- 99 trees anni train avutay ? slightly more time takes than single tree
- Kani result chala better ?

````python
RF = RandomForestClassifier(n_estimators = 99, class_weight= "balanced")
RF.fit(X_train, y_train)
````
