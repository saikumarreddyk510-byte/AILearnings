## Logistic Regression ? Core Model

**Ee section lo:**
> Airline passenger data meeda Logistic Regression apply chesi satisfaction predict chestam.

**How LR works internally:**

```
Step 1: z = w?x? + w?x? + ... + w?x? + b  (linear combination)
         (weights ? features + bias)

Step 2: p = sigmoid(z) = 1 / (1 + e??)    (probability)
         Output: 0 to 1

Step 3: if p > 0.5 ? satisfied (1)
         if p < 0.5 ? not satisfied (0)
```

**Training lo emi jarigithundi?**
> LR training lo **weights (w)** optimize chesthundi ? correct predictions vastunaay ani.
> Uses **Maximum Likelihood Estimation** (not MSE like Linear Regression).

````python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
````

## Cell 18: Initialize and Train Model

```python

```

**LogisticRegression() default params:**
- `penalty='l2'` ? L2 regularization (prevents overfitting)
- `C=1.0` ? Regularization strength (lower = stronger regularization)
- `max_iter=100` ? Maximum iterations for convergence
- `solver='lbfgs'` ? Optimization algorithm

**LR.fit() lo emi jarigithundi?**
> Model features (X_train) chusi satisfaction (y_train) nerchukuntundi.
> Prathi feature ki optimal weight (w) calculate chestundi using gradient descent.
> "Learning" = weights adjust cheyyadam until error minimize avutundi.

**LR vs Decision Tree training difference:**
> Decision Tree: Splits build chestundi (GINI use chesi)
> Logistic Regression: Weights optimize chestundi (Loss function minimize chesi)
